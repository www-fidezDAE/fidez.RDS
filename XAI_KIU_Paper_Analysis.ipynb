```python
# ================================================================
# SMART WORKFORCE ANALYTICS
# EXPLAINABLE MACHINE LEARNING FOR EMPLOYEE PRODUCTIVITY PREDICTION
#
# Study:
# Explainable Machine Learning for Employee Productivity Prediction:
# A Data Driven Human Capital Decision Support Framework in Uganda
#
# Objectives:
# 1. Examine relationships between employee characteristics,
#    digital capability, organisational factors, work environment,
#    training, technology adoption and employee productivity.
#
# 2. Develop statistical and machine learning models for predicting
#    employee productivity.
#
# 3. Compare predictive performance using RMSE, MAE and R².
#
# 4. Interpret predictions of the best performing machine learning
#    model using SHAP and identify principal predictive factors.
#
# DATA EXPECTED:
# Respondent_ID, Age, Education, Experience, Tenure,
# DC, TA, OS, WE, TR, EPI
#
# OPTIONAL RAW QUESTIONNAIRE ITEMS:
# B1:B10, C1:C10, D1:D10, E1:E10, F1:F6, G1:G10
#
# ================================================================


# ================================================================
# 1. MOUNT GOOGLE DRIVE
# ================================================================

from google.colab import drive
drive.mount('/content/drive')


# ================================================================
# 2. INSTALL / UPDATE REQUIRED PACKAGES
# ================================================================

!pip -q install pandas numpy scipy statsmodels scikit-learn \
    xgboost shap openpyxl matplotlib seaborn statsmodels


# ================================================================
# 3. IMPORT LIBRARIES
# ================================================================

import os
import warnings
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import pearsonr

import statsmodels.api as sm
import statsmodels.formula.api as smf

from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.diagnostic import (
    het_breuschpagan,
    linear_reset,
    acorr_breusch_godfrey
)
from statsmodels.stats.stattools import durbin_watson, jarque_bera

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_validate,
    RandomizedSearchCV,
    GridSearchCV
)

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

from sklearn.inspection import permutation_importance

from xgboost import XGBRegressor

import shap

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
pd.set_option("display.max_rows", 100)

print("All libraries imported successfully.")


# ================================================================
# 4. CREATE RESULTS DIRECTORY
# ================================================================

RESULTS_DIR = "/content/drive/MyDrive/Smart_Workforce_Analytics_Results"

os.makedirs(RESULTS_DIR, exist_ok=True)

print("Results directory:", RESULTS_DIR)


# ================================================================
# 5. UPLOAD EXCEL DATASET
# ================================================================

from google.colab import files

print("Please upload the Excel dataset containing the 612 employee records.")

uploaded = files.upload()

uploaded_files = list(uploaded.keys())

if len(uploaded_files) == 0:
    raise FileNotFoundError("No Excel file was uploaded.")

excel_file = uploaded_files[0]

print("Uploaded file:", excel_file)


# ================================================================
# 6. READ EXCEL DATA
# ================================================================

df = pd.read_excel(excel_file)

print("\nDataset successfully loaded.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())


# ================================================================
# 7. CLEAN COLUMN NAMES
# ================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print("\nColumn names:")
print(df.columns.tolist())


# ================================================================
# 8. EXPECTED CORE VARIABLES
# ================================================================

id_col = "Respondent_ID"

core_predictors = [
    "Age",
    "Education",
    "Experience",
    "Tenure",
    "DC",
    "TA",
    "OS",
    "WE",
    "TR"
]

target = "EPI"

required_columns = [id_col] + core_predictors + [target]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"The following required columns are missing: {missing_columns}"
    )

print("\nAll required analytical variables are available.")


# ================================================================
# 9. VERIFY EXPECTED SAMPLE SIZE
# ================================================================

print("\nSample size verification")

print("Observed records:", len(df))

if len(df) == 612:
    print("PASS: The dataset contains the expected 612 employee records.")
else:
    print(
        "WARNING: The methodology reports 612 observations, "
        f"but the uploaded file contains {len(df)} records."
    )


# ================================================================
# 10. CHECK DUPLICATE RESPONDENT IDS
# ================================================================

duplicate_ids = df[id_col].duplicated().sum()

print("\nDuplicate Respondent IDs:", duplicate_ids)

if duplicate_ids > 0:
    print("Duplicate IDs detected. Inspecting duplicates...")
    display(
        df[df[id_col].duplicated(keep=False)]
        .sort_values(id_col)
    )
else:
    print("No duplicate Respondent IDs detected.")


# ================================================================
# 11. CHECK COMPLETE DUPLICATE ROWS
# ================================================================

duplicate_rows = df.duplicated().sum()

print("Completely duplicated rows:", duplicate_rows)

if duplicate_rows > 0:
    df = df.drop_duplicates()
    print("Exact duplicate rows removed.")

print("Current dataset size:", df.shape)


# ================================================================
# 12. DATA TYPE CONVERSION
# ================================================================

for col in core_predictors + [target]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("\nData types after conversion:")
print(df[core_predictors + [target]].dtypes)


# ================================================================
# 13. MISSING DATA ASSESSMENT
# ================================================================

missing_summary = pd.DataFrame({
    "Variable": df.columns,
    "Missing_N": df.isna().sum().values,
    "Missing_Percent": (
        df.isna().mean().values * 100
    )
})

display(
    missing_summary
    .sort_values("Missing_Percent", ascending=False)
)

missing_summary.to_excel(
    os.path.join(RESULTS_DIR, "01_missing_data_summary.xlsx"),
    index=False
)


# ================================================================
# 14. RANGE AND VALIDITY CHECK
# ================================================================

print("\nDescriptive range checks")

for col in ["DC", "TA", "OS", "WE", "TR", "EPI"]:
    print(
        f"{col}: minimum = {df[col].min()}, "
        f"maximum = {df[col].max()}"
    )

# Likert based composite scores should normally lie between 1 and 5.

invalid_construct_rows = df[
    (
        (df[["DC", "TA", "OS", "WE", "TR", "EPI"]] < 1) |
        (df[["DC", "TA", "OS", "WE", "TR", "EPI"]] > 5)
    ).any(axis=1)
]

print(
    "\nRows with construct scores outside the expected 1 to 5 range:",
    len(invalid_construct_rows)
)

if len(invalid_construct_rows) > 0:
    display(invalid_construct_rows)


# ================================================================
# 15. FINAL CLEAN ANALYTICAL DATASET
# ================================================================

analysis_cols = [id_col] + core_predictors + [target]

analysis_df = df[analysis_cols].copy()

before = len(analysis_df)

analysis_df = analysis_df.dropna(
    subset=core_predictors + [target]
).reset_index(drop=True)

after = len(analysis_df)

print("\nComplete case screening")
print("Before:", before)
print("After:", after)
print("Removed:", before - after)


# ================================================================
# 16. DESCRIPTIVE STATISTICS
# ================================================================

descriptive = analysis_df[
    core_predictors + [target]
].describe().T

descriptive["skewness"] = analysis_df[
    core_predictors + [target]
].skew()

descriptive["kurtosis"] = analysis_df[
    core_predictors + [target]
].kurtosis()

print("\nDescriptive statistics")

display(descriptive)

descriptive.to_excel(
    os.path.join(
        RESULTS_DIR,
        "02_descriptive_statistics.xlsx"
    )
)


# ================================================================
# 17. FREQUENCY DISTRIBUTIONS FOR ORDINAL CHARACTERISTICS
# ================================================================

ordinal_variables = [
    "Age",
    "Education",
    "Experience",
    "Tenure"
]

for col in ordinal_variables:

    print("\nFrequency distribution:", col)

    freq = (
        analysis_df[col]
        .value_counts(dropna=False)
        .sort_index()
        .rename_axis(col)
        .reset_index(name="Frequency")
    )

    freq["Percentage"] = (
        freq["Frequency"] /
        freq["Frequency"].sum()
    ) * 100

    display(freq)

    freq.to_excel(
        os.path.join(
            RESULTS_DIR,
            f"03_frequency_{col}.xlsx"
        ),
        index=False
    )


# ================================================================
# 18. HISTOGRAM OF EMPLOYEE PRODUCTIVITY
# ================================================================

plt.figure(figsize=(9, 6))

sns.histplot(
    analysis_df["EPI"],
    kde=True,
    bins=15
)

plt.title("Distribution of Employee Productivity Index")
plt.xlabel("Employee Productivity Index")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "04_EPI_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 19. BOX PLOTS FOR MAIN CONTINUOUS VARIABLES
# ================================================================

plot_variables = [
    "DC",
    "TA",
    "OS",
    "WE",
    "TR",
    "EPI"
]

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=analysis_df[plot_variables]
)

plt.title("Distribution and Potential Outliers of Study Constructs")
plt.ylabel("Score")
plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "05_construct_boxplots.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 20. CORRELATION ANALYSIS
# ================================================================

correlation_variables = [
    "Age",
    "Education",
    "Experience",
    "Tenure",
    "DC",
    "TA",
    "OS",
    "WE",
    "TR",
    "EPI"
]

corr_matrix = analysis_df[
    correlation_variables
].corr(method="pearson")

print("\nPearson correlation matrix")

display(corr_matrix)

corr_matrix.to_excel(
    os.path.join(
        RESULTS_DIR,
        "06_pearson_correlation_matrix.xlsx"
    )
)


# ================================================================
# 21. CORRELATION HEATMAP
# ================================================================

plt.figure(figsize=(12, 9))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True
)

plt.title(
    "Correlation Matrix of Employee Characteristics, "
    "Workforce Constructs and Productivity"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "07_correlation_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 22. CRONBACH'S ALPHA FUNCTION
# ================================================================

def cronbach_alpha(data):

    data = data.dropna()

    item_variances = data.var(axis=0, ddof=1)

    total_score = data.sum(axis=1)

    total_variance = total_score.var(ddof=1)

    k = data.shape[1]

    if k <= 1:
        return np.nan

    alpha = (
        k / (k - 1)
    ) * (
        1 -
        item_variances.sum() /
        total_variance
    )

    return alpha


# ================================================================
# 23. RELIABILITY ANALYSIS
#
# This section only runs when item level questionnaire variables
# actually exist.
# ================================================================

construct_items = {
    "Digital Capability": [f"B{i}" for i in range(1, 11)],
    "Technology Adoption": [f"C{i}" for i in range(1, 11)],
    "Organisational Support": [f"D{i}" for i in range(1, 11)],
    "Work Environment": [f"E{i}" for i in range(1, 11)],
    "Training": [f"F{i}" for i in range(1, 7)],
    "Employee Productivity": [f"G{i}" for i in range(1, 11)]
}

reliability_results = []

for construct, items in construct_items.items():

    available_items = [
        item for item in items
        if item in df.columns
    ]

    if len(available_items) == len(items):

        item_data = df[available_items].apply(
            pd.to_numeric,
            errors="coerce"
        )

        alpha = cronbach_alpha(item_data)

        reliability_results.append({
            "Construct": construct,
            "Number_of_items": len(available_items),
            "Cronbach_alpha": alpha
        })

    else:

        print(
            f"\nReliability not calculated for {construct}."
        )

        print(
            "Required item level variables are not present."
        )

if reliability_results:

    reliability_table = pd.DataFrame(
        reliability_results
    )

    display(reliability_table)

    reliability_table.to_excel(
        os.path.join(
            RESULTS_DIR,
            "08_cronbach_alpha.xlsx"
        ),
        index=False
    )

else:

    print(
        "\nNOTE:"
        "\nThe uploaded analytical dataset contains composite "
        "construct scores only."
        "\nCronbach's alpha cannot be legitimately reconstructed "
        "from DC, TA, OS, WE, TR and EPI alone."
    )


# ================================================================
# 24. PRINCIPAL COMPONENT ANALYSIS
#
# If item level data are available, PCA is conducted on the
# six construct scores as specified in the methodology.
# ================================================================

pca_variables = [
    "DC",
    "TA",
    "OS",
    "WE",
    "TR",
    "EPI"
]

pca_data = analysis_df[pca_variables].dropna().copy()

# Standardise for PCA
pca_scaled = StandardScaler().fit_transform(pca_data)

from sklearn.decomposition import PCA

pca_model = PCA()

pca_model.fit(pca_scaled)

eigenvalues = pca_model.explained_variance_

explained_variance = (
    pca_model.explained_variance_ratio_ * 100
)

pca_summary = pd.DataFrame({
    "Component": np.arange(
        1,
        len(eigenvalues) + 1
    ),
    "Eigenvalue": eigenvalues,
    "Variance_Percent": explained_variance,
    "Cumulative_Percent": np.cumsum(
        explained_variance
    )
})

print("\nPCA summary")

display(pca_summary)

pca_summary.to_excel(
    os.path.join(
        RESULTS_DIR,
        "09_PCA_summary.xlsx"
    ),
    index=False
)


# ================================================================
# 25. PCA LOADINGS
# ================================================================

loadings = pd.DataFrame(
    pca_model.components_.T,
    index=pca_variables,
    columns=[
        f"PC{i}"
        for i in range(
            1,
            len(pca_variables) + 1
        )
    ]
)

print("\nPCA component loadings")

display(loadings)

loadings.to_excel(
    os.path.join(
        RESULTS_DIR,
        "10_PCA_loadings.xlsx"
    )
)


# ================================================================
# 26. KMO AND BARTLETT'S TEST
# ================================================================

# KMO calculation
from scipy.linalg import inv

corr = pca_data.corr().values

inv_corr = inv(corr)

partial_corr = np.zeros_like(corr)

for i in range(corr.shape[0]):
    for j in range(corr.shape[1]):

        if i == j:
            partial_corr[i, j] = 0

        else:

            partial_corr[i, j] = (
                -inv_corr[i, j] /
                np.sqrt(
                    inv_corr[i, i] *
                    inv_corr[j, j]
                )
            )

r_squared = corr ** 2
partial_squared = partial_corr ** 2

np.fill_diagonal(r_squared, 0)
np.fill_diagonal(partial_squared, 0)

kmo_total = (
    r_squared.sum() /
    (
        r_squared.sum() +
        partial_squared.sum()
    )
)

print(
    f"\nOverall KMO measure: {kmo_total:.3f}"
)


# Bartlett's test
n_pca = len(pca_data)
pca_corr_det = np.linalg.det(corr)
p = len(pca_variables)

chi_square = (
    -(n_pca - 1 - (2*p + 5)/6)
    * np.log(pca_corr_det)
)

df_bartlett = p * (p - 1) / 2

bartlett_p = stats.chi2.sf(
    chi_square,
    df_bartlett
)

print(
    f"Bartlett's test: "
    f"chi-square = {chi_square:.3f}, "
    f"df = {int(df_bartlett)}, "
    f"p = {bartlett_p:.6f}"
)


# ================================================================
# 27. PREPARE MODELING DATA
# ================================================================

features = [
    "Age",
    "Education",
    "Experience",
    "Tenure",
    "DC",
    "TA",
    "OS",
    "WE",
    "TR"
]

X = analysis_df[features].copy()
y = analysis_df[target].copy()


# ================================================================
# 28. TRAIN TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTrain/test split")
print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))

print(
    "Training percentage:",
    round(len(X_train) / len(X) * 100, 2),
    "%"
)

print(
    "Testing percentage:",
    round(len(X_test) / len(X) * 100, 2),
    "%"
)


# ================================================================
# 29. MULTIPLE LINEAR REGRESSION FOR OBJECTIVE 1
#
# This is the inferential statistical model.
# ================================================================

X_ols = sm.add_constant(X)

ols_model = sm.OLS(
    y,
    X_ols
).fit()

print("\n================================================")
print("MULTIPLE LINEAR REGRESSION")
print("================================================")

print(ols_model.summary())


# ================================================================
# 30. REGRESSION COEFFICIENT TABLE
# ================================================================

conf_int = ols_model.conf_int()

regression_table = pd.DataFrame({
    "Variable": ols_model.params.index,
    "Coefficient": ols_model.params.values,
    "Std_Error": ols_model.bse.values,
    "t_value": ols_model.tvalues.values,
    "p_value": ols_model.pvalues.values,
    "CI_Lower_95": conf_int[0].values,
    "CI_Upper_95": conf_int[1].values
})

display(regression_table)

regression_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "11_multiple_linear_regression.xlsx"
    ),
    index=False
)


# ================================================================
# 31. REGRESSION MODEL SUMMARY
# ================================================================

regression_summary = pd.DataFrame({
    "Statistic": [
        "R_squared",
        "Adjusted_R_squared",
        "F_statistic",
        "F_p_value",
        "AIC",
        "BIC"
    ],
    "Value": [
        ols_model.rsquared,
        ols_model.rsquared_adj,
        ols_model.fvalue,
        ols_model.f_pvalue,
        ols_model.aic,
        ols_model.bic
    ]
})

display(regression_summary)

regression_summary.to_excel(
    os.path.join(
        RESULTS_DIR,
        "12_regression_model_summary.xlsx"
    ),
    index=False
)


# ================================================================
# 32. MULTICOLLINEARITY USING VIF
# ================================================================

X_vif = X.copy()

vif_table = pd.DataFrame()

vif_table["Variable"] = X_vif.columns

vif_table["VIF"] = [
    variance_inflation_factor(
        X_vif.values,
        i
    )
    for i in range(X_vif.shape[1])
]

display(vif_table)

vif_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "13_variance_inflation_factors.xlsx"
    ),
    index=False
)


# ================================================================
# 33. REGRESSION RESIDUALS AND FITTED VALUES
# ================================================================

fitted_values = ols_model.fittedvalues

residuals = ols_model.resid

residual_df = pd.DataFrame({
    "Observed_EPI": y,
    "Fitted_EPI": fitted_values,
    "Residual": residuals,
    "Standardised_Residual": (
        residuals /
        residuals.std()
    )
})

display(residual_df.head())

residual_df.to_excel(
    os.path.join(
        RESULTS_DIR,
        "14_regression_residuals.xlsx"
    ),
    index=False
)


# ================================================================
# 34. LINEARITY DIAGNOSTIC
# ================================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x=fitted_values,
    y=residuals
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Fitted Employee Productivity")
plt.ylabel("Regression Residuals")

plt.title(
    "Residuals versus Fitted Values: Linearity and Model Adequacy"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "15_residuals_vs_fitted.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 35. LOWESS RESIDUAL PLOT
# ================================================================

plt.figure(figsize=(9, 6))

sns.residplot(
    x=fitted_values,
    y=y,
    lowess=True
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Fitted Employee Productivity")
plt.ylabel("Residuals")

plt.title(
    "Lowess Residual Diagnostic for Linearity"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "16_lowess_residual_diagnostic.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 36. HOMOSCEDASTICITY: BREUSCH PAGAN TEST
# ================================================================

bp_test = het_breuschpagan(
    residuals,
    X_ols
)

bp_results = pd.DataFrame({
    "Statistic": [
        "Lagrange multiplier",
        "LM p value",
        "F statistic",
        "F p value"
    ],
    "Value": [
        bp_test[0],
        bp_test[1],
        bp_test[2],
        bp_test[3]
    ]
})

print("\nBreusch Pagan test")

display(bp_results)

bp_results.to_excel(
    os.path.join(
        RESULTS_DIR,
        "17_breusch_pagan_test.xlsx"
    ),
    index=False
)


# ================================================================
# 37. NORMALITY OF RESIDUALS: JARQUE BERA
# ================================================================

jb_stat, jb_pvalue, skewness, kurtosis = jarque_bera(
    residuals
)

jb_results = pd.DataFrame({
    "Statistic": [
        "Jarque Bera",
        "p value",
        "Skewness",
        "Kurtosis"
    ],
    "Value": [
        jb_stat,
        jb_pvalue,
        skewness,
        kurtosis
    ]
})

display(jb_results)

jb_results.to_excel(
    os.path.join(
        RESULTS_DIR,
        "18_jarque_bera_test.xlsx"
    ),
    index=False
)


# ================================================================
# 38. Q Q PLOT
# ================================================================

plt.figure(figsize=(8, 7))

sm.qqplot(
    residuals,
    line="45",
    fit=True
)

plt.title(
    "Normal Q Q Plot of Regression Residuals"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "19_QQ_plot_residuals.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 39. DURBIN WATSON TEST
# ================================================================

dw_value = durbin_watson(
    residuals
)

print(
    f"\nDurbin Watson statistic: {dw_value:.3f}"
)


# ================================================================
# 40. RAMSEY RESET TEST
# ================================================================

try:

    reset_result = linear_reset(
        ols_model,
        power=2,
        use_f=True
    )

    print("\nRamsey RESET test")
    print(reset_result)

except Exception as e:

    print(
        "Ramsey RESET test could not be completed:",
        e
    )


# ================================================================
# 41. TRAINING DATA PREPROCESSING PIPELINE
#
# Numerical features are retained as numerical variables because
# the methodology specifies the coded ordinal employee
# characteristics as predictors.
#
# Imputation is included inside the pipeline to prevent leakage.
# Standardisation is used for the linear model.
# ================================================================

linear_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        LinearRegression()
    )
])


# ================================================================
# 42. MACHINE LEARNING MODELS
# ================================================================

rf_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),
    (
        "model",
        RandomForestRegressor(
            random_state=42,
            n_jobs=-1
        )
    )
])


gb_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),
    (
        "model",
        GradientBoostingRegressor(
            random_state=42
        )
    )
])


xgb_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),
    (
        "model",
        XGBRegressor(
            objective="reg:squarederror",
            random_state=42,
            n_jobs=-1,
            verbosity=0
        )
    )
])


# ================================================================
# 43. FIVE FOLD CROSS VALIDATION
# ================================================================

cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "RMSE": "neg_root_mean_squared_error",
    "MAE": "neg_mean_absolute_error",
    "R2": "r2"
}


# ================================================================
# 44. CROSS VALIDATION FUNCTION
# ================================================================

def evaluate_cv(model, X_train, y_train, model_name):

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
        return_train_score=False
    )

    results = {
        "Model": model_name,
        "CV_RMSE_Mean": -scores[
            "test_RMSE"
        ].mean(),

        "CV_RMSE_SD": scores[
            "test_RMSE"
        ].std(),

        "CV_MAE_Mean": -scores[
            "test_MAE"
        ].mean(),

        "CV_MAE_SD": scores[
            "test_MAE"
        ].std(),

        "CV_R2_Mean": scores[
            "test_R2"
        ].mean(),

        "CV_R2_SD": scores[
            "test_R2"
        ].std()
    }

    return results


# ================================================================
# 45. BASELINE FIVE FOLD CROSS VALIDATION
# ================================================================

cv_results = []

cv_results.append(
    evaluate_cv(
        linear_pipeline,
        X_train,
        y_train,
        "Multiple Linear Regression"
    )
)

cv_results.append(
    evaluate_cv(
        rf_pipeline,
        X_train,
        y_train,
        "Random Forest"
    )
)

cv_results.append(
    evaluate_cv(
        gb_pipeline,
        X_train,
        y_train,
        "Gradient Boosting"
    )
)

cv_results.append(
    evaluate_cv(
        xgb_pipeline,
        X_train,
        y_train,
        "XGBoost"
    )
)

cv_table = pd.DataFrame(cv_results)

print("\nFive fold cross validation results")

display(cv_table)

cv_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "20_five_fold_cross_validation.xlsx"
    ),
    index=False
)


# ================================================================
# 46. RANDOM FOREST HYPERPARAMETER SEARCH
# ================================================================

rf_param_grid = {
    "model__n_estimators": [
        200,
        400,
        600,
        800
    ],

    "model__max_depth": [
        None,
        5,
        10,
        15,
        20
    ],

    "model__min_samples_split": [
        2,
        5,
        10
    ],

    "model__min_samples_leaf": [
        1,
        2,
        4
    ],

    "model__max_features": [
        0.6,
        0.8,
        1.0
    ]
}

rf_search = RandomizedSearchCV(
    rf_pipeline,
    param_distributions=rf_param_grid,
    n_iter=30,
    scoring="neg_root_mean_squared_error",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=0
)

rf_search.fit(
    X_train,
    y_train
)

print("\nBest Random Forest parameters:")
print(rf_search.best_params_)

print(
    "Best RF CV RMSE:",
    -rf_search.best_score_
)


# ================================================================
# 47. GRADIENT BOOSTING HYPERPARAMETER SEARCH
# ================================================================

gb_param_grid = {
    "model__n_estimators": [
        100,
        200,
        300,
        500
    ],

    "model__learning_rate": [
        0.01,
        0.03,
        0.05,
        0.10
    ],

    "model__max_depth": [
        2,
        3,
        4,
        5
    ],

    "model__min_samples_split": [
        2,
        5,
        10
    ],

    "model__min_samples_leaf": [
        1,
        2,
        4
    ],

    "model__subsample": [
        0.7,
        0.8,
        1.0
    ]
}

gb_search = RandomizedSearchCV(
    gb_pipeline,
    param_distributions=gb_param_grid,
    n_iter=30,
    scoring="neg_root_mean_squared_error",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=0
)

gb_search.fit(
    X_train,
    y_train
)

print("\nBest Gradient Boosting parameters:")
print(gb_search.best_params_)

print(
    "Best GB CV RMSE:",
    -gb_search.best_score_
)


# ================================================================
# 48. XGBOOST HYPERPARAMETER SEARCH
# ================================================================

xgb_param_grid = {
    "model__n_estimators": [
        100,
        200,
        300,
        500,
        700
    ],

    "model__learning_rate": [
        0.01,
        0.03,
        0.05,
        0.10
    ],

    "model__max_depth": [
        2,
        3,
        4,
        5,
        6
    ],

    "model__min_child_weight": [
        1,
        3,
        5
    ],

    "model__subsample": [
        0.7,
        0.8,
        1.0
    ],

    "model__colsample_bytree": [
        0.7,
        0.8,
        1.0
    ],

    "model__gamma": [
        0,
        0.1,
        0.3
    ]
}

xgb_search = RandomizedSearchCV(
    xgb_pipeline,
    param_distributions=xgb_param_grid,
    n_iter=40,
    scoring="neg_root_mean_squared_error",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=0
)

xgb_search.fit(
    X_train,
    y_train
)

print("\nBest XGBoost parameters:")
print(xgb_search.best_params_)

print(
    "Best XGBoost CV RMSE:",
    -xgb_search.best_score_
)


# ================================================================
# 49. FINAL LINEAR REGRESSION PREDICTION MODEL
# ================================================================

linear_pipeline.fit(
    X_train,
    y_train
)

linear_pred = linear_pipeline.predict(
    X_test
)


# ================================================================
# 50. FINAL OPTIMISED MODEL PREDICTIONS
# ================================================================

rf_best = rf_search.best_estimator_
gb_best = gb_search.best_estimator_
xgb_best = xgb_search.best_estimator_

rf_pred = rf_best.predict(X_test)
gb_pred = gb_best.predict(X_test)
xgb_pred = xgb_best.predict(X_test)


# ================================================================
# 51. MODEL EVALUATION FUNCTION
# ================================================================

def evaluate_model(
    model_name,
    y_true,
    predictions
):

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            predictions
        )
    )

    mae = mean_absolute_error(
        y_true,
        predictions
    )

    r2 = r2_score(
        y_true,
        predictions
    )

    return {
        "Model": model_name,
        "RMSE": rmse,
        "MAE": mae,
        "R2": r2
    }


# ================================================================
# 52. FINAL TEST SET MODEL COMPARISON
# ================================================================

test_results = []

test_results.append(
    evaluate_model(
        "Multiple Linear Regression",
        y_test,
        linear_pred
    )
)

test_results.append(
    evaluate_model(
        "Random Forest",
        y_test,
        rf_pred
    )
)

test_results.append(
    evaluate_model(
        "Gradient Boosting",
        y_test,
        gb_pred
    )
)

test_results.append(
    evaluate_model(
        "XGBoost",
        y_test,
        xgb_pred
    )
)

performance_table = pd.DataFrame(
    test_results
)

print("\n================================================")
print("FINAL INDEPENDENT TEST SET PERFORMANCE")
print("================================================")

display(
    performance_table.sort_values(
        "RMSE",
        ascending=True
    )
)

performance_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "21_final_model_performance.xlsx"
    ),
    index=False
)


# ================================================================
# 53. IDENTIFY BEST PERFORMING MACHINE LEARNING MODEL
#
# Primary criterion: lowest RMSE.
# Secondary information: MAE and R².
# Only the three tree based models are considered for the SHAP
# stage because Objective 4 refers to the best performing machine
# learning model.
# ================================================================

ml_performance = performance_table[
    performance_table["Model"].isin([
        "Random Forest",
        "Gradient Boosting",
        "XGBoost"
    ])
].copy()

best_ml_row = ml_performance.loc[
    ml_performance["RMSE"].idxmin()
]

best_ml_name = best_ml_row["Model"]

print(
    "\nBest performing machine learning model based on "
    f"test RMSE: {best_ml_name}"
)


# ================================================================
# 54. SAVE TEST SET PREDICTIONS
# ================================================================

prediction_table = X_test.copy()

prediction_table["Observed_EPI"] = y_test.values

prediction_table[
    "MLR_Predicted_EPI"
] = linear_pred

prediction_table[
    "RF_Predicted_EPI"
] = rf_pred

prediction_table[
    "GB_Predicted_EPI"
] = gb_pred

prediction_table[
    "XGB_Predicted_EPI"
] = xgb_pred

display(
    prediction_table.head(20)
)

prediction_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "22_test_set_predictions.xlsx"
    ),
    index=False
)


# ================================================================
# 55. OBSERVED VERSUS PREDICTED PLOT
# ================================================================

predictions_dictionary = {
    "Multiple Linear Regression": linear_pred,
    "Random Forest": rf_pred,
    "Gradient Boosting": gb_pred,
    "XGBoost": xgb_pred
}

for model_name, pred in predictions_dictionary.items():

    plt.figure(figsize=(7, 6))

    plt.scatter(
        y_test,
        pred,
        alpha=0.7
    )

    min_value = min(
        y_test.min(),
        pred.min()
    )

    max_value = max(
        y_test.max(),
        pred.max()
    )

    plt.plot(
        [min_value, max_value],
        [min_value, max_value],
        linestyle="--"
    )

    plt.xlabel("Observed EPI")
    plt.ylabel("Predicted EPI")

    plt.title(
        f"Observed versus Predicted Employee Productivity: "
        f"{model_name}"
    )

    plt.tight_layout()

    safe_name = (
        model_name
        .replace(" ", "_")
    )

    plt.savefig(
        os.path.join(
            RESULTS_DIR,
            f"23_observed_vs_predicted_{safe_name}.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ================================================================
# 56. CROSS VALIDATED PERFORMANCE OF OPTIMISED MODELS
# ================================================================

optimised_models = {
    "Random Forest": rf_best,
    "Gradient Boosting": gb_best,
    "XGBoost": xgb_best
}

optimised_cv_results = []

for model_name, model in optimised_models.items():

    result = evaluate_cv(
        model,
        X_train,
        y_train,
        model_name
    )

    optimised_cv_results.append(result)

optimised_cv_table = pd.DataFrame(
    optimised_cv_results
)

display(
    optimised_cv_table
)

optimised_cv_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "24_optimised_model_cross_validation.xlsx"
    ),
    index=False
)


# ================================================================
# 57. FEATURE IMPORTANCE FOR ALL TREE MODELS
# ================================================================

tree_models = {
    "Random Forest": rf_best,
    "Gradient Boosting": gb_best,
    "XGBoost": xgb_best
}

feature_importance_results = []

for model_name, pipeline in tree_models.items():

    model = pipeline.named_steps["model"]

    importance = model.feature_importances_

    for feature, value in zip(
        features,
        importance
    ):

        feature_importance_results.append({
            "Model": model_name,
            "Feature": feature,
            "Importance": value
        })

feature_importance_df = pd.DataFrame(
    feature_importance_results
)

display(
    feature_importance_df
)

feature_importance_df.to_excel(
    os.path.join(
        RESULTS_DIR,
        "25_tree_model_feature_importance.xlsx"
    ),
    index=False
)


# ================================================================
# 58. PLOT FEATURE IMPORTANCE FOR THE BEST MODEL
# ================================================================

best_pipeline = tree_models[
    best_ml_name
]

best_tree_model = best_pipeline.named_steps[
    "model"
]

best_importance = pd.DataFrame({
    "Feature": features,
    "Importance": best_tree_model.feature_importances_
}).sort_values(
    "Importance",
    ascending=True
)

plt.figure(figsize=(9, 6))

plt.barh(
    best_importance["Feature"],
    best_importance["Importance"]
)

plt.xlabel("Feature Importance")
plt.ylabel("Predictor")

plt.title(
    f"Feature Importance: {best_ml_name}"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "26_best_model_feature_importance.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 59. SHAP ANALYSIS
#
# SHAP is conducted only on the best performing machine learning
# model, as specified in the methodology.
# ================================================================

print(
    "\n================================================"
)
print(
    f"SHAP ANALYSIS FOR {best_ml_name}"
)
print(
    "================================================"
)


# Extract fitted tree model
best_tree = best_pipeline.named_steps["model"]

# Extract imputer
best_imputer = best_pipeline.named_steps["imputer"]

# Transform the training and testing data using ONLY the fitted
# training imputer.

X_train_imputed = pd.DataFrame(
    best_imputer.transform(X_train),
    columns=features,
    index=X_train.index
)

X_test_imputed = pd.DataFrame(
    best_imputer.transform(X_test),
    columns=features,
    index=X_test.index
)


# ================================================================
# 60. CREATE SHAP EXPLAINER
# ================================================================

explainer = shap.TreeExplainer(
    best_tree
)

shap_values = explainer.shap_values(
    X_test_imputed
)


# ================================================================
# 61. SHAP SUMMARY / BEESWARM PLOT
# ================================================================

plt.figure()

shap.summary_plot(
    shap_values,
    X_test_imputed,
    show=False
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "27_SHAP_summary_beeswarm.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 62. SHAP BAR PLOT
# ================================================================

plt.figure()

shap.summary_plot(
    shap_values,
    X_test_imputed,
    plot_type="bar",
    show=False
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        RESULTS_DIR,
        "28_SHAP_global_feature_importance.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 63. CALCULATE MEAN ABSOLUTE SHAP VALUES
# ================================================================

mean_abs_shap = np.abs(
    shap_values
).mean(axis=0)

shap_importance = pd.DataFrame({
    "Feature": features,
    "Mean_Absolute_SHAP": mean_abs_shap
}).sort_values(
    "Mean_Absolute_SHAP",
    ascending=False
)

shap_importance[
    "Relative_Importance_Percent"
] = (
    shap_importance[
        "Mean_Absolute_SHAP"
    ] /
    shap_importance[
        "Mean_Absolute_SHAP"
    ].sum()
) * 100

print(
    "\nGlobal SHAP feature importance"
)

display(shap_importance)

shap_importance.to_excel(
    os.path.join(
        RESULTS_DIR,
        "29_SHAP_global_importance.xlsx"
    ),
    index=False
)


# ================================================================
# 64. SHAP DEPENDENCE PLOTS FOR TOP FEATURES
# ================================================================

top_shap_features = (
    shap_importance
    .head(5)["Feature"]
    .tolist()
)

print(
    "\nTop five SHAP features:"
)

print(top_shap_features)


for feature in top_shap_features:

    plt.figure()

    shap.dependence_plot(
        feature,
        shap_values,
        X_test_imputed,
        show=False
    )

    plt.tight_layout()

    safe_feature = (
        feature
        .replace(" ", "_")
    )

    plt.savefig(
        os.path.join(
            RESULTS_DIR,
            f"30_SHAP_dependence_{safe_feature}.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ================================================================
# 65. SHAP LOCAL EXPLANATION FOR SELECTED TEST OBSERVATIONS
# ================================================================

# Select up to five test observations.

number_local_cases = min(
    5,
    len(X_test_imputed)
)

local_indices = X_test_imputed.index[
    :number_local_cases
]

local_shap_records = []

expected_value = explainer.expected_value

if np.ndim(expected_value) > 0:
    expected_value = expected_value[0]

for idx in local_indices:

    row_position = X_test_imputed.index.get_loc(
        idx
    )

    row_shap = shap_values[
        row_position
    ]

    for feature, shap_value in zip(
        features,
        row_shap
    ):

        local_shap_records.append({
            "Test_Index": idx,
            "Feature": feature,
            "SHAP_Value": shap_value,
            "Direction": (
                "Positive contribution"
                if shap_value > 0
                else
                "Negative contribution"
                if shap_value < 0
                else
                "No contribution"
            )
        })

local_shap_df = pd.DataFrame(
    local_shap_records
)

display(
    local_shap_df.head(50)
)

local_shap_df.to_excel(
    os.path.join(
        RESULTS_DIR,
        "31_SHAP_local_explanations.xlsx"
    ),
    index=False
)


# ================================================================
# 66. SHAP WATERFALL PLOT FOR FIRST TEST OBSERVATION
# ================================================================

if len(X_test_imputed) > 0:

    first_position = 0

    shap_explanation = shap.Explanation(
        values=shap_values[first_position],
        base_values=expected_value,
        data=X_test_imputed.iloc[
            first_position
        ].values,
        feature_names=features
    )

    plt.figure()

    shap.plots.waterfall(
        shap_explanation,
        show=False,
        max_display=10
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            RESULTS_DIR,
            "32_SHAP_waterfall_first_test_case.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ================================================================
# 67. SHAP DIRECTIONAL SUMMARY
#
# For each feature:
# positive_mean = average positive contribution
# negative_mean = average negative contribution
# mean_SHAP = average signed contribution
# ================================================================

directional_shap = []

for i, feature in enumerate(features):

    values = shap_values[:, i]

    positive_values = values[
        values > 0
    ]

    negative_values = values[
        values < 0
    ]

    directional_shap.append({
        "Feature": feature,
        "Mean_SHAP": np.mean(values),
        "Mean_Absolute_SHAP": np.mean(
            np.abs(values)
        ),
        "Mean_Positive_SHAP": (
            np.mean(positive_values)
            if len(positive_values) > 0
            else 0
        ),
        "Mean_Negative_SHAP": (
            np.mean(negative_values)
            if len(negative_values) > 0
            else 0
        ),
        "Positive_Contribution_Count": (
            len(positive_values)
        ),
        "Negative_Contribution_Count": (
            len(negative_values)
        )
    })

directional_shap_df = pd.DataFrame(
    directional_shap
).sort_values(
    "Mean_Absolute_SHAP",
    ascending=False
)

display(
    directional_shap_df
)

directional_shap_df.to_excel(
    os.path.join(
        RESULTS_DIR,
        "33_SHAP_directional_summary.xlsx"
    ),
    index=False
)


# ================================================================
# 68. PERMUTATION IMPORTANCE
#
# This provides a second model agnostic check of predictive
# importance.
# ================================================================

permutation = permutation_importance(
    best_pipeline,
    X_test,
    y_test,
    n_repeats=30,
    random_state=42,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

permutation_df = pd.DataFrame({
    "Feature": features,
    "Importance_Mean": permutation.importances_mean,
    "Importance_SD": permutation.importances_std
}).sort_values(
    "Importance_Mean",
    ascending=False
)

display(
    permutation_df
)

permutation_df.to_excel(
    os.path.join(
        RESULTS_DIR,
        "34_permutation_importance.xlsx"
    ),
    index=False
)


# ================================================================
# 69. COMPARE SHAP AND PERMUTATION IMPORTANCE
# ================================================================

comparison_importance = (
    shap_importance[
        [
            "Feature",
            "Mean_Absolute_SHAP"
        ]
    ]
    .merge(
        permutation_df[
            [
                "Feature",
                "Importance_Mean"
            ]
        ],
        on="Feature",
        how="left"
    )
)

display(
    comparison_importance
)

comparison_importance.to_excel(
    os.path.join(
        RESULTS_DIR,
        "35_SHAP_vs_permutation_importance.xlsx"
    ),
    index=False
)


# ================================================================
# 70. CREATE FINAL MODEL SUMMARY
# ================================================================

best_model_summary = pd.DataFrame({
    "Item": [
        "Best machine learning model",
        "Test RMSE",
        "Test MAE",
        "Test R2",
        "Training observations",
        "Testing observations",
        "Cross validation folds",
        "Primary SHAP analysis"
    ],
    "Value": [
        best_ml_name,
        best_ml_row["RMSE"],
        best_ml_row["MAE"],
        best_ml_row["R2"],
        len(X_train),
        len(X_test),
        5,
        "Global and local SHAP interpretation"
    ]
})

display(
    best_model_summary
)

best_model_summary.to_excel(
    os.path.join(
        RESULTS_DIR,
        "36_best_model_summary.xlsx"
    ),
    index=False
)


# ================================================================
# 71. OBJECTIVE 1 SUMMARY
#
# Statistical relationships are extracted from the OLS results.
# ================================================================

objective_1_table = regression_table.copy()

objective_1_table["Significance"] = np.select(
    [
        objective_1_table["p_value"] < 0.001,
        objective_1_table["p_value"] < 0.01,
        objective_1_table["p_value"] < 0.05
    ],
    [
        "***",
        "**",
        "*"
    ],
    default=""
)

print(
    "\nOBJECTIVE 1: Statistical relationship summary"
)

display(
    objective_1_table
)

objective_1_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "37_objective_1_relationships.xlsx"
    ),
    index=False
)


# ================================================================
# 72. OBJECTIVE 2 SUMMARY
#
# Models developed:
# Multiple Linear Regression
# Random Forest
# Gradient Boosting
# XGBoost
# ================================================================

objective_2_table = pd.DataFrame({
    "Model": [
        "Multiple Linear Regression",
        "Random Forest",
        "Gradient Boosting",
        "XGBoost"
    ],
    "Model_Type": [
        "Statistical regression",
        "Machine learning",
        "Machine learning",
        "Machine learning"
    ],
    "Developed": [
        "Yes",
        "Yes",
        "Yes",
        "Yes"
    ]
})

display(
    objective_2_table
)

objective_2_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "38_objective_2_models.xlsx"
    ),
    index=False
)


# ================================================================
# 73. OBJECTIVE 3 SUMMARY
# ================================================================

objective_3_table = performance_table.sort_values(
    "RMSE",
    ascending=True
).reset_index(drop=True)

objective_3_table[
    "RMSE_Rank"
] = np.arange(
    1,
    len(objective_3_table) + 1
)

print(
    "\nOBJECTIVE 3: Predictive performance comparison"
)

display(
    objective_3_table
)

objective_3_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "39_objective_3_performance_comparison.xlsx"
    ),
    index=False
)


# ================================================================
# 74. OBJECTIVE 4 SUMMARY
# ================================================================

objective_4_table = shap_importance.copy()

objective_4_table[
    "SHAP_Rank"
] = np.arange(
    1,
    len(objective_4_table) + 1
)

print(
    "\nOBJECTIVE 4: Explainable AI summary"
)

display(
    objective_4_table
)

objective_4_table.to_excel(
    os.path.join(
        RESULTS_DIR,
        "40_objective_4_SHAP_results.xlsx"
    ),
    index=False
)


# ================================================================
# 75. AUTOMATIC TEXTUAL SUMMARY OF KEY RESULTS
#
# This does NOT invent values. It reads values generated by the
# actual analysis.
# ================================================================

print("\n")
print("=" * 80)
print("AUTOMATED ANALYTICAL SUMMARY")
print("=" * 80)

print(
    f"\nThe analytical dataset contained {len(analysis_df)} "
    "complete observations."
)

print(
    f"The multiple linear regression model produced an "
    f"R² of {ols_model.rsquared:.3f} and an adjusted "
    f"R² of {ols_model.rsquared_adj:.3f}."
)

print(
    f"The best performing machine learning model based on "
    f"test RMSE was {best_ml_name}."
)

print(
    f"Its test RMSE was {best_ml_row['RMSE']:.4f}, "
    f"MAE was {best_ml_row['MAE']:.4f}, "
    f"and R² was {best_ml_row['R2']:.4f}."
)

print(
    "\nThe five most important predictors according to "
    "mean absolute SHAP value were:"
)

for rank, row in enumerate(
    shap_importance.head(5).itertuples(),
    start=1
):

    print(
        f"{rank}. {row.Feature}: "
        f"mean absolute SHAP = "
        f"{row.Mean_Absolute_SHAP:.4f}"
    )


# ================================================================
# 76. DIAGNOSTIC SUMMARY
# ================================================================

print("\n")
print("=" * 80)
print("REGRESSION DIAGNOSTIC SUMMARY")
print("=" * 80)

print(
    f"Breusch Pagan p value: {bp_test[1]:.6f}"
)

print(
    f"Jarque Bera p value: {jb_pvalue:.6f}"
)

print(
    f"Durbin Watson statistic: {dw_value:.4f}"
)

print(
    f"Overall KMO: {kmo_total:.4f}"
)

print(
    f"Bartlett chi square: {chi_square:.4f}"
)

print(
    f"Bartlett p value: {bartlett_p:.6f}"
)


# ================================================================
# 77. EXPORT COMPLETE ANALYTICAL WORKBOOK
# ================================================================

complete_excel_path = os.path.join(
    RESULTS_DIR,
    "Smart_Workforce_Analytics_Complete_Results.xlsx"
)

with pd.ExcelWriter(
    complete_excel_path,
    engine="openpyxl"
) as writer:

    descriptive.to_excel(
        writer,
        sheet_name="Descriptive",
    )

    corr_matrix.to_excel(
        writer,
        sheet_name="Correlation"
    )

    regression_table.to_excel(
        writer,
        sheet_name="OLS_Coefficients",
        index=False
    )

    regression_summary.to_excel(
        writer,
        sheet_name="OLS_Summary",
        index=False
    )

    vif_table.to_excel(
        writer,
        sheet_name="VIF",
        index=False
    )

    bp_results.to_excel(
        writer,
        sheet_name="Breusch_Pagan",
        index=False
    )

    jb_results.to_excel(
        writer,
        sheet_name="Jarque_Bera",
        index=False
    )

    pca_summary.to_excel(
        writer,
        sheet_name="PCA_Summary",
        index=False
    )

    loadings.to_excel(
        writer,
        sheet_name="PCA_Loadings"
    )

    cv_table.to_excel(
        writer,
        sheet_name="Cross_Validation",
        index=False
    )

    performance_table.to_excel(
        writer,
        sheet_name="Test_Performance",
        index=False
    )

    shap_importance.to_excel(
        writer,
        sheet_name="SHAP_Importance",
        index=False
    )

    directional_shap_df.to_excel(
        writer,
        sheet_name="SHAP_Direction",
        index=False
    )

    permutation_df.to_excel(
        writer,
        sheet_name="Permutation",
        index=False
    )

    prediction_table.to_excel(
        writer,
        sheet_name="Predictions",
        index=False
    )

print(
    "\nComplete results workbook saved to:"
)

print(
    complete_excel_path
)


# ================================================================
# 78. ZIP ALL RESULTS
# ================================================================

import shutil

zip_path = shutil.make_archive(
    RESULTS_DIR,
    "zip",
    RESULTS_DIR
)

print(
    "\nAll analysis outputs have been compressed into:"
)

print(zip_path)


# ================================================================
# 79. FINAL CHECK
# ================================================================

print("\n")
print("=" * 80)
print("ANALYSIS PIPELINE COMPLETED")
print("=" * 80)

print(
    "\nObjectives addressed:"
)

print(
    "Objective 1: Descriptive statistics, correlations, "
    "OLS regression, VIF and residual diagnostics."
)

print(
    "Objective 2: Multiple Linear Regression, Random Forest, "
    "Gradient Boosting and XGBoost."
)

print(
    "Objective 3: Five fold cross validation and independent "
    "test set evaluation using RMSE, MAE and R²."
)

print(
    "Objective 4: SHAP global importance, directional "
    "contributions, dependence plots and local explanations."
)

print(
    "\nAll generated tables, figures and model outputs are "
    "available in the Results directory."
)

print(
    RESULTS_DIR
)
```
