```python
# ================================================================
# FORECASTING ECONOMIC GROWTH IN SOMALIA USING EXTERNAL CAPITAL
# FLOWS: A STRUCTURAL BREAK AND CAUSALITY BASED TIME SERIES METHODS
#
# Study period: 1990 to 2025
# Forecast period: 2026 to 2030
#
# Objectives:
# 1. Examine historical trends and patterns of FDI, Foreign Aid
#    and GDP Growth.
# 2. Assess long run equilibrium relationships using cointegration.
# 3. Investigate causality using Toda Yamamoto and Granger tests.
# 4. Develop and compare ARIMAX and Markov Switching forecasts.
#
# Main software:
# Python
#
# ================================================================


# ================================================================
# 0. INSTALLATION OF REQUIRED PACKAGES
# ================================================================
#
# Run this section once in Google Colab or Jupyter Notebook.
# If the packages are already installed, this section can be skipped.
#
# ================================================================

# Uncomment the next line if packages need to be installed.

# !pip install pandas numpy scipy matplotlib seaborn statsmodels openpyxl


# ================================================================
# 1. IMPORT REQUIRED LIBRARIES
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

import statsmodels.api as sm

from statsmodels.tsa.stattools import (
    adfuller,
    zivot_andrews,
    grangercausalitytests
)

from statsmodels.tsa.vector_ar.var_model import VAR
from statsmodels.tsa.vector_ar.vecm import coint_johansen

from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.arima.model import ARIMA

from statsmodels.tsa.regime_switching.markov_autoregression import (
    MarkovAutoregression
)

from statsmodels.stats.diagnostic import (
    acorr_breusch_godfrey,
    het_breuschpagan,
    acorr_ljungbox
)

from statsmodels.stats.outliers_influence import variance_inflation_factor

from pathlib import Path


# ================================================================
# 2. GENERAL SETTINGS
# ================================================================

ALPHA = 0.05

STUDY_START = 1970
STUDY_END = 2025

FORECAST_START = 2026
FORECAST_END = 2030

TEST_YEARS = 3

MAX_VAR_LAGS = 4

MAX_ARIMAX_P = 2
MAX_ARIMAX_Q = 2

MAX_EXOG_ARIMA_P = 2
MAX_EXOG_ARIMA_Q = 2

RANDOM_STATE = 123


# ================================================================
# 3. ENTER THE DATA
# ================================================================
#
# The original dataset supplied by the researcher is entered here.
#
# Variable names:
# Years
# FDI
# Foreign_Aid
# GDP_Growth
#
# FDI = Foreign Direct Investment, net inflows (% of GDP)
# Foreign_Aid = Net ODA received (% of GNI)
# GDP_Growth = GDP growth, annual %
#
# ================================================================

data = """
1970 1.293157723 8.543779453 4.581407921
1971 0.489857859 9.233327154 2.13836478
1972 1.00472457 7.115849477 14.88095238
1973 0.124043929 10.00728887 -12.14936573
1974 0.12226988 17.21588035 0.752491357
1975 1.557430991 23.2309895 14.6951958
1976 0.252483449 12.49628154 -1.847940866
1977 1.560523256 47.16773288 13.64532903
1978 0.053137615 46.62528245 5.799418605
1979 0.00169616 46.42288891 -4.986948757
1980 -0.004979125 79.68692538 1.792943898
1981 0.01572996 54.74970653 6.264204545
1982 -0.098117825 61.45559773 2.767009758
1983 -1.119105696 46.46677215 -11.57648283
1984 -1.897300564 44.77136748 3.265666372
1985 -0.081077787 40.95067467 9.52991453
1986 0.333845817 57.3707794 -2.145922747
1987 6.31420275 59.54118544 10.0877193
1988 -4.126105472 41.1922299 3.415247721
1989 -3.4841175 37.00160701 3.80766467
1990 0.542043744 53.04526118 2.326930486
1991 -0.020892094 26.78992256 2.950570844
1992 -0.008459073 94.94092113 -8.488323481
1993 0.158713411 72.88960947 3.990596419
1994 0.067512595 37.39299111 -17.84698949
1995 0.065156931 12.65479395 3.990596419
1996 0.072535159 5.056856863 8.04601505
1997 0.050188228 3.803885654 1.180491846
1998 0.001350417 2.817754261 6.586493254
1999 -0.026594422 3.896414703 7.318295504
2000 0.007911398 3.066965066 7.110314311
2001 0.001775531 6.772637779 7.110314311
2002 0.006384869 7.109754822 7.630267293
2003 -0.029964138 6.332094265 7.630267293
2004 -0.124106356 5.309933517 7.110314311
2005 0.512465023 5.214982426 7.110314311
2006 1.909785215 8.001576089 6.486370733
2007 2.596343815 7.351287876 6.694351926
2008 1.471180016 13.11377238 6.694351926
2009 3.662761973 22.69019919 6.694351926
2010 4.166965851 18.90630431 6.694351926
2011 3.509979236 38.13572399 6.694351926
2012 2.459063253 22.6925891 6.694351926
2013 5.095912178 20.96221037 10.95929862
2014 4.556246295 19.46726595 8.978005906
2015 4.925108203 20.58761375 10.46218812
2016 4.989610218 17.97901853 7.0348726
2017 4.841565534 23.19655237 14.7357956
2018 5.18197836 20.0932881 2.054315073
2019 5.164630417 19.96148757 2.800333125
2020 6.188869184 35.45403239 -2.764556962
2021 6.336990356 25.50132916 3.4471985
2022 6.233602935 19.20389889 2.718212021
2023 6.167652331 20.87668881 4.175242576
2024 6.317867988 27.26881044 4.111393358
2025 1.624807901 27.26881044 2.991144045
"""


# ================================================================
# 4. CONVERT DATA TO DATAFRAME
# ================================================================

rows = []

for line in data.strip().split("\n"):
    parts = line.split()

    rows.append([
        int(parts[0]),
        float(parts[1]),
        float(parts[2]),
        float(parts[3])
    ])

df = pd.DataFrame(
    rows,
    columns=[
        "Year",
        "FDI",
        "Foreign_Aid",
        "GDP_Growth"
    ]
)


# ================================================================
# 5. BASIC DATA VALIDATION
# ================================================================

print("\n============================================================")
print("DATA VALIDATION")
print("============================================================")

print("\nComplete dataset:")
print(df.head())

print("\nLast observations:")
print(df.tail())

print("\nNumber of observations:")
print(len(df))

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate years:")
print(df["Year"].duplicated().sum())

print("\nYear range:")
print(df["Year"].min(), "to", df["Year"].max())


# Check that years are unique and ordered.

df = df.sort_values("Year").drop_duplicates("Year")

df = df.set_index("Year")

df.index = pd.Index(df.index.astype(int), name="Year")


# ================================================================
# 6. RESTRICT ANALYSIS TO THE ACTUAL STUDY PERIOD
# ================================================================

study = df.loc[STUDY_START:STUDY_END].copy()

print("\n============================================================")
print("STUDY SAMPLE")
print("============================================================")

print(study)

print("\nNumber of annual observations:")
print(len(study))

if len(study) != 36:
    print(
        "\nWARNING: The study period should contain 36 annual "
        "observations from 1990 to 2025."
    )


# ================================================================
# 7. CREATE OUTPUT DIRECTORY
# ================================================================

output_dir = Path("Somalia_Economic_Growth_Analysis")

output_dir.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# 8. SAVE CLEANED STUDY DATA
# ================================================================

study.to_csv(
    output_dir / "Somalia_Study_Data_1990_2025.csv"
)


# ================================================================
# OBJECTIVE ONE
# HISTORICAL TRENDS AND PATTERNS
# ================================================================


# ================================================================
# 9. DESCRIPTIVE STATISTICS
# ================================================================

print("\n============================================================")
print("OBJECTIVE ONE: DESCRIPTIVE STATISTICS")
print("============================================================")

descriptive = study[
    ["FDI", "Foreign_Aid", "GDP_Growth"]
].describe().T

descriptive["Median"] = study[
    ["FDI", "Foreign_Aid", "GDP_Growth"]
].median()

descriptive["Skewness"] = study[
    ["FDI", "Foreign_Aid", "GDP_Growth"]
].skew()

descriptive["Kurtosis"] = study[
    ["FDI", "Foreign_Aid", "GDP_Growth"]
].kurtosis()

descriptive = descriptive[
    [
        "count",
        "mean",
        "Median",
        "std",
        "min",
        "max",
        "Skewness",
        "Kurtosis"
    ]
]

print(descriptive.round(4))

descriptive.to_csv(
    output_dir / "Table_4_1_Descriptive_Statistics.csv"
)


# ================================================================
# 10. YEARLY CHANGE
# ================================================================

annual_change = study.pct_change() * 100

annual_change.to_csv(
    output_dir / "Annual_Percentage_Changes.csv"
)


# ================================================================
# 11. PLOT HISTORICAL TRENDS
# ================================================================

plt.figure(figsize=(12, 6))

plt.plot(
    study.index,
    study["FDI"],
    marker="o",
    label="FDI"
)

plt.plot(
    study.index,
    study["Foreign_Aid"],
    marker="s",
    label="Foreign Aid"
)

plt.xlabel("Year")
plt.ylabel("Percentage")
plt.title(
    "Historical Trends in FDI and Foreign Aid in Somalia, 1990 to 2025"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir / "Figure_4_1_FDI_Foreign_Aid_Trends.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 12. GDP GROWTH TREND
# ================================================================

plt.figure(figsize=(12, 6))

plt.plot(
    study.index,
    study["GDP_Growth"],
    marker="o"
)

plt.axhline(
    y=0,
    linestyle="--",
    linewidth=1
)

plt.xlabel("Year")
plt.ylabel("GDP Growth (%)")

plt.title(
    "Historical Pattern of Somalia's GDP Growth, 1990 to 2025"
)

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir / "Figure_4_2_GDP_Growth_Trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 13. ALL VARIABLES IN ONE STANDARDIZED GRAPH
# ================================================================

standardized = (
    study[["FDI", "Foreign_Aid", "GDP_Growth"]]
    -
    study[["FDI", "Foreign_Aid", "GDP_Growth"]].mean()
) / study[
    ["FDI", "Foreign_Aid", "GDP_Growth"]
].std()

plt.figure(figsize=(12, 6))

for column in standardized.columns:

    plt.plot(
        standardized.index,
        standardized[column],
        marker="o",
        label=column
    )

plt.axhline(
    y=0,
    linestyle="--",
    linewidth=1
)

plt.xlabel("Year")
plt.ylabel("Standardized Value")

plt.title(
    "Standardized Historical Behaviour of Study Variables"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir / "Figure_4_3_Standardized_Variables.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 14. CORRELATION MATRIX
# ================================================================

correlation = study[
    ["FDI", "Foreign_Aid", "GDP_Growth"]
].corr()

print("\nCorrelation matrix:")
print(correlation.round(4))

correlation.to_csv(
    output_dir / "Correlation_Matrix.csv"
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Matrix of Study Variables"
)

plt.tight_layout()

plt.savefig(
    output_dir / "Figure_4_4_Correlation_Matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 15. IDENTIFY EXTREME OBSERVATIONS
# ================================================================

print("\n============================================================")
print("EXTREME OBSERVATIONS")
print("============================================================")

for column in ["FDI", "Foreign_Aid", "GDP_Growth"]:

    print("\n", column)

    print(
        "Minimum:",
        study[column].idxmin(),
        study[column].min()
    )

    print(
        "Maximum:",
        study[column].idxmax(),
        study[column].max()
    )


# ================================================================
# OBJECTIVE TWO
# STRUCTURAL BREAKS, UNIT ROOTS AND COINTEGRATION
# ================================================================


# ================================================================
# 16. AUGMENTED DICKEY FULLER TEST
# ================================================================

def adf_test(series, variable_name, regression="ct"):

    series = series.dropna()

    result = adfuller(
        series,
        regression=regression,
        autolag="AIC"
    )

    output = {
        "Variable": variable_name,
        "Regression": regression,
        "ADF_Statistic": result[0],
        "P_Value": result[1],
        "Used_Lag": result[2],
        "Observations": result[3],
        "Critical_1%": result[4]["1%"],
        "Critical_5%": result[4]["5%"],
        "Critical_10%": result[4]["10%"],
        "Stationary_5%": "Yes" if result[1] < ALPHA else "No"
    }

    return output


print("\n============================================================")
print("ADF UNIT ROOT TESTS")
print("============================================================")


adf_results = []

for variable in [
    "FDI",
    "Foreign_Aid",
    "GDP_Growth"
]:

    level_result = adf_test(
        study[variable],
        variable,
        regression="ct"
    )

    adf_results.append(level_result)

    difference = study[variable].diff().dropna()

    diff_result = adf_test(
        difference,
        variable + "_First_Difference",
        regression="c"
    )

    adf_results.append(diff_result)


adf_table = pd.DataFrame(adf_results)

print(
    adf_table.round(4).to_string(index=False)
)

adf_table.to_csv(
    output_dir / "Table_4_2_ADF_Unit_Root_Test.csv",
    index=False
)


# ================================================================
# 17. DETERMINE ORDER OF INTEGRATION
# ================================================================

integration_orders = {}

for variable in [
    "FDI",
    "Foreign_Aid",
    "GDP_Growth"
]:

    level_p = adf_table.loc[
        adf_table["Variable"] == variable,
        "P_Value"
    ].iloc[0]

    diff_p = adf_table.loc[
        adf_table["Variable"] == variable + "_First_Difference",
        "P_Value"
    ].iloc[0]

    if level_p < ALPHA:

        integration_orders[variable] = 0

    elif diff_p < ALPHA:

        integration_orders[variable] = 1

    else:

        integration_orders[variable] = None


print("\n============================================================")
print("ORDER OF INTEGRATION")
print("============================================================")

for variable, order in integration_orders.items():

    print(
        variable,
        ": I(",
        order,
        ")"
    )


# ================================================================
# 18. ZIVOT ANDREWS STRUCTURAL BREAK TEST
# ================================================================
#
# The test permits one endogenous structural break.
#
# regression="ct" means:
# constant plus trend with a possible structural break.
#
# ================================================================

def zivot_andrews_test(
    series,
    variable_name,
    regression="ct"
):

    series = series.dropna()

    result = zivot_andrews(
        series,
        maxlag=4,
        regression=regression,
        autolag="AIC"
    )

    test_stat = result[0]
    p_value = result[1]
    critical_values = result[2]
    used_lag = result[3]
    break_index = result[4]

    break_year = series.index[break_index]

    return {
        "Variable": variable_name,
        "ZA_Statistic": test_stat,
        "P_Value": p_value,
        "Used_Lag": used_lag,
        "Break_Year": break_year,
        "Critical_1%": critical_values["1%"],
        "Critical_5%": critical_values["5%"],
        "Critical_10%": critical_values["10%"],
        "Reject_Unit_Root_5%":
            "Yes" if p_value < ALPHA else "No"
    }


print("\n============================================================")
print("ZIVOT ANDREWS STRUCTURAL BREAK TEST")
print("============================================================")


za_results = []

for variable in [
    "FDI",
    "Foreign_Aid",
    "GDP_Growth"
]:

    result = zivot_andrews_test(
        study[variable],
        variable,
        regression="ct"
    )

    za_results.append(result)


za_table = pd.DataFrame(za_results)

print(
    za_table.round(4).to_string(index=False)
)

za_table.to_csv(
    output_dir / "Table_4_3_Zivot_Andrews_Test.csv",
    index=False
)


# ================================================================
# 19. PLOT STRUCTURAL BREAK DATES
# ================================================================

for _, row in za_table.iterrows():

    variable = row["Variable"]

    break_year = int(row["Break_Year"])

    plt.figure(figsize=(12, 5))

    plt.plot(
        study.index,
        study[variable],
        marker="o"
    )

    plt.axvline(
        break_year,
        linestyle="--",
        linewidth=2,
        label=f"Estimated break: {break_year}"
    )

    plt.xlabel("Year")
    plt.ylabel(variable)

    plt.title(
        f"Zivot Andrews Structural Break: {variable}"
    )

    plt.legend()

    plt.grid(alpha=0.3)

    plt.tight_layout()

    plt.savefig(
        output_dir /
        f"Figure_ZA_Break_{variable}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ================================================================
# 20. VAR LAG LENGTH SELECTION
# ================================================================

print("\n============================================================")
print("VAR LAG LENGTH SELECTION")
print("============================================================")

var_data = study[
    ["GDP_Growth", "FDI", "Foreign_Aid"]
].copy()

max_allowed_lag = min(
    MAX_VAR_LAGS,
    max(1, int(len(var_data) / 8))
)

print(
    "Maximum lag considered:",
    max_allowed_lag
)

var_model = VAR(var_data)

lag_selection = var_model.select_order(
    maxlags=max_allowed_lag
)

print(lag_selection.summary())


# ================================================================
# 21. SELECT OPTIMAL LAG
# ================================================================

try:

    selected_lag = lag_selection.aic

except Exception:

    selected_lag = None


if selected_lag is None or selected_lag < 1:

    selected_lag = 1


selected_lag = int(selected_lag)

print(
    "\nSelected VAR lag based on AIC:",
    selected_lag
)


# ================================================================
# 22. JOHANSEN COINTEGRATION TEST
# ================================================================
#
# Johansen is only interpreted when all variables are I(1).
#
# If the variables are not all I(1), the code will clearly report
# that Johansen should not be used as the primary long run test.
#
# ================================================================

print("\n============================================================")
print("JOHANSEN COINTEGRATION TEST")
print("============================================================")


all_I1 = all(
    integration_orders[v] == 1
    for v in [
        "FDI",
        "Foreign_Aid",
        "GDP_Growth"
    ]
)


if all_I1:

    johansen_data = var_data.dropna()

    johansen_lags = max(
        selected_lag - 1,
        0
    )

    johansen_result = coint_johansen(
        johansen_data,
        det_order=0,
        k_ar_diff=johansen_lags
    )

    trace_table = pd.DataFrame({
        "Rank": [
            "None",
            "At most 1",
            "At most 2"
        ],
        "Trace_Statistic":
            johansen_result.lr1,
        "Critical_90":
            johansen_result.cvt[:, 0],
        "Critical_95":
            johansen_result.cvt[:, 1],
        "Critical_99":
            johansen_result.cvt[:, 2]
    })

    trace_table["Cointegration_At_5%"] = np.where(
        trace_table["Trace_Statistic"]
        >
        trace_table["Critical_95"],
        "Yes",
        "No"
    )

    print("\nTrace test:")
    print(
        trace_table.round(4).to_string(index=False)
    )


    max_eigen_table = pd.DataFrame({
        "Rank":
            [
                "None versus 1",
                "1 versus 2",
                "2 versus 3"
            ],
        "Max_Eigen_Statistic":
            johansen_result.lr2,
        "Critical_90":
            johansen_result.cvm[:, 0],
        "Critical_95":
            johansen_result.cvm[:, 1],
        "Critical_99":
            johansen_result.cvm[:, 2]
    })

    max_eigen_table["Cointegration_At_5%"] = np.where(
        max_eigen_table["Max_Eigen_Statistic"]
        >
        max_eigen_table["Critical_95"],
        "Yes",
        "No"
    )

    print("\nMaximum eigenvalue test:")
    print(
        max_eigen_table.round(4).to_string(index=False)
    )


    trace_table.to_csv(
        output_dir /
        "Table_4_4_Johansen_Trace_Test.csv",
        index=False
    )

    max_eigen_table.to_csv(
        output_dir /
        "Table_4_5_Johansen_Maximum_Eigenvalue_Test.csv",
        index=False
    )


else:

    print(
        "\nJohansen cointegration test was not automatically "
        "interpreted because the ADF results do not show that "
        "all three variables are I(1)."
    )

    print(
        "\nThis is intentional. Do not force a Johansen result "
        "when the integration orders do not satisfy the required "
        "conditions."
    )

    print(
        "\nObserved integration orders:"
    )

    print(integration_orders)


# ================================================================
# OBJECTIVE THREE
# GRANGER AND TODA YAMAMOTO CAUSALITY
# ================================================================


# ================================================================
# 23. CONVENTIONAL GRANGER CAUSALITY
# ================================================================
#
# The conventional Granger test requires stationary variables.
#
# Therefore, each variable is transformed according to the
# stationarity results obtained above.
#
# If a variable is I(0), its level is retained.
# If it is I(1), its first difference is used.
#
# ================================================================

print("\n============================================================")
print("CONVENTIONAL GRANGER CAUSALITY")
print("============================================================")


stationary_data = pd.DataFrame(index=study.index)

for variable in [
    "GDP_Growth",
    "FDI",
    "Foreign_Aid"
]:

    order = integration_orders[variable]

    if order == 0:

        stationary_data[variable] = study[variable]

    elif order == 1:

        stationary_data[variable] = study[variable].diff()

    else:

        stationary_data[variable] = study[variable].diff()


stationary_data = stationary_data.dropna()


# Keep the maximum lag conservative because the sample is small.

granger_maxlag = min(
    selected_lag,
    2
)

print(
    "Granger maximum lag:",
    granger_maxlag
)


def run_granger_test(
    data,
    caused,
    causing,
    maxlag
):

    test_data = data[
        [caused, causing]
    ].dropna()

    results = []

    for lag in range(1, maxlag + 1):

        try:

            output = grangercausalitytests(
                test_data,
                maxlag=[lag],
                verbose=False
            )

            ssr_ftest = output[
                lag
            ][0]["ssr_ftest"]

            results.append({
                "Caused": caused,
                "Causing": causing,
                "Lag": lag,
                "F_Statistic": ssr_ftest[0],
                "P_Value": ssr_ftest[1],
                "Decision_5%":
                    "Reject H0"
                    if ssr_ftest[1] < ALPHA
                    else "Fail to Reject H0"
            })

        except Exception as error:

            print(
                f"Granger test failed for "
                f"{causing} -> {caused}, lag {lag}: {error}"
            )

    return pd.DataFrame(results)


granger_results = []

pairs = [
    ("GDP_Growth", "FDI"),
    ("FDI", "GDP_Growth"),
    ("GDP_Growth", "Foreign_Aid"),
    ("Foreign_Aid", "GDP_Growth"),
    ("FDI", "Foreign_Aid"),
    ("Foreign_Aid", "FDI")
]


for caused, causing in pairs:

    result = run_granger_test(
        stationary_data,
        caused,
        causing,
        granger_maxlag
    )

    granger_results.append(result)


granger_table = pd.concat(
    granger_results,
    ignore_index=True
)

print(
    granger_table.round(4).to_string(index=False)
)

granger_table.to_csv(
    output_dir /
    "Table_4_6_Granger_Causality.csv",
    index=False
)


# ================================================================
# 24. TODA YAMAMOTO CAUSALITY
# ================================================================
#
# Toda Yamamoto estimates an augmented VAR:
#
# VAR order = k + dmax
#
# where:
#
# k = optimal VAR lag
# dmax = maximum order of integration
#
# The Wald test is applied only to the first k lags.
#
# ================================================================


print("\n============================================================")
print("TODA YAMAMOTO CAUSALITY")
print("============================================================")


valid_orders = [
    value
    for value in integration_orders.values()
    if value is not None
]

if len(valid_orders) > 0:

    dmax = max(valid_orders)

else:

    dmax = 1


print(
    "Maximum integration order:",
    dmax
)


# Conservative lower bound.

k = max(
    1,
    selected_lag
)

augmented_lag = k + dmax

print(
    "Optimal VAR lag k:",
    k
)

print(
    "Augmented lag k + dmax:",
    augmented_lag
)


# ================================================================
# 25. CREATE AUGMENTED VAR DATASET
# ================================================================

ty_data = var_data.copy()

variables = [
    "GDP_Growth",
    "FDI",
    "Foreign_Aid"
]


def create_augmented_var_matrix(
    data,
    variables,
    lags
):

    temp = data[variables].copy()

    X = pd.DataFrame(
        index=temp.index
    )

    X["const"] = 1.0

    for lag in range(1, lags + 1):

        for variable in variables:

            X[
                f"{variable}_L{lag}"
            ] = temp[variable].shift(lag)

    return X


X_ty = create_augmented_var_matrix(
    ty_data,
    variables,
    augmented_lag
)


# ================================================================
# 26. TODA YAMAMOTO WALD FUNCTION
# ================================================================

def toda_yamamoto_wald(
    data,
    causing,
    caused,
    k,
    dmax,
    variables
):

    augmented_lag = k + dmax

    Y = data[caused].copy()

    X = create_augmented_var_matrix(
        data,
        variables,
        augmented_lag
    )

    combined = pd.concat(
        [Y.rename("Y"), X],
        axis=1
    ).dropna()

    Y_clean = combined["Y"]

    X_clean = combined.drop(
        columns=["Y"]
    )

    model = sm.OLS(
        Y_clean,
        X_clean
    ).fit()

    beta = model.params.values

    covariance = model.cov_params().values

    parameter_names = list(
        X_clean.columns
    )

    restriction_indices = []

    for lag in range(1, k + 1):

        parameter_name = (
            f"{causing}_L{lag}"
        )

        if parameter_name in parameter_names:

            restriction_indices.append(
                parameter_names.index(
                    parameter_name
                )
            )


    if len(restriction_indices) == 0:

        return None


    R = np.zeros(
        (
            len(restriction_indices),
            len(beta)
        )
    )

    for row, column in enumerate(
        restriction_indices
    ):

        R[row, column] = 1.0


    restriction = R @ beta

    middle = (
        R
        @ covariance
        @ R.T
    )

    try:

        inverse_middle = np.linalg.pinv(
            middle
        )

        wald_statistic = (
            restriction.T
            @ inverse_middle
            @ restriction
        )

    except Exception:

        wald_statistic = np.nan


    df_wald = len(
        restriction_indices
    )

    p_value = (
        1
        -
        stats.chi2.cdf(
            wald_statistic,
            df_wald
        )
    )


    return {
        "Caused": caused,
        "Causing": causing,
        "k": k,
        "dmax": dmax,
        "Augmented_Lag": augmented_lag,
        "Wald_Statistic": wald_statistic,
        "DF": df_wald,
        "P_Value": p_value,
        "Decision_5%":
            "Reject H0"
            if p_value < ALPHA
            else "Fail to Reject H0"
    }


ty_results = []

for caused, causing in pairs:

    result = toda_yamamoto_wald(
        ty_data,
        causing,
        caused,
        k,
        dmax,
        variables
    )

    if result is not None:

        ty_results.append(result)


ty_table = pd.DataFrame(
    ty_results
)

print(
    ty_table.round(4).to_string(index=False)
)

ty_table.to_csv(
    output_dir /
    "Table_4_7_Toda_Yamamoto_Causality.csv",
    index=False
)


# ================================================================
# OBJECTIVE FOUR
# ARIMAX FORECASTING
# ================================================================


# ================================================================
# 27. TRAINING AND TESTING DATA
# ================================================================
#
# Last three observed years:
#
# 2023
# 2024
# 2025
#
# are retained as a holdout test set.
#
# Training:
# 1990 to 2022
#
# Testing:
# 2023 to 2025
#
# Final forecast:
# 2026 to 2030
#
# ================================================================


test_start = STUDY_END - TEST_YEARS + 1

train = study.loc[
    :test_start - 1
].copy()

test = study.loc[
    test_start:STUDY_END
].copy()

print("\n============================================================")
print("FORECAST TRAINING AND TESTING")
print("============================================================")

print(
    "Training period:",
    train.index.min(),
    "to",
    train.index.max()
)

print(
    "Testing period:",
    test.index.min(),
    "to",
    test.index.max()
)


# ================================================================
# 28. ARIMAX MODEL SELECTION
# ================================================================
#
# Since GDP growth is a growth rate, d=0 will normally be preferred
# if the stationarity test confirms that GDP Growth is I(0).
#
# However, the code uses the ADF result to determine the differencing
# order.
#
# ================================================================

gdp_order = integration_orders["GDP_Growth"]

if gdp_order is None:

    gdp_order = 0


def fit_arimax_grid(
    y,
    exog,
    max_p=2,
    max_q=2,
    d=0
):

    results = []

    best_model = None
    best_aic = np.inf
    best_order = None

    for p in range(
        0,
        max_p + 1
    ):

        for q in range(
            0,
            max_q + 1
        ):

            order = (
                p,
                d,
                q
            )

            try:

                model = SARIMAX(
                    y,
                    exog=exog,
                    order=order,
                    trend="c",
                    enforce_stationarity=False,
                    enforce_invertibility=False
                )

                fitted = model.fit(
                    disp=False
                )

                aic = fitted.aic
                bic = fitted.bic

                results.append({
                    "p": p,
                    "d": d,
                    "q": q,
                    "AIC": aic,
                    "BIC": bic
                })

                if np.isfinite(aic):

                    if aic < best_aic:

                        best_aic = aic
                        best_model = fitted
                        best_order = order

            except Exception as error:

                print(
                    "ARIMAX failure:",
                    order,
                    error
                )


    return (
        best_model,
        best_order,
        pd.DataFrame(results)
    )


arimax_train_y = train["GDP_Growth"]

arimax_train_x = train[
    ["FDI", "Foreign_Aid"]
]


print("\n============================================================")
print("ARIMAX MODEL SELECTION")
print("============================================================")


arimax_model, arimax_order, arimax_selection = fit_arimax_grid(
    arimax_train_y,
    arimax_train_x,
    max_p=MAX_ARIMAX_P,
    max_q=MAX_ARIMAX_Q,
    d=gdp_order
)


print(
    "\nSelected ARIMAX order:",
    arimax_order
)

print(
    "\nCandidate ARIMAX models:"
)

print(
    arimax_selection.sort_values(
        "AIC"
    ).round(4).to_string(index=False)
)


arimax_selection.to_csv(
    output_dir /
    "Table_4_8_ARIMAX_Model_Selection.csv",
    index=False
)


# ================================================================
# 29. ARIMAX MODEL SUMMARY
# ================================================================

if arimax_model is not None:

    print(
        "\nARIMAX model summary:"
    )

    print(
        arimax_model.summary()
    )

else:

    raise RuntimeError(
        "No valid ARIMAX model could be estimated."
    )


# ================================================================
# 30. ARIMAX HOLDOUT FORECAST
# ================================================================

arimax_test_forecast = arimax_model.get_forecast(
    steps=len(test),
    exog=test[
        ["FDI", "Foreign_Aid"]
    ]
)

arimax_test_prediction = (
    arimax_test_forecast.predicted_mean
)

arimax_test_ci = (
    arimax_test_forecast.conf_int()
)


# ================================================================
# 31. FORECAST ACCURACY FUNCTIONS
# ================================================================

def rmse(actual, predicted):

    actual = np.asarray(actual)
    predicted = np.asarray(predicted)

    return np.sqrt(
        np.mean(
            (actual - predicted) ** 2
        )
    )


def mae(actual, predicted):

    actual = np.asarray(actual)
    predicted = np.asarray(predicted)

    return np.mean(
        np.abs(
            actual - predicted
        )
    )


def mape(actual, predicted):

    actual = np.asarray(actual)
    predicted = np.asarray(predicted)

    denominator = np.where(
        np.abs(actual) < 1e-8,
        np.nan,
        np.abs(actual)
    )

    return np.nanmean(
        np.abs(
            (actual - predicted)
            /
            denominator
        )
    ) * 100


def forecast_accuracy(
    actual,
    predicted
):

    return {
        "RMSE": rmse(
            actual,
            predicted
        ),
        "MAE": mae(
            actual,
            predicted
        ),
        "MAPE": mape(
            actual,
            predicted
        )
    }


arimax_accuracy = forecast_accuracy(
    test["GDP_Growth"],
    arimax_test_prediction
)


print("\n============================================================")
print("ARIMAX HOLDOUT FORECAST ACCURACY")
print("============================================================")

print(
    pd.Series(
        arimax_accuracy
    ).round(4)
)


# ================================================================
# 32. MARKOV SWITCHING MODEL SELECTION
# ================================================================
#
# Two regimes are specified:
#
# Regime 0 = lower growth regime
# Regime 1 = higher growth regime
#
# The interpretation is determined after estimation rather than
# imposed before estimation.
#
# ================================================================


print("\n============================================================")
print("MARKOV SWITCHING MODEL SELECTION")
print("============================================================")


def fit_markov_models(
    y,
    max_order=2
):

    results = []

    best_model = None
    best_aic = np.inf
    best_specification = None

    for regimes in [2, 3]:

        for order in range(
            1,
            max_order + 1
        ):

            try:

                model = MarkovAutoregression(
                    y,
                    k_regimes=regimes,
                    order=order,
                    trend="c",
                    switching_ar=False
                )

                fitted = model.fit(
                    em_iter=20,
                    search_reps=20,
                    disp=False
                )

                results.append({
                    "Regimes": regimes,
                    "AR_Order": order,
                    "AIC": fitted.aic,
                    "BIC": fitted.bic,
                    "LogLikelihood":
                        fitted.llf
                })

                if np.isfinite(
                    fitted.aic
                ):

                    if fitted.aic < best_aic:

                        best_aic = fitted.aic
                        best_model = fitted
                        best_specification = (
                            regimes,
                            order
                        )

            except Exception as error:

                print(
                    "Markov model failure:",
                    regimes,
                    order,
                    error
                )


    return (
        best_model,
        best_specification,
        pd.DataFrame(results)
    )


ms_train_y = train[
    "GDP_Growth"
].copy()


ms_model, ms_specification, ms_selection = fit_markov_models(
    ms_train_y,
    max_order=2
)


print(
    "\nSelected Markov Switching specification:",
    ms_specification
)

print(
    "\nCandidate Markov Switching models:"
)

print(
    ms_selection.sort_values(
        "AIC"
    ).round(4).to_string(index=False)
)


ms_selection.to_csv(
    output_dir /
    "Table_4_9_Markov_Switching_Model_Selection.csv",
    index=False
)


# ================================================================
# 33. MARKOV SWITCHING SUMMARY
# ================================================================

if ms_model is not None:

    print(
        "\nMarkov Switching model summary:"
    )

    print(
        ms_model.summary()
    )

else:

    raise RuntimeError(
        "No valid Markov Switching model could be estimated."
    )


# ================================================================
# 34. MARKOV SWITCHING HOLDOUT FORECAST
# ================================================================
#
# statsmodels permits out of sample prediction through predict().
#
# Because annual data are being used, the forecast begins immediately
# after the final training observation.
#
# ================================================================

try:

    ms_test_prediction = ms_model.predict(
        start=len(train),
        end=len(train) + len(test) - 1,
        probabilities="predicted"
    )

    ms_test_prediction = pd.Series(
        np.asarray(
            ms_test_prediction
        ),
        index=test.index
    )

except Exception as error:

    print(
        "\nDirect Markov Switching forecasting failed:"
    )

    print(error)

    print(
        "\nA recursive regime probability forecast will be attempted."
    )

    # Last available predicted probabilities

    transition_matrix = np.asarray(
        ms_model.regime_transition
    )

    if transition_matrix.ndim == 3:

        P = transition_matrix[:, :, -1]

    else:

        P = transition_matrix

    last_prob = np.asarray(
        ms_model.filtered_marginal_probabilities.iloc[-1]
    )

    probabilities = last_prob.copy()

    regime_means = []

    try:

        params = ms_model.params

        for regime in range(
            ms_model.k_regimes
        ):

            regime_mean = (
                ms_model.params[
                    regime
                ]
                if regime < len(
                    ms_model.params
                )
                else ms_train_y.mean()
            )

            regime_means.append(
                regime_mean
            )

        regime_means = np.asarray(
            regime_means
        )

        forecasts = []

        for step in range(
            len(test)
        ):

            probabilities = (
                probabilities @ P
            )

            forecasts.append(
                np.sum(
                    probabilities
                    *
                    regime_means
                )
            )

        ms_test_prediction = pd.Series(
            forecasts,
            index=test.index
        )

    except Exception:

        ms_test_prediction = pd.Series(
            np.repeat(
                ms_train_y.mean(),
                len(test)
            ),
            index=test.index
        )


# ================================================================
# 35. MARKOV SWITCHING HOLDOUT ACCURACY
# ================================================================

ms_accuracy = forecast_accuracy(
    test["GDP_Growth"],
    ms_test_prediction
)


print("\n============================================================")
print("MARKOV SWITCHING HOLDOUT FORECAST ACCURACY")
print("============================================================")

print(
    pd.Series(
        ms_accuracy
    ).round(4)
)


# ================================================================
# 36. COMPARE ARIMAX AND MARKOV SWITCHING
# ================================================================

accuracy_table = pd.DataFrame({
    "Model": [
        "ARIMAX",
        "Markov Switching"
    ],
    "RMSE": [
        arimax_accuracy["RMSE"],
        ms_accuracy["RMSE"]
    ],
    "MAE": [
        arimax_accuracy["MAE"],
        ms_accuracy["MAE"]
    ],
    "MAPE": [
        arimax_accuracy["MAPE"],
        ms_accuracy["MAPE"]
    ]
})


print("\n============================================================")
print("FORECAST ACCURACY COMPARISON")
print("============================================================")

print(
    accuracy_table.round(4).to_string(index=False)
)


accuracy_table.to_csv(
    output_dir /
    "Table_4_10_ARIMAX_vs_Markov_Switching_Accuracy.csv",
    index=False
)


# ================================================================
# 37. IDENTIFY BEST FORECASTING MODEL
# ================================================================

best_rmse_model = accuracy_table.loc[
    accuracy_table["RMSE"].idxmin(),
    "Model"
]

best_mae_model = accuracy_table.loc[
    accuracy_table["MAE"].idxmin(),
    "Model"
]

best_mape_model = accuracy_table.loc[
    accuracy_table["MAPE"].idxmin(),
    "Model"
]


print(
    "\nBest model based on RMSE:",
    best_rmse_model
)

print(
    "Best model based on MAE:",
    best_mae_model
)

print(
    "Best model based on MAPE:",
    best_mape_model
)


# ================================================================
# 38. HOLDOUT FORECAST COMPARISON TABLE
# ================================================================

holdout_forecasts = pd.DataFrame({
    "Actual_GDP_Growth":
        test["GDP_Growth"],
    "ARIMAX_Forecast":
        arimax_test_prediction,
    "Markov_Switching_Forecast":
        ms_test_prediction
})


holdout_forecasts.to_csv(
    output_dir /
    "Table_4_11_Holdout_Forecasts_2023_2025.csv"
)


print(
    "\nHoldout forecasts:"
)

print(
    holdout_forecasts.round(4)
)


# ================================================================
# 39. PLOT HOLDOUT FORECASTS
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    test.index,
    test["GDP_Growth"],
    marker="o",
    linewidth=2,
    label="Actual GDP Growth"
)

plt.plot(
    test.index,
    arimax_test_prediction,
    marker="s",
    linestyle="--",
    label="ARIMAX"
)

plt.plot(
    test.index,
    ms_test_prediction,
    marker="^",
    linestyle=":",
    label="Markov Switching"
)

plt.xlabel("Year")
plt.ylabel("GDP Growth (%)")

plt.title(
    "Holdout Forecast Comparison, 2023 to 2025"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir /
    "Figure_4_5_Holdout_Forecast_Comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# FUTURE FORECASTING: 2026 TO 2030
# ================================================================
#
# IMPORTANT:
#
# ARIMAX requires future values of the exogenous variables.
#
# Since FDI and Foreign Aid for 2026 to 2030 are not supplied,
# they must first be forecast.
#
# This code forecasts FDI and Foreign Aid using simple ARIMA models.
#
# These are auxiliary forecasts used only to provide future
# exogenous inputs to ARIMAX.
#
# ================================================================


# ================================================================
# 40. AUXILIARY FORECAST FUNCTION FOR EXOGENOUS VARIABLES
# ================================================================

def select_exog_arima(
    series,
    max_p=2,
    max_q=2
):

    results = []

    best_model = None
    best_order = None
    best_aic = np.inf

    for p in range(
        0,
        max_p + 1
    ):

        for q in range(
            0,
            max_q + 1
        ):

            try:

                model = ARIMA(
                    series,
                    order=(p, 0, q),
                    trend="c"
                )

                fitted = model.fit()

                results.append({
                    "p": p,
                    "d": 0,
                    "q": q,
                    "AIC": fitted.aic,
                    "BIC": fitted.bic
                })

                if fitted.aic < best_aic:

                    best_aic = fitted.aic
                    best_model = fitted
                    best_order = (
                        p,
                        0,
                        q
                    )

            except Exception:

                pass


    return (
        best_model,
        best_order,
        pd.DataFrame(results)
    )


# ================================================================
# 41. FORECAST FDI
# ================================================================

fdi_aux_model, fdi_aux_order, fdi_aux_selection = (
    select_exog_arima(
        study["FDI"]
    )
)


fdi_future = fdi_aux_model.forecast(
    steps=5
)

fdi_future.index = range(
    FORECAST_START,
    FORECAST_END + 1
)


# ================================================================
# 42. FORECAST FOREIGN AID
# ================================================================

aid_aux_model, aid_aux_order, aid_aux_selection = (
    select_exog_arima(
        study["Foreign_Aid"]
    )
)


aid_future = aid_aux_model.forecast(
    steps=5
)

aid_future.index = range(
    FORECAST_START,
    FORECAST_END + 1
)


future_exog = pd.DataFrame({
    "FDI": fdi_future,
    "Foreign_Aid": aid_future
})


print("\n============================================================")
print("FORECASTS OF EXOGENOUS VARIABLES")
print("============================================================")

print(
    future_exog.round(4)
)


future_exog.to_csv(
    output_dir /
    "Table_4_12_Future_Exogenous_Forecasts.csv"
)


# ================================================================
# 43. FINAL ARIMAX MODEL USING 1990 TO 2025
# ================================================================

final_arimax_model = SARIMAX(
    study["GDP_Growth"],
    exog=study[
        ["FDI", "Foreign_Aid"]
    ],
    order=arimax_order,
    trend="c",
    enforce_stationarity=False,
    enforce_invertibility=False
).fit(
    disp=False
)


print(
    "\n============================================================"
)

print(
    "FINAL ARIMAX MODEL"
)

print(
    "============================================================"
)

print(
    final_arimax_model.summary()
)


# ================================================================
# 44. FINAL ARIMAX FORECAST 2026 TO 2030
# ================================================================

final_arimax_forecast_result = (
    final_arimax_model.get_forecast(
        steps=5,
        exog=future_exog[
            ["FDI", "Foreign_Aid"]
        ]
    )
)


arimax_future_forecast = (
    final_arimax_forecast_result.predicted_mean
)

arimax_future_ci = (
    final_arimax_forecast_result.conf_int()
)


arimax_future_table = pd.DataFrame({
    "Year":
        range(
            FORECAST_START,
            FORECAST_END + 1
        ),
    "FDI_Forecast":
        future_exog["FDI"].values,
    "Foreign_Aid_Forecast":
        future_exog["Foreign_Aid"].values,
    "ARIMAX_GDP_Growth_Forecast":
        arimax_future_forecast.values,
    "Lower_95":
        arimax_future_ci.iloc[:, 0].values,
    "Upper_95":
        arimax_future_ci.iloc[:, 1].values
})


print("\n============================================================")
print("ARIMAX GDP GROWTH FORECAST: 2026 TO 2030")
print("============================================================")

print(
    arimax_future_table.round(4).to_string(index=False)
)


arimax_future_table.to_csv(
    output_dir /
    "Table_4_13_ARIMAX_GDP_Forecast_2026_2030.csv",
    index=False
)


# ================================================================
# 45. FINAL MARKOV SWITCHING MODEL USING 1990 TO 2025
# ================================================================

final_ms_model = MarkovAutoregression(
    study["GDP_Growth"],
    k_regimes=ms_specification[0],
    order=ms_specification[1],
    trend="c",
    switching_ar=False
).fit(
    em_iter=30,
    search_reps=30,
    disp=False
)


print(
    "\n============================================================"
)

print(
    "FINAL MARKOV SWITCHING MODEL"
)

print(
    "============================================================"
)

print(
    final_ms_model.summary()
)


# ================================================================
# 46. MARKOV SWITCHING FORECAST 2026 TO 2030
# ================================================================

try:

    ms_future_prediction = final_ms_model.predict(
        start=len(study),
        end=len(study) + 4,
        probabilities="predicted"
    )

    ms_future_prediction = np.asarray(
        ms_future_prediction
    )

except Exception as error:

    print(
        "\nMarkov Switching direct forecast error:"
    )

    print(error)

    # Fallback forecast based on last estimated regime
    # probability distribution.

    try:

        P = np.asarray(
            final_ms_model.regime_transition
        )

        if P.ndim == 3:

            P = P[:, :, -1]

        probabilities = np.asarray(
            final_ms_model
            .filtered_marginal_probabilities
            .iloc[-1]
        )

        regime_means = []

        for regime in range(
            final_ms_model.k_regimes
        ):

            regime_observations = study[
                final_ms_model
                .smoothed_marginal_probabilities[
                    regime
                ] > 0.5
            ]["GDP_Growth"]

            if len(
                regime_observations
            ) > 0:

                regime_means.append(
                    regime_observations.mean()
                )

            else:

                regime_means.append(
                    study["GDP_Growth"].mean()
                )

        regime_means = np.asarray(
            regime_means
        )

        forecasts = []

        for step in range(5):

            probabilities = (
                probabilities @ P
            )

            forecasts.append(
                np.sum(
                    probabilities
                    *
                    regime_means
                )
            )

        ms_future_prediction = np.asarray(
            forecasts
        )

    except Exception:

        ms_future_prediction = np.repeat(
            study["GDP_Growth"].mean(),
            5
        )


# ================================================================
# 47. FINAL FORECAST COMPARISON
# ================================================================

future_comparison = pd.DataFrame({
    "Year":
        range(
            FORECAST_START,
            FORECAST_END + 1
        ),
    "ARIMAX_Forecast":
        arimax_future_forecast.values,
    "ARIMAX_Lower_95":
        arimax_future_ci.iloc[:, 0].values,
    "ARIMAX_Upper_95":
        arimax_future_ci.iloc[:, 1].values,
    "Markov_Switching_Forecast":
        ms_future_prediction
})


print("\n============================================================")
print("FINAL GDP GROWTH FORECAST COMPARISON: 2026 TO 2030")
print("============================================================")

print(
    future_comparison.round(4).to_string(index=False)
)


future_comparison.to_csv(
    output_dir /
    "Table_4_14_Final_GDP_Forecast_2026_2030.csv",
    index=False
)


# ================================================================
# 48. FINAL FORECAST GRAPH
# ================================================================

plt.figure(figsize=(12, 6))

plt.plot(
    study.index,
    study["GDP_Growth"],
    marker="o",
    label="Historical GDP Growth"
)

plt.plot(
    future_comparison["Year"],
    future_comparison["ARIMAX_Forecast"],
    marker="s",
    linestyle="--",
    label="ARIMAX Forecast"
)

plt.plot(
    future_comparison["Year"],
    future_comparison["Markov_Switching_Forecast"],
    marker="^",
    linestyle=":",
    label="Markov Switching Forecast"
)

plt.axvline(
    2025,
    linestyle="--",
    linewidth=1
)

plt.xlabel("Year")

plt.ylabel("GDP Growth (%)")

plt.title(
    "Somalia GDP Growth Forecasts, 2026 to 2030"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir /
    "Figure_4_6_GDP_Growth_Forecast_2026_2030.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 49. MARKOV SWITCHING REGIME PROBABILITIES
# ================================================================

try:

    probabilities = (
        final_ms_model
        .smoothed_marginal_probabilities
        .copy()
    )

    probabilities.index = study.index[
        -len(probabilities):
    ]

    plt.figure(figsize=(12, 6))

    for regime in probabilities.columns:

        plt.plot(
            probabilities.index,
            probabilities[regime],
            marker="o",
            label=f"Regime {regime + 1}"
        )

    plt.xlabel("Year")

    plt.ylabel("Smoothed Probability")

    plt.title(
        "Markov Switching Smoothed Regime Probabilities"
    )

    plt.legend()

    plt.grid(alpha=0.3)

    plt.tight_layout()

    plt.savefig(
        output_dir /
        "Figure_4_7_Markov_Switching_Regime_Probabilities.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    probabilities.to_csv(
        output_dir /
        "Table_4_15_Markov_Switching_Regime_Probabilities.csv"
    )

except Exception as error:

    print(
        "\nRegime probability plot could not be produced:"
    )

    print(error)


# ================================================================
# 50. MARKOV SWITCHING REGIME CLASSIFICATION
# ================================================================

try:

    smoothed_probabilities = (
        final_ms_model
        .smoothed_marginal_probabilities
    )

    regime_assignment = (
        smoothed_probabilities
        .idxmax(axis=1)
    )

    regime_table = pd.DataFrame({
        "Year":
            study.index[-len(regime_assignment):],
        "GDP_Growth":
            study["GDP_Growth"]
            .iloc[-len(regime_assignment):]
            .values,
        "Most_Likely_Regime":
            regime_assignment.values
    })

    print(
        "\nMost likely economic regime by year:"
    )

    print(
        regime_table.to_string(index=False)
    )

    regime_table.to_csv(
        output_dir /
        "Table_4_16_Most_Likely_Economic_Regime.csv",
        index=False
    )

except Exception as error:

    print(
        "\nRegime classification could not be produced:"
    )

    print(error)


# ================================================================
# DIAGNOSTIC TESTS
# ================================================================


# ================================================================
# 51. ARIMAX RESIDUAL DIAGNOSTICS
# ================================================================

print("\n============================================================")
print("ARIMAX DIAGNOSTIC TESTS")
print("============================================================")


arimax_residuals = (
    final_arimax_model.resid
    .dropna()
)


# Jarque Bera

jb_stat, jb_pvalue, skewness, kurtosis = (
    stats.jarque_bera(
        arimax_residuals
    )[0],
    stats.jarque_bera(
        arimax_residuals
    )[1],
    stats.skew(
        arimax_residuals
    ),
    stats.kurtosis(
        arimax_residuals,
        fisher=False
    )
)


# Ljung Box

ljung_box = acorr_ljungbox(
    arimax_residuals,
    lags=[
        min(
            1,
            max(
                1,
                len(arimax_residuals) // 5
            )
        )
    ],
    return_df=True
)


diagnostics = pd.DataFrame({
    "Test": [
        "Jarque Bera",
        "Ljung Box"
    ],
    "Statistic": [
        jb_stat,
        ljung_box["lb_stat"].iloc[-1]
    ],
    "P_Value": [
        jb_pvalue,
        ljung_box["lb_pvalue"].iloc[-1]
    ]
})


diagnostics["Decision_5%"] = np.where(
    diagnostics["P_Value"] < ALPHA,
    "Reject H0",
    "Fail to Reject H0"
)


print(
    diagnostics.round(4)
)


diagnostics.to_csv(
    output_dir /
    "Table_4_17_ARIMAX_Diagnostic_Tests.csv",
    index=False
)


# ================================================================
# 52. RESIDUAL PLOT
# ================================================================

plt.figure(figsize=(12, 5))

plt.plot(
    arimax_residuals.index,
    arimax_residuals
)

plt.axhline(
    y=0,
    linestyle="--",
    linewidth=1
)

plt.xlabel("Year")

plt.ylabel("Residual")

plt.title(
    "ARIMAX Residuals"
)

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_dir /
    "Figure_4_8_ARIMAX_Residuals.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 53. RESIDUAL ACF
# ================================================================

print("\n============================================================")
print("ARIMAX RESIDUAL AUTOCORRELATION FUNCTION")
print("============================================================")

# Determine a safe number of lags.
# The sample is small, so the number of lags is deliberately limited.
acf_lags = min(
    10,
    max(1, len(arimax_residuals) // 3)
)

# Create the figure first.
# figsize belongs to plt.figure(), not plot_acf().
plt.figure(figsize=(10, 5))

# Plot the ACF without passing figsize to plot_acf().
sm.graphics.tsa.plot_acf(
    arimax_residuals,
    lags=acf_lags,
    alpha=0.05,
    zero=True
)

plt.title("ACF of ARIMAX Residuals")
plt.xlabel("Lag")
plt.ylabel("Autocorrelation")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_dir /
    "Figure_4_9_ARIMAX_Residual_ACF.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ================================================================
# 53. OPTIONAL NUMERICAL CONFIRMATION OF RESIDUAL AUTOCORRELATION
# ================================================================

# Ljung Box test is already part of the diagnostic analysis.
# Here we provide a more detailed Ljung Box table for several lags.

ljung_lags = [
    lag for lag in [1, 2, 3, 4, 5]
    if lag < len(arimax_residuals)
]

if len(ljung_lags) > 0:

    ljung_box_detailed = acorr_ljungbox(
        arimax_residuals,
        lags=ljung_lags,
        return_df=True
    )

    ljung_box_detailed = (
        ljung_box_detailed
        .reset_index()
        .rename(
            columns={
                "index": "Lag",
                "lb_stat": "Ljung_Box_Statistic",
                "lb_pvalue": "P_Value"
            }
        )
    )

    ljung_box_detailed["Decision_5%"] = np.where(
        ljung_box_detailed["P_Value"] < ALPHA,
        "Reject H0",
        "Fail to Reject H0"
    )

    print("\nDetailed Ljung Box Test:")
    print(
        ljung_box_detailed.round(4).to_string(
            index=False
        )
    )

    ljung_box_detailed.to_csv(
        output_dir /
        "Table_4_20_Detailed_Ljung_Box_Test.csv",
        index=False
    )

else:

    print(
        "\nInsufficient residual observations "
        "for the detailed Ljung Box test."
    )


# ================================================================
# 54. HETEROSKEDASTICITY CHECK
# ================================================================
#
# Breusch Pagan is more naturally associated with regression
# residuals than pure ARIMA errors. It is included because it
# appears in the proposed Chapter Three methodology.
#
# The result should therefore be interpreted cautiously.
#
# ================================================================

try:

    X_bp = sm.add_constant(
        study[
            ["FDI", "Foreign_Aid"]
        ]
    )

    bp_lm, bp_lm_pvalue, bp_f, bp_f_pvalue = (
        het_breuschpagan(
            arimax_residuals.reindex(
                study.index
            ).fillna(0),
            X_bp
        )
    )

    bp_table = pd.DataFrame({
        "Test": [
            "Breusch Pagan LM",
            "Breusch Pagan F"
        ],
        "Statistic": [
            bp_lm,
            bp_f
        ],
        "P_Value": [
            bp_lm_pvalue,
            bp_f_pvalue
        ]
    })

    bp_table["Decision_5%"] = np.where(
        bp_table["P_Value"] < ALPHA,
        "Reject H0",
        "Fail to Reject H0"
    )

    print(
        "\nBreusch Pagan test:"
    )

    print(
        bp_table.round(4)
    )

    bp_table.to_csv(
        output_dir /
        "Table_4_18_Breusch_Pagan_Test.csv",
        index=False
    )

except Exception as error:

    print(
        "\nBreusch Pagan test failed:"
    )

    print(error)


# ================================================================
# 55. MULTICOLLINEARITY CHECK
# ================================================================
#
# This is applied to FDI and Foreign Aid because they are the
# explanatory variables in the ARIMAX specification.
#
# ================================================================

vif_data = study[
    ["FDI", "Foreign_Aid"]
].copy()

vif_data = sm.add_constant(
    vif_data
)


vif_results = []

for i, column in enumerate(
    vif_data.columns
):

    if column == "const":

        continue

    vif_results.append({
        "Variable": column,
        "VIF":
            variance_inflation_factor(
                vif_data.values,
                i
            )
    })


vif_table = pd.DataFrame(
    vif_results
)

print(
    "\n============================================================"
)

print(
    "VARIANCE INFLATION FACTOR"
)

print(
    "============================================================"
)

print(
    vif_table.round(4)
)


vif_table.to_csv(
    output_dir /
    "Table_4_19_VIF_Test.csv",
    index=False
)


# ================================================================
# 56. SAVE ALL IMPORTANT RESULTS TO ONE EXCEL WORKBOOK
# ================================================================

excel_file = (
    output_dir /
    "Somalia_Economic_Growth_Complete_Results.xlsx"
)


with pd.ExcelWriter(
    excel_file,
    engine="openpyxl"
) as writer:

    study.to_excel(
        writer,
        sheet_name="Study_Data"
    )

    descriptive.to_excel(
        writer,
        sheet_name="Descriptive"
    )

    correlation.to_excel(
        writer,
        sheet_name="Correlation"
    )

    adf_table.to_excel(
        writer,
        sheet_name="ADF",
        index=False
    )

    za_table.to_excel(
        writer,
        sheet_name="Zivot_Andrews",
        index=False
    )

    if all_I1:

        trace_table.to_excel(
            writer,
            sheet_name="Johansen_Trace",
            index=False
        )

        max_eigen_table.to_excel(
            writer,
            sheet_name="Johansen_MaxEig",
            index=False
        )

    granger_table.to_excel(
        writer,
        sheet_name="Granger",
        index=False
    )

    ty_table.to_excel(
        writer,
        sheet_name="Toda_Yamamoto",
        index=False
    )

    arimax_selection.to_excel(
        writer,
        sheet_name="ARIMAX_Selection",
        index=False
    )

    ms_selection.to_excel(
        writer,
        sheet_name="MS_Selection",
        index=False
    )

    accuracy_table.to_excel(
        writer,
        sheet_name="Accuracy",
        index=False
    )

    holdout_forecasts.to_excel(
        writer,
        sheet_name="Holdout_Forecasts"
    )

    future_exog.to_excel(
        writer,
        sheet_name="Future_Exogenous"
    )

    arimax_future_table.to_excel(
        writer,
        sheet_name="ARIMAX_Forecast",
        index=False
    )

    future_comparison.to_excel(
        writer,
        sheet_name="Final_Forecast",
        index=False
    )

    diagnostics.to_excel(
        writer,
        sheet_name="Diagnostics",
        index=False
    )

    vif_table.to_excel(
        writer,
        sheet_name="VIF",
        index=False
    )


# ================================================================
# 57. FINAL AUTOMATED RESULTS SUMMARY
# ================================================================

print("\n")
print("============================================================")
print("FINAL ANALYSIS SUMMARY")
print("============================================================")

print(
    "\nStudy period:",
    f"{STUDY_START} to {STUDY_END}"
)

print(
    "Number of observations:",
    len(study)
)

print(
    "\nIntegration orders:"
)

for variable, order in integration_orders.items():

    print(
        f"{variable}: I({order})"
    )


print(
    "\nSelected VAR lag:",
    selected_lag
)

print(
    "Maximum integration order:",
    dmax
)

print(
    "\nARIMAX selected order:",
    arimax_order
)

print(
    "Markov Switching specification:",
    ms_specification
)

print(
    "\nBest holdout model by RMSE:",
    best_rmse_model
)

print(
    "Best holdout model by MAE:",
    best_mae_model
)

print(
    "Best holdout model by MAPE:",
    best_mape_model
)

print(
    "\n2026 to 2030 GDP growth forecasts:"
)

print(
    future_comparison.round(4)
)

print(
    "\nAll results have been saved in:"
)

print(
    output_dir.resolve()
)

print(
    "\nExcel workbook:"
)

print(
    excel_file.resolve()
)

# ================================================================
# END OF COMPLETE ANALYSIS CODE
# ================================================================
```
