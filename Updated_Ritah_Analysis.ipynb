```python
# ================================================================
# STOCHASTIC MODELLING OF MALNUTRITION STATUS AMONG CHILDREN
# UNDER FIVE YEARS IN UGANDA
#
# COMPLETE BOOTSTRAP ANALYSIS PIPELINE
#
# Objectives addressed:
#
# Objective 1:
# Identify and quantify demographic, socioeconomic, health and
# environmental factors associated with malnutrition status.
#
# Objective 2:
# Develop and apply stochastic models for representing and
# simulating probabilistic transitions among nutritional states.
#
# Objective 3:
# Evaluate predictive performance and uncertainty of the fitted
# statistical and stochastic models.
#
# Objective 4:
# Examine spatial patterns and regional heterogeneity and identify
# areas with elevated predicted nutritional risk.
#
# IMPORTANT:
# The supplied dataset contains only 14 child/household level rows.
# Bootstrap resampling is used to quantify uncertainty, but it does
# NOT create new empirical information or convert the dataset into
# a nationally representative UDHS sample.
#
# The Markov component is treated as MODEL BASED STOCHASTIC
# SIMULATION because the supplied data are cross sectional and the
# transition counts are not verified longitudinal observations of
# the same children.
#
# Every result is classified as:
#   A. OBSERVED EMPIRICAL
#   B. ESTIMATED MODEL
#   C. SIMULATED
# ================================================================


# ================================================================
# 0. INSTALL / IMPORT PACKAGES
# ================================================================

!pip -q install pandas numpy scipy scikit-learn matplotlib seaborn openpyxl statsmodels

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.special import expit
from scipy.stats import chi2

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import (
    roc_auc_score,
    roc_curve,
    brier_score_loss,
    log_loss,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from sklearn.model_selection import StratifiedKFold, cross_val_predict

import statsmodels.api as sm

np.random.seed(2026)

print("Packages loaded successfully.")


# ================================================================
# 1. ENTER THE SUPPLIED DATASET
# ================================================================

data = pd.DataFrame({

    "Child_ID": range(1, 15),

    "Household_ID": [
        "HH01","HH02","HH03","HH04","HH05","HH06","HH07",
        "HH08","HH09","HH10","HH11","HH12","HH13","HH14"
    ],

    "Region": [
        "Kampala","Buganda","Busoga","Bukedi","Elgon","Teso",
        "Karamoja","Lango","Acholi","West Nile","Bunyoro","Toro",
        "Ankole","Kigezi"
    ],

    "Age_Months": [
        "<6","12–23","24–35","36–47","18–23","06–11",
        "48-59","<6","12–23","24–35","36–47","18–23",
        "06–11","48-59"
    ],

    "Sex": [
        "Male","Female","Male","Female","Male","Female","Male",
        "Female","Male","Female","Male","Female","Male","Female"
    ],

    "Nutrition_Status": [
        "Normal","Stunted","Underweight","Normal","Normal","Stunted",
        "Stunted","Overweight","Overweight","Overweight","Underweight",
        "Underweight","Underweight","Overweight"
    ],

    "Stunted": [
        218,735,480,374,306,441,311,298,265,372,366,344,313,189
    ],

    "Overweight": [
        220,735,480,374,309,445,313,299,264,377,367,346,312,190
    ],

    "Underweight": [
        221,739,480,375,309,444,314,301,266,375,369,346,314,189
    ],

    "Maternal_Education": [
        "Higher","None","Primary","Secondary","Primary","Higher","None",
        "Primary","Secondary","Primary","None","Primary","Secondary","Primary"
    ],

    "Wealth_Quintile": [
        "Fifth","Lowest","Second","Fourth","Third","Fifth","Lowest",
        "Second","Fourth","Third","Fifth","Lowest","Second","Fourth"
    ],

    "VitaminA_24h": [
        24.2,34.6,42.4,33.8,28.6,56.1,56.9,22.6,44.5,71.3,
        47.5,28.0,34.4,34.3
    ],

    "Iron_24h": [
        45.1,41.0,33.0,31.3,22.0,44.7,12.8,12.3,23.4,54.4,
        29.0,33.2,18.9,12.1
    ],

    "Iodized_Salt": [
        99.8,98.7,100.0,94.1,100.0,97.7,99.8,100.0,99.5,99.6,
        96.8,98.6,99.3,100.0
    ],

    "Immunization_Status": [
        "Yes","No","Yes","Yes","Yes","Yes","No","Yes","Yes","Yes",
        "No","Yes","Yes","Yes"
    ],

    "Birth_Order": [
        1,3,2,4,1,2,5,2,3,2,4,3,2,1
    ],

    "Rainfall_Index": [
        0.75,0.45,0.55,0.65,0.70,0.68,0.50,0.62,0.58,0.66,
        0.60,0.64,0.57,0.69
    ],

    "Health_Access_Index": [
        0.90,0.40,0.60,0.70,0.80,0.75,0.45,0.65,0.55,0.72,
        0.62,0.70,0.60,0.78
    ],

    "Time_Point": [
        2022,2023,2022,2023,2022,2023,2022,2023,2022,2023,
        2022,2023,2022,2023
    ],

    "From_State": [
        "Normal","Normal","Normal","Normal","Stunted","Stunted",
        "Stunted","Stunted","Underweight","Underweight",
        "Underweight","Underweight","Overweight","Overweight"
    ],

    "To_State": [
        "Normal","Stunted","Wasted","Underweight","Normal","Stunted",
        "Underweight","Overweight","Normal","Stunted","Underweight",
        "Overweight","Normal","Overweight"
    ],

    "Count": [
        350,120,60,90,100,210,85,70,95,80,250,60,110,200
    ]
})


# ================================================================
# 2. BASIC DATA QUALITY CHECK
# ================================================================

print("\n============================================================")
print("DATA QUALITY CHECK")
print("============================================================")

print("\nDataset dimensions:")
print(data.shape)

print("\nFirst five observations:")
display(data.head())

print("\nMissing values:")
print(data.isnull().sum())

print("\nNutrition status frequencies:")
print(data["Nutrition_Status"].value_counts())

print("\nRegions:")
print(data["Region"].tolist())

print("\nNumber of unique children:", data["Child_ID"].nunique())
print("Number of unique households:", data["Household_ID"].nunique())


# ================================================================
# 3. IMPORTANT DATA STRUCTURE WARNING
# ================================================================

print("\n============================================================")
print("METHODOLOGICAL DATA STRUCTURE WARNING")
print("============================================================")

print("""
The supplied dataset contains 14 child/household records.

Bootstrap resampling will be used to quantify sampling uncertainty
around the supplied observations.

However, bootstrap resampling does NOT create new children, does NOT
increase the underlying information content, and does NOT make the
dataset nationally representative.

The transition counts are treated as model inputs for stochastic
simulation rather than observed longitudinal transitions.

Therefore:
A = Observed empirical results
B = Estimated model results
C = Simulated results
""")


# ================================================================
# 4. STANDARDIZE AGE GROUPS
# ================================================================

age_mapping = {
    "<6": 3,
    "06–11": 8,
    "06-11": 8,
    "12–23": 18,
    "12-23": 18,
    "18–23": 21,
    "18-23": 21,
    "24–35": 30,
    "24-35": 30,
    "36–47": 42,
    "36-47": 42,
    "48-59": 54,
    "48–59": 54
}

data["Age_Months_Numeric"] = data["Age_Months"].map(age_mapping)

if data["Age_Months_Numeric"].isnull().any():
    print("WARNING: Some age categories could not be converted.")


# ================================================================
# 5. CREATE BINARY MALNUTRITION OUTCOME
# ================================================================

# For this analysis, malnutrition includes:
# Stunted, Underweight and Overweight.
#
# This definition should be stated explicitly in the thesis.
# If the study intends to define malnutrition differently, modify
# this line accordingly.

malnutrition_states = [
    "Stunted",
    "Underweight",
    "Overweight"
]

data["Malnutrition"] = (
    data["Nutrition_Status"]
    .isin(malnutrition_states)
    .astype(int)
)

data["Malnutrition_Label"] = data["Malnutrition"].map({
    0: "Normal",
    1: "Malnourished"
})

print("\nObserved empirical malnutrition classification:")
print(data["Malnutrition_Label"].value_counts())


# ================================================================
# 6. OBJECTIVE 1
# DESCRIPTIVE STATISTICS
# ================================================================

print("\n============================================================")
print("OBJECTIVE 1: OBSERVED EMPIRICAL DESCRIPTIVE ANALYSIS")
print("============================================================")


# Categorical variables
categorical_vars = [
    "Nutrition_Status",
    "Sex",
    "Maternal_Education",
    "Wealth_Quintile",
    "Immunization_Status",
    "Region"
]

print("\nCategorical variable distributions:")

categorical_tables = {}

for var in categorical_vars:

    tab = (
        data[var]
        .value_counts(dropna=False)
        .rename_axis(var)
        .reset_index(name="Frequency")
    )

    tab["Percentage"] = (
        tab["Frequency"] / len(data) * 100
    ).round(2)

    categorical_tables[var] = tab

    print("\n", var)
    display(tab)


# Continuous variables
continuous_vars = [
    "VitaminA_24h",
    "Iron_24h",
    "Iodized_Salt",
    "Birth_Order",
    "Rainfall_Index",
    "Health_Access_Index"
]

continuous_summary = data[continuous_vars].describe().T

continuous_summary["Median"] = data[continuous_vars].median()

continuous_summary["IQR"] = (
    data[continuous_vars].quantile(0.75)
    -
    data[continuous_vars].quantile(0.25)
)

continuous_summary = continuous_summary[
    ["mean","std","Median","IQR","min","max"]
]

continuous_summary.columns = [
    "Mean",
    "SD",
    "Median",
    "IQR",
    "Minimum",
    "Maximum"
]

print("\nContinuous variable descriptive statistics:")
display(continuous_summary.round(3))


# ================================================================
# 7. BOOTSTRAP FUNCTION
# ================================================================

def bootstrap_sample(df, random_state=None):
    """
    Nonparametric bootstrap:
    sample rows with replacement using the same number of rows.
    """

    return df.sample(
        n=len(df),
        replace=True,
        random_state=random_state
    ).reset_index(drop=True)


# Number of bootstrap replications
B = 2000

print("\nNumber of bootstrap replications:", B)


# ================================================================
# 8. BOOTSTRAP MALNUTRITION PREVALENCE
# ================================================================

bootstrap_prevalence = []

rng = np.random.default_rng(2026)

for b in range(B):

    indices = rng.integers(
        low=0,
        high=len(data),
        size=len(data)
    )

    boot = data.iloc[indices]

    prevalence = boot["Malnutrition"].mean()

    bootstrap_prevalence.append(prevalence)

bootstrap_prevalence = np.array(bootstrap_prevalence)

observed_prev = data["Malnutrition"].mean()

ci_low = np.percentile(
    bootstrap_prevalence,
    2.5
)

ci_high = np.percentile(
    bootstrap_prevalence,
    97.5
)

print("\nObserved malnutrition prevalence:",
      round(observed_prev * 100, 2), "%")

print(
    "Bootstrap 95% CI:",
    round(ci_low * 100, 2),
    "% to",
    round(ci_high * 100, 2),
    "%"
)


# ================================================================
# 9. PLOT BOOTSTRAP DISTRIBUTION
# ================================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    bootstrap_prevalence,
    bins=30,
    kde=True
)

plt.axvline(
    observed_prev,
    linestyle="--",
    linewidth=2,
    label="Observed prevalence"
)

plt.axvline(
    ci_low,
    linestyle=":",
    linewidth=2,
    label="2.5th percentile"
)

plt.axvline(
    ci_high,
    linestyle=":",
    linewidth=2,
    label="97.5th percentile"
)

plt.xlabel("Malnutrition prevalence")
plt.ylabel("Bootstrap frequency")
plt.title(
    "Bootstrap Distribution of Observed Malnutrition Prevalence"
)

plt.legend()
plt.tight_layout()
plt.show()


# ================================================================
# 10. OBJECTIVE 1
# REGULARIZED LOGISTIC REGRESSION
# ================================================================

print("\n============================================================")
print("OBJECTIVE 1: ESTIMATED REGULARIZED LOGISTIC MODEL")
print("============================================================")

# Predictors
numeric_predictors = [
    "Age_Months_Numeric",
    "VitaminA_24h",
    "Iron_24h",
    "Iodized_Salt",
    "Birth_Order",
    "Rainfall_Index",
    "Health_Access_Index"
]

categorical_predictors = [
    "Sex",
    "Maternal_Education",
    "Wealth_Quintile",
    "Immunization_Status"
]

predictors = numeric_predictors + categorical_predictors

X = data[predictors]
y = data["Malnutrition"]


# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            StandardScaler(),
            numeric_predictors
        ),
        (
            "categorical",
            OneHotEncoder(
                drop="first",
                handle_unknown="ignore"
            ),
            categorical_predictors
        )
    ]
)


# L2 regularized logistic regression
logistic_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            LogisticRegression(
                penalty="l2",
                C=1.0,
                solver="liblinear",
                max_iter=5000
            )
        )
    ]
)


logistic_model.fit(X, y)

pred_prob = logistic_model.predict_proba(X)[:, 1]

pred_class = (
    pred_prob >= 0.5
).astype(int)

print("\nRegularized logistic model fitted successfully.")


# ================================================================
# 11. EXTRACT REGULARIZED LOGISTIC COEFFICIENTS
# ================================================================

feature_names = (
    logistic_model
    .named_steps["preprocessor"]
    .get_feature_names_out()
)

coefficients = (
    logistic_model
    .named_steps["model"]
    .coef_[0]
)

logistic_results = pd.DataFrame({
    "Variable": feature_names,
    "B": coefficients,
    "Odds_Ratio": np.exp(coefficients)
})

logistic_results["Direction"] = np.where(
    logistic_results["B"] > 0,
    "Higher odds",
    "Lower odds"
)

print("\nEstimated regularized logistic regression coefficients:")
display(
    logistic_results
    .sort_values(
        "Odds_Ratio",
        ascending=False
    )
    .round(4)
)


# ================================================================
# 12. BOOTSTRAP LOGISTIC REGRESSION
# ================================================================

# ================================================================
# CORRECTED BOOTSTRAP LOGISTIC REGRESSION
# FIXES:
# ValueError: setting an array element with a sequence
#
# Reason:
# Different bootstrap samples can contain different categorical
# levels. Therefore, OneHotEncoder may produce different numbers
# of columns in different bootstrap replications.
#
# Solution:
# 1. Freeze the categorical levels using the ORIGINAL dataset.
# 2. Reuse the same fixed design structure in every bootstrap.
# 3. Store coefficients in a fixed length vector.
# 4. Calculate bootstrap ORs and 95% percentile confidence intervals.
# ================================================================

print("\n============================================================")
print("CORRECTED BOOTSTRAP LOGISTIC REGRESSION")
print("============================================================")

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# ------------------------------------------------
# 1. DEFINE VARIABLES
# ------------------------------------------------

numeric_predictors = [
    "Age_Months_Numeric",
    "VitaminA_24h",
    "Iron_24h",
    "Iodized_Salt",
    "Birth_Order",
    "Rainfall_Index",
    "Health_Access_Index"
]

categorical_predictors = [
    "Sex",
    "Maternal_Education",
    "Wealth_Quintile",
    "Immunization_Status"
]

predictors = (
    numeric_predictors +
    categorical_predictors
)

X = data[predictors].copy()
y = data["Malnutrition"].astype(int).copy()


# ------------------------------------------------
# 2. FREEZE CATEGORICAL LEVELS FROM THE ORIGINAL DATA
# ------------------------------------------------
#
# This is the key correction.
#
# The encoder will always use exactly the same categories,
# even if a particular bootstrap sample does not contain
# one of those categories.

fixed_categories = []

for variable in categorical_predictors:

    categories = sorted(
        data[variable]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    fixed_categories.append(categories)


# ------------------------------------------------
# 3. CREATE A FIXED PREPROCESSOR
# ------------------------------------------------

fixed_preprocessor = ColumnTransformer(

    transformers=[

        (
            "numeric",
            StandardScaler(),
            numeric_predictors
        ),

        (
            "categorical",
            OneHotEncoder(
                categories=fixed_categories,
                drop="first",
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_predictors
        )

    ],

    remainder="drop"
)


# ------------------------------------------------
# 4. FIT THE PREPROCESSOR ON THE ORIGINAL DATA ONLY
# ------------------------------------------------
#
# This establishes one fixed design matrix structure
# for every bootstrap replication.

fixed_preprocessor.fit(X)


# ------------------------------------------------
# 5. GET FIXED FEATURE NAMES
# ------------------------------------------------

feature_names = (
    fixed_preprocessor
    .get_feature_names_out()
)

n_features = len(feature_names)

print(
    "Number of fixed model features:",
    n_features
)

print("\nFixed model features:")

for i, feature in enumerate(
    feature_names,
    start=1
):

    print(
        f"{i:02d}. {feature}"
    )


# ------------------------------------------------
# 6. FIT THE ORIGINAL REGULARIZED LOGISTIC MODEL
# ------------------------------------------------

X_original_transformed = (
    fixed_preprocessor
    .transform(X)
)

original_logistic = LogisticRegression(
    penalty="l2",
    C=1.0,
    solver="liblinear",
    max_iter=5000,
    random_state=2026
)

original_logistic.fit(
    X_original_transformed,
    y
)

original_coefficients = (
    original_logistic.coef_[0]
)

original_intercept = (
    original_logistic.intercept_[0]
)


# ------------------------------------------------
# 7. BOOTSTRAP
# ------------------------------------------------

B = 2000

rng = np.random.default_rng(2026)

# Preallocate a fixed rectangular matrix.
#
# This is another important protection against the
# inhomogeneous shape error.

bootstrap_coefficients = np.full(
    (B, n_features),
    np.nan,
    dtype=float
)

bootstrap_intercepts = np.full(
    B,
    np.nan,
    dtype=float
)

successful_bootstrap_models = 0


# ------------------------------------------------
# 8. RUN BOOTSTRAP MODELS
# ------------------------------------------------

for b in range(B):

    # Draw bootstrap observations
    bootstrap_indices = rng.integers(
        low=0,
        high=len(data),
        size=len(data)
    )

    boot = data.iloc[
        bootstrap_indices
    ].copy()

    Xb = boot[predictors]
    yb = boot["Malnutrition"].astype(int)

    # A logistic model requires both outcome classes
    if yb.nunique() < 2:
        continue

    try:

        # Transform using the FIXED preprocessor.
        #
        # Do NOT refit the encoder here.

        Xb_transformed = (
            fixed_preprocessor
            .transform(Xb)
        )

        model_b = LogisticRegression(
            penalty="l2",
            C=1.0,
            solver="liblinear",
            max_iter=5000,
            random_state=2026 + b
        )

        model_b.fit(
            Xb_transformed,
            yb
        )

        coef_b = (
            model_b.coef_[0]
        )

        # Safety check
        if len(coef_b) != n_features:

            print(
                f"Bootstrap {b}: "
                f"unexpected coefficient length "
                f"{len(coef_b)}"
            )

            continue

        # Store into fixed size matrix
        bootstrap_coefficients[
            b,
            :
        ] = coef_b

        bootstrap_intercepts[
            b
        ] = model_b.intercept_[0]

        successful_bootstrap_models += 1

    except Exception as e:

        # Continue rather than terminating the entire
        # bootstrap analysis.

        continue


# ------------------------------------------------
# 9. REMOVE UNSUCCESSFUL BOOTSTRAP REPLICATIONS
# ------------------------------------------------

valid_rows = (
    ~np.isnan(
        bootstrap_intercepts
    )
)

bootstrap_coefficients_valid = (
    bootstrap_coefficients[
        valid_rows,
        :
    ]
)

bootstrap_intercepts_valid = (
    bootstrap_intercepts[
        valid_rows
    ]
)


# ------------------------------------------------
# 10. REPORT BOOTSTRAP SUCCESS
# ------------------------------------------------

print(
    "\nSuccessful bootstrap logistic models:",
    successful_bootstrap_models,
    "out of",
    B
)

print(
    "Valid coefficient matrix shape:",
    bootstrap_coefficients_valid.shape
)


# ------------------------------------------------
# 11. VERIFY HOMOGENEOUS SHAPE
# ------------------------------------------------

if (
    bootstrap_coefficients_valid.ndim != 2
    or
    bootstrap_coefficients_valid.shape[1]
    != n_features
):

    raise ValueError(
        "Bootstrap coefficient matrix is not rectangular."
    )

print(
    "\nCoefficient matrix successfully constructed."
)

print(
    "Rows = successful bootstrap replications"
)

print(
    "Columns = fixed model predictors"
)


# ------------------------------------------------
# 12. BOOTSTRAP COEFFICIENT SUMMARY
# ------------------------------------------------

bootstrap_logistic_summary = []


for j, feature in enumerate(
    feature_names
):

    values = (
        bootstrap_coefficients_valid[
            :,
            j
        ]
    )

    beta_mean = np.mean(
        values
    )

    beta_median = np.median(
        values
    )

    beta_sd = np.std(
        values,
        ddof=1
    )

    beta_lower = np.percentile(
        values,
        2.5
    )

    beta_upper = np.percentile(
        values,
        97.5
    )

    # Convert coefficient to odds ratio
    or_mean = np.exp(
        beta_mean
    )

    or_median = np.exp(
        beta_median
    )

    or_lower = np.exp(
        beta_lower
    )

    or_upper = np.exp(
        beta_upper
    )

    # Proportion of bootstrap coefficients
    # greater than zero

    probability_positive = np.mean(
        values > 0
    )

    probability_negative = np.mean(
        values < 0
    )

    # Direction based on bootstrap mean

    if beta_mean > 0:

        direction = "Higher odds"

    elif beta_mean < 0:

        direction = "Lower odds"

    else:

        direction = "No directional effect"


    bootstrap_logistic_summary.append({

        "Variable": feature,

        "Bootstrap_B_Mean": beta_mean,

        "Bootstrap_B_Median": beta_median,

        "Bootstrap_B_SD": beta_sd,

        "B_Lower_95CI": beta_lower,

        "B_Upper_95CI": beta_upper,

        "OR_Mean": or_mean,

        "OR_Median": or_median,

        "OR_Lower_95CI": or_lower,

        "OR_Upper_95CI": or_upper,

        "Posterior_like_P(B>0)": probability_positive,

        "Posterior_like_P(B<0)": probability_negative,

        "Direction": direction

    })


# ------------------------------------------------
# 13. CREATE RESULTS TABLE
# ------------------------------------------------

bootstrap_logistic_summary = pd.DataFrame(
    bootstrap_logistic_summary
)


# ------------------------------------------------
# 14. ADD INTERCEPT
# ------------------------------------------------

intercept_values = (
    bootstrap_intercepts_valid
)

intercept_row = pd.DataFrame({

    "Variable": ["Intercept"],

    "Bootstrap_B_Mean": [
        np.mean(intercept_values)
    ],

    "Bootstrap_B_Median": [
        np.median(intercept_values)
    ],

    "Bootstrap_B_SD": [
        np.std(
            intercept_values,
            ddof=1
        )
    ],

    "B_Lower_95CI": [
        np.percentile(
            intercept_values,
            2.5
        )
    ],

    "B_Upper_95CI": [
        np.percentile(
            intercept_values,
            97.5
        )
    ],

    "OR_Mean": [
        np.exp(
            np.mean(intercept_values)
        )
    ],

    "OR_Median": [
        np.exp(
            np.median(intercept_values)
        )
    ],

    "OR_Lower_95CI": [
        np.exp(
            np.percentile(
                intercept_values,
                2.5
            )
        )
    ],

    "OR_Upper_95CI": [
        np.exp(
            np.percentile(
                intercept_values,
                97.5
            )
        )
    ],

    "Posterior_like_P(B>0)": [
        np.mean(
            intercept_values > 0
        )
    ],

    "Posterior_like_P(B<0)": [
        np.mean(
            intercept_values < 0
        )
    ],

    "Direction": [
        "Intercept"
    ]

})


bootstrap_logistic_summary = pd.concat(
    [
        intercept_row,
        bootstrap_logistic_summary
    ],
    ignore_index=True
)


# ------------------------------------------------
# 15. DISPLAY RESULTS
# ------------------------------------------------

print(
    "\n============================================================"
)

print(
    "BOOTSTRAP LOGISTIC REGRESSION RESULTS"
)

print(
    "============================================================"
)

display(
    bootstrap_logistic_summary.round(4)
)


# ------------------------------------------------
# 16. ORIGINAL MODEL COEFFICIENTS AND ODDS RATIOS
# ------------------------------------------------

original_logistic_results = pd.DataFrame({

    "Variable": feature_names,

    "B": original_coefficients,

    "Odds_Ratio": np.exp(
        original_coefficients
    )

})

print(
    "\n============================================================"
)

print(
    "ORIGINAL REGULARIZED LOGISTIC MODEL"
)

print(
    "============================================================"
)

display(
    original_logistic_results.round(4)
)


# ------------------------------------------------
# 17. BOOTSTRAP DISTRIBUTION PLOTS
# ------------------------------------------------

# Plot the bootstrap distributions of the first few
# predictors to assess coefficient stability.

number_to_plot = min(
    6,
    n_features
)

for j in range(
    number_to_plot
):

    plt.figure(
        figsize=(8, 5)
    )

    plt.hist(
        bootstrap_coefficients_valid[
            :,
            j
        ],
        bins=35
    )

    plt.axvline(
        0,
        linestyle="--",
        linewidth=2
    )

    plt.axvline(
        original_coefficients[j],
        linestyle=":",
        linewidth=2,
        label="Original estimate"
    )

    plt.xlabel(
        "Bootstrap coefficient"
    )

    plt.ylabel(
        "Frequency"
    )

    plt.title(
        "Bootstrap Distribution: "
        + feature_names[j]
    )

    plt.legend()

    plt.tight_layout()

    plt.show()


# ------------------------------------------------
# 18. SAVE BOOTSTRAP OBJECTS FOR LATER SECTIONS
# ------------------------------------------------

# These objects are intentionally kept with the names that can
# be used by the subsequent prediction sections.

bootstrap_coefficients = (
    bootstrap_coefficients_valid
)

successful_bootstrap_models = (
    len(bootstrap_coefficients)
)


print(
    "\n============================================================"
)

print(
    "BOOTSTRAP LOGISTIC ANALYSIS COMPLETED SUCCESSFULLY"
)

print(
    "============================================================"
)

print(
    f"Valid bootstrap replications: "
    f"{successful_bootstrap_models}"
)

print(
    f"Fixed coefficient dimensions: "
    f"{bootstrap_coefficients.shape}"
)

print(
    "\nThe previous inhomogeneous-shape error has been eliminated."
)

print("""
THESIS INTERPRETATION NOTE:

The bootstrap confidence intervals quantify uncertainty under
resampling of the supplied 14 observations. They do not convert
the 14 observations into a nationally representative UDHS sample.

For the final thesis, the same procedure should be applied to the
complete appropriate UDHS analytical sample, incorporating the
survey sampling design, weights, clustering and stratification
where available.
""")

# ================================================================
# 13. OBJECTIVE 2
# STOCHASTIC TRANSITION MODEL
# ================================================================

print("\n============================================================")
print("OBJECTIVE 2: STOCHASTIC NUTRITIONAL STATE TRANSITION MODEL")
print("============================================================")

# Define nutritional states
states = [
    "Normal",
    "Stunted",
    "Wasted",
    "Underweight",
    "Overweight"
]


# Aggregate transition counts
transition_counts = (
    data
    .groupby(
        ["From_State", "To_State"]
    )["Count"]
    .sum()
    .reset_index()
)

print("\nObserved transition count inputs:")
display(transition_counts)


# Transition count matrix
count_matrix = pd.DataFrame(
    0,
    index=states,
    columns=states,
    dtype=float
)

for _, row in transition_counts.iterrows():

    source = row["From_State"]
    target = row["To_State"]
    count = row["Count"]

    if source in states and target in states:

        count_matrix.loc[
            source,
            target
        ] += count


print("\nTransition count matrix:")
display(count_matrix)


# ================================================================
# 14. ESTIMATE TRANSITION PROBABILITIES
# ================================================================

P = pd.DataFrame(
    0.0,
    index=states,
    columns=states
)

for source in states:

    total = count_matrix.loc[
        source
    ].sum()

    if total > 0:

        P.loc[source] = (
            count_matrix.loc[source] / total
        )

print(
    "\nESTIMATED transition probability matrix:"
)

display(P.round(4))


# ================================================================
# 15. HANDLE UNOBSERVED OUTGOING WASTED STATE
# ================================================================

# No transition originating from Wasted is supplied.
#
# We therefore DO NOT pretend to estimate a Wasted transition row.
#
# For simulation only, Wasted is treated as absorbing.
#
# This is an explicit simulation assumption and must be reported
# as such in the thesis.

P_sim = P.copy()

if P_sim.loc["Wasted"].sum() == 0:

    P_sim.loc["Wasted", "Wasted"] = 1.0

    print("""
Wasted has no observed outgoing transition counts.

For stochastic simulation only, Wasted has therefore been treated
as an absorbing state.

This is a simulation assumption, not an observed empirical result.
""")


print("\nTransition matrix used for simulation:")
display(P_sim.round(4))


# ================================================================
# 16. CHECK THAT MATRIX ROWS SUM TO ONE
# ================================================================

print("\nTransition matrix row sums:")
display(
    P_sim.sum(axis=1).round(4)
)


# ================================================================
# 17. BOOTSTRAP STOCHASTIC TRANSITION MATRIX
# ================================================================

print("\n============================================================")
print("BOOTSTRAP UNCERTAINTY OF TRANSITION PROBABILITIES")
print("============================================================")

bootstrap_transition_matrices = []

for b in range(B):

    P_b = pd.DataFrame(
        0.0,
        index=states,
        columns=states
    )

    for source in states:

        row_counts = count_matrix.loc[source].values

        total = int(row_counts.sum())

        if total == 0:
            continue

        # Multinomial bootstrap from the estimated transition
        # probabilities
        probabilities = (
            row_counts / total
        )

        simulated_counts = rng.multinomial(
            total,
            probabilities
        )

        P_b.loc[source] = (
            simulated_counts / total
        )

    # Wasted has no observed outgoing transitions
    P_b.loc["Wasted"] = 0

    P_b.loc[
        "Wasted",
        "Wasted"
    ] = 1.0

    bootstrap_transition_matrices.append(P_b)


# ================================================================
# 18. TRANSITION PROBABILITY UNCERTAINTY TABLE
# ================================================================

transition_uncertainty = []

for source in states:

    for target in states:

        values = np.array([
            P_b.loc[source, target]
            for P_b in bootstrap_transition_matrices
        ])

        transition_uncertainty.append({

            "From_State": source,

            "To_State": target,

            "Estimated_Probability":
                P_sim.loc[source, target],

            "Bootstrap_Mean":
                values.mean(),

            "Lower_95CI":
                np.percentile(values, 2.5),

            "Upper_95CI":
                np.percentile(values, 97.5)

        })


transition_uncertainty = pd.DataFrame(
    transition_uncertainty
)

print(
    "\nBootstrap uncertainty for stochastic transition probabilities:"
)

display(
    transition_uncertainty.round(4)
)


# ================================================================
# 19. OBJECTIVE 2
# SIMULATE FUTURE NUTRITIONAL STATES
# ================================================================

def simulate_markov(
    P,
    initial_state,
    n_steps=12,
    n_simulations=5000,
    seed=2026
):

    rng_local = np.random.default_rng(seed)

    state_to_index = {
        state: i
        for i, state in enumerate(states)
    }

    current = np.repeat(
        state_to_index[initial_state],
        n_simulations
    )

    simulated = np.zeros(
        (n_simulations, n_steps + 1),
        dtype=int
    )

    simulated[:, 0] = current

    for t in range(1, n_steps + 1):

        next_states = []

        for current_state in current:

            current_name = states[
                current_state
            ]

            probabilities = P.loc[
                current_name
            ].values

            next_state = rng_local.choice(
                len(states),
                p=probabilities
            )

            next_states.append(
                next_state
            )

        current = np.array(
            next_states
        )

        simulated[:, t] = current

    return simulated


# Simulate from each state
simulation_results = {}

for initial_state in states:

    simulation_results[
        initial_state
    ] = simulate_markov(
        P_sim,
        initial_state,
        n_steps=12,
        n_simulations=5000,
        seed=2026
    )


print(
    "\nStochastic simulations completed."
)

print("""
IMPORTANT:
These trajectories are SIMULATED model based trajectories.

They are NOT observed longitudinal trajectories of the same children.
""")


# ================================================================
# 20. SIMULATED STATE DISTRIBUTION
# ================================================================

def state_distribution(
    simulation_matrix,
    states
):

    records = []

    for t in range(
        simulation_matrix.shape[1]
    ):

        for j, state in enumerate(states):

            probability = np.mean(
                simulation_matrix[:, t] == j
            )

            records.append({

                "Time_Step": t,

                "State": state,

                "Probability": probability

            })

    return pd.DataFrame(records)


simulation_distribution = {}

for initial_state in states:

    simulation_distribution[
        initial_state
    ] = state_distribution(
        simulation_results[initial_state],
        states
    )


# Display simulation from Normal
print(
    "\nSimulated nutritional state distribution starting from Normal:"
)

display(
    simulation_distribution[
        "Normal"
    ].round(4)
)


# ================================================================
# 21. PLOT STOCHASTIC SIMULATION
# ================================================================

normal_sim = simulation_distribution[
    "Normal"
]

plt.figure(figsize=(11, 6))

for state in states:

    subset = normal_sim[
        normal_sim["State"] == state
    ]

    plt.plot(
        subset["Time_Step"],
        subset["Probability"],
        marker="o",
        label=state
    )

plt.xlabel("Simulation time step")
plt.ylabel("Simulated probability")
plt.title(
    "Simulated Nutritional State Probabilities Starting from Normal"
)

plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()


# ================================================================
# 22. STEADY STATE DISTRIBUTION
# ================================================================

print("\n============================================================")
print("MODEL BASED LONG RUN STOCHASTIC DISTRIBUTION")
print("============================================================")

P_matrix = P_sim.values

# Repeated matrix multiplication
P_power = np.linalg.matrix_power(
    P_matrix,
    100
)

steady_state = pd.DataFrame(
    P_power,
    index=states,
    columns=states
)

print("""
The following represents the model based long run transition
distribution under the estimated transition structure.

It is a SIMULATED / MODEL BASED result, not an observed national
malnutrition prevalence estimate.
""")

display(
    steady_state.round(4)
)


# ================================================================
# 23. OBJECTIVE 3
# PREDICTIVE PERFORMANCE
# ================================================================

print("\n============================================================")
print("OBJECTIVE 3: PREDICTIVE PERFORMANCE")
print("============================================================")

# Apparent predictive performance on the supplied data.
#
# Because n = 14, this should NOT be interpreted as external
# validation.

auc = roc_auc_score(
    y,
    pred_prob
)

brier = brier_score_loss(
    y,
    pred_prob
)

ll = log_loss(
    y,
    pred_prob,
    labels=[0,1]
)

accuracy = accuracy_score(
    y,
    pred_class
)

precision = precision_score(
    y,
    pred_class,
    zero_division=0
)

recall = recall_score(
    y,
    pred_class,
    zero_division=0
)

f1 = f1_score(
    y,
    pred_class,
    zero_division=0
)

tn, fp, fn, tp = confusion_matrix(
    y,
    pred_class,
    labels=[0,1]
).ravel()

specificity = (
    tn / (tn + fp)
    if (tn + fp) > 0
    else np.nan
)

predictive_results = pd.DataFrame({

    "Metric": [
        "AUC",
        "Brier Score",
        "Log Loss",
        "Accuracy",
        "Sensitivity",
        "Specificity",
        "Precision",
        "F1 Score"
    ],

    "Value": [
        auc,
        brier,
        ll,
        accuracy,
        recall,
        specificity,
        precision,
        f1
    ]

})

display(
    predictive_results.round(4)
)


# ================================================================
# 24. CAUTIOUS AUC INTERPRETATION
# ================================================================

print("\nInterpretation:")

if auc >= 0.999:

    print("""
The apparent AUC is approximately 1.000.

This should NOT be described as evidence of perfect generalisable
predictive performance.

Given the very small analytical sample and potential separation,
the result is more appropriately described as apparent perfect
discrimination within the supplied analytical sample.

Independent validation or repeated cross validation using the
complete appropriate survey sample is required before claiming
generalisable predictive accuracy.
""")

elif auc >= 0.80:

    print("""
The model demonstrates good apparent discrimination within the
supplied analytical sample. Because the sample contains only
14 observations, external validation is still required.
""")

elif auc >= 0.70:

    print("""
The model demonstrates moderate apparent discrimination.
Validation using the complete appropriate survey sample is
required before generalisation.
""")

else:

    print("""
The model demonstrates limited apparent discrimination within
the supplied analytical sample.
""")


# ================================================================
# 25. ROC CURVE
# ================================================================

fpr, tpr, thresholds = roc_curve(
    y,
    pred_prob
)

plt.figure(figsize=(7, 6))

plt.plot(
    fpr,
    tpr,
    linewidth=2,
    label=f"AUC = {auc:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="No discrimination"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "ROC Curve for Regularized Logistic Prediction"
)

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()


# ================================================================
# 26. CALIBRATION DATA
# ================================================================

calibration_table = pd.DataFrame({

    "Observed_Status": y.values,

    "Predicted_Probability": pred_prob,

    "Predicted_Class": pred_class

})

print(
    "\nObserved versus predicted probabilities:"
)

display(
    calibration_table.round(4)
)


# ================================================================
# 27. BOOTSTRAP PREDICTIVE PERFORMANCE
# ================================================================

print("\n============================================================")
print("BOOTSTRAP PREDICTIVE PERFORMANCE")
print("============================================================")

bootstrap_metrics = []

successful_prediction_bootstraps = 0

for b in range(B):

    boot_indices = rng.integers(
        0,
        len(data),
        len(data)
    )

    boot = data.iloc[
        boot_indices
    ].copy()

    yb = boot["Malnutrition"]

    # Need both classes for AUC
    if yb.nunique() < 2:
        continue

    try:

        model_b = Pipeline(
            steps=[
                (
                    "preprocessor",
                    preprocessor
                ),
                (
                    "model",
                    LogisticRegression(
                        penalty="l2",
                        C=1.0,
                        solver="liblinear",
                        max_iter=5000
                    )
                )
            ]
        )

        model_b.fit(
            boot[predictors],
            yb
        )

        pb = model_b.predict_proba(
            boot[predictors]
        )[:, 1]

        cb = (
            pb >= 0.5
        ).astype(int)

        auc_b = roc_auc_score(
            yb,
            pb
        )

        brier_b = brier_score_loss(
            yb,
            pb
        )

        logloss_b = log_loss(
            yb,
            pb,
            labels=[0,1]
        )

        f1_b = f1_score(
            yb,
            cb,
            zero_division=0
        )

        bootstrap_metrics.append({

            "AUC": auc_b,

            "Brier": brier_b,

            "Log_Loss": logloss_b,

            "F1": f1_b

        })

        successful_prediction_bootstraps += 1

    except Exception:
        continue


bootstrap_metrics = pd.DataFrame(
    bootstrap_metrics
)

print(
    "Successful predictive bootstrap samples:",
    successful_prediction_bootstraps
)


# ================================================================
# 28. BOOTSTRAP PERFORMANCE SUMMARY
# ================================================================

metric_summary = []

for metric in [
    "AUC",
    "Brier",
    "Log_Loss",
    "F1"
]:

    vals = bootstrap_metrics[
        metric
    ].dropna()

    metric_summary.append({

        "Metric": metric,

        "Mean": vals.mean(),

        "SD": vals.std(),

        "Lower_95CI":
            vals.quantile(0.025),

        "Upper_95CI":
            vals.quantile(0.975)

    })


metric_summary = pd.DataFrame(
    metric_summary
)

print(
    "\nBootstrap predictive performance summary:"
)

display(
    metric_summary.round(4)
)


# ================================================================
# 29. OBJECTIVE 4
# REGIONAL HETEROGENEITY
# ================================================================

print("\n============================================================")
print("OBJECTIVE 4: REGIONAL HETEROGENEITY")
print("============================================================")


# Observed regional status
regional_observed = (
    data
    .groupby("Region")
    .agg(
        Number_of_Children=(
            "Child_ID",
            "count"
        ),

        Malnutrition_Risk=(
            "Malnutrition",
            "mean"
        ),

        Mean_VitaminA=(
            "VitaminA_24h",
            "mean"
        ),

        Mean_Iron=(
            "Iron_24h",
            "mean"
        ),

        Mean_Rainfall=(
            "Rainfall_Index",
            "mean"
        ),

        Mean_Health_Access=(
            "Health_Access_Index",
            "mean"
        )
    )
    .reset_index()
)

regional_observed[
    "Malnutrition_Risk_Percent"
] = (
    regional_observed[
        "Malnutrition_Risk"
    ] * 100
)

print(
    "\nObserved empirical regional nutritional risk:"
)

display(
    regional_observed.round(3)
)


# ================================================================
# 30. MODEL BASED REGIONAL PREDICTION
# ================================================================

data["Predicted_Malnutrition_Risk"] = pred_prob

regional_predicted = (
    data
    .groupby("Region")
    .agg(
        Observed_Risk=(
            "Malnutrition",
            "mean"
        ),

        Posterior_or_Model_Predicted_Risk=(
            "Predicted_Malnutrition_Risk",
            "mean"
        )
    )
    .reset_index()
)

regional_predicted[
    "Observed_Risk_Percent"
] = (
    regional_predicted[
        "Observed_Risk"
    ] * 100
)

regional_predicted[
    "Predicted_Risk_Percent"
] = (
    regional_predicted[
        "Posterior_or_Model_Predicted_Risk"
    ] * 100
)


# Define elevated risk using the median predicted risk
risk_threshold = (
    regional_predicted[
        "Posterior_or_Model_Predicted_Risk"
    ].median()
)

regional_predicted[
    "Elevated_Predicted_Risk"
] = np.where(
    regional_predicted[
        "Posterior_or_Model_Predicted_Risk"
    ] >= risk_threshold,
    "Yes",
    "No"
)

print(
    "\nModel estimated regional nutritional risk:"
)

display(
    regional_predicted.round(3)
)


# ================================================================
# 31. REGIONAL RISK RANKING
# ================================================================

regional_ranking = (
    regional_predicted
    .sort_values(
        "Posterior_or_Model_Predicted_Risk",
        ascending=False
    )
    .reset_index(drop=True)
)

regional_ranking[
    "Risk_Rank"
] = np.arange(
    1,
    len(regional_ranking) + 1
)

print(
    "\nRegional ranking according to estimated nutritional risk:"
)

display(
    regional_ranking[
        [
            "Risk_Rank",
            "Region",
            "Predicted_Risk_Percent",
            "Elevated_Predicted_Risk"
        ]
    ].round(2)
)


# ================================================================
# 32. REGIONAL RISK PLOT
# ================================================================

plt.figure(figsize=(12, 7))

plot_data = regional_ranking.sort_values(
    "Predicted_Risk_Percent"
)

plt.barh(
    plot_data["Region"],
    plot_data["Predicted_Risk_Percent"]
)

plt.xlabel(
    "Model estimated malnutrition risk (%)"
)

plt.ylabel(
    "Region"
)

plt.title(
    "Estimated Regional Malnutrition Risk"
)

plt.tight_layout()
plt.show()


# ================================================================
# 33. BOOTSTRAP REGIONAL RISK
# ================================================================

print("\n============================================================")
print("BOOTSTRAP REGIONAL RISK UNCERTAINTY")
print("============================================================")

regional_bootstrap = []

for region in data["Region"].unique():

    region_values = data.loc[
        data["Region"] == region,
        "Malnutrition"
    ].values

    # Since each region has one supplied observation,
    # the bootstrap within-region CI has very limited meaning.
    #
    # We therefore report it explicitly but do not pretend it
    # provides reliable regional population inference.

    boot_values = rng.choice(
        region_values,
        size=(B, len(region_values)),
        replace=True
    ).mean(axis=1)

    regional_bootstrap.append({

        "Region": region,

        "Observed_Risk":
            region_values.mean(),

        "Bootstrap_Mean":
            boot_values.mean(),

        "Lower_95CI":
            np.percentile(
                boot_values,
                2.5
            ),

        "Upper_95CI":
            np.percentile(
                boot_values,
                97.5
            )

    })


regional_bootstrap = pd.DataFrame(
    regional_bootstrap
)

print("""
IMPORTANT:
Each region has only one supplied child level observation.
Consequently, regional bootstrap confidence intervals are not
informative population level confidence intervals.

The regional results should therefore be interpreted as
exploratory evidence of regional heterogeneity.
""")

display(
    regional_bootstrap.round(3)
)


# ================================================================
# 34. SPATIAL AUTOCORRELATION
# ================================================================

print("\n============================================================")
print("SPATIAL AUTOCORRELATION CHECK")
print("============================================================")

print("""
Moran's I cannot be validly calculated from region names alone.

The supplied dataset contains no geographic coordinates,
geographic polygons, adjacency matrix or verified spatial weights.

Therefore, this pipeline does NOT invent coordinates.

If valid coordinates or a spatial weights matrix are available,
the Moran's I section below can be activated.
""")


# ================================================================
# 35. OPTIONAL MORAN'S I FUNCTION
# ================================================================

def morans_i(values, weights):

    values = np.asarray(values, dtype=float)

    weights = np.asarray(weights, dtype=float)

    n = len(values)

    mean_value = np.mean(values)

    deviations = (
        values - mean_value
    )

    numerator = 0.0

    for i in range(n):

        for j in range(n):

            numerator += (
                weights[i, j]
                *
                deviations[i]
                *
                deviations[j]
            )

    denominator = np.sum(
        deviations ** 2
    )

    weight_sum = np.sum(weights)

    if denominator == 0 or weight_sum == 0:

        return np.nan

    I = (
        n / weight_sum
    ) * (
        numerator / denominator
    )

    return I


# ================================================================
# 36. OPTIONAL SPATIAL WEIGHTS INPUT
# ================================================================

# To perform Moran's I:
#
# 1. Prepare a CSV file called:
#       region_coordinates.csv
#
# 2. It should contain:
#
# Region,Latitude,Longitude
# Kampala,...
# Buganda,...
#
# 3. Upload the CSV into Colab.
#
# This code intentionally does not invent coordinates.

import os

if os.path.exists(
    "/content/region_coordinates.csv"
):

    coordinates = pd.read_csv(
        "/content/region_coordinates.csv"
    )

    print(
        "Coordinate file detected."
    )

    print(
        coordinates.head()
    )

    spatial_data = regional_predicted.merge(
        coordinates,
        on="Region",
        how="inner"
    )

    if len(spatial_data) >= 3:

        # Simple inverse distance weights
        # for exploratory Moran's I.
        #
        # For final thesis analysis, preferably use a
        # scientifically justified spatial weights matrix.

        coords = spatial_data[
            ["Latitude", "Longitude"]
        ].values

        n = len(coords)

        W = np.zeros(
            (n, n)
        )

        for i in range(n):

            for j in range(n):

                if i != j:

                    distance = np.sqrt(
                        (
                            coords[i, 0]
                            -
                            coords[j, 0]
                        ) ** 2
                        +
                        (
                            coords[i, 1]
                            -
                            coords[j, 1]
                        ) ** 2
                    )

                    if distance > 0:

                        W[i, j] = (
                            1 / distance
                        )

        # Row standardisation
        row_sums = W.sum(axis=1)

        for i in range(n):

            if row_sums[i] > 0:

                W[i, :] /= row_sums[i]

        spatial_values = (
            spatial_data[
                "Posterior_or_Model_Predicted_Risk"
            ].values
        )

        I = morans_i(
            spatial_values,
            W
        )

        print(
            "\nMoran's I:",
            round(I, 4)
        )

        print("""
Interpretation:
A positive Moran's I suggests that regions with similar
estimated nutritional risk tend to be spatially clustered.

A negative Moran's I suggests spatial dispersion.

However, Moran's I is evidence concerning SPATIAL
AUTOCORRELATION. It does NOT establish temporal dependence
or longitudinal nutritional progression.
""")

    else:

        print(
            "Insufficient regions with valid coordinates."
        )

else:

    print(
        "No coordinate file found. Moran's I not calculated."
    )


# ================================================================
# 37. OPTIONAL SPATIAL RISK MAP
# ================================================================

print("\n============================================================")
print("SPATIAL MAP REQUIREMENT")
print("============================================================")

print("""
A valid regional map requires a geographic boundary file whose
geographical units correspond to the regions actually analysed.

Do not create a map from region names alone.

Once an appropriate Uganda subregional shapefile or GeoJSON is
available, the estimated regional risk can be joined to it and
mapped.

The map title should distinguish the source:

Observed:
    Observed Regional Malnutrition

Estimated:
    Posterior Predicted Regional Malnutrition Risk

Simulated:
    Simulated Spatial Distribution of Nutritional Risk
""")


# ================================================================
# 38. MODEL DIAGNOSTIC SUMMARY
# ================================================================

print("\n============================================================")
print("MODEL DIAGNOSTIC SUMMARY")
print("============================================================")

diagnostic_summary = pd.DataFrame({

    "Component": [

        "Dataset size",

        "Bootstrap replications",

        "Logistic model",

        "Regularization",

        "Stochastic model",

        "Longitudinal observations",

        "Predictive validation",

        "Spatial coordinates"

    ],

    "Result": [

        f"{len(data)} observations",

        f"{B}",

        "Regularized logistic regression",

        "L2",

        "Model based stochastic transition simulation",

        "Not supplied",

        "Bootstrap apparent performance",

        "Not supplied"

    ]

})

display(
    diagnostic_summary
)


# ================================================================
# 39. EXPORT RESULTS
# ================================================================

print("\n============================================================")
print("EXPORTING RESULTS")
print("============================================================")

output_file = (
    "/content/"
    "Malnutrition_Stochastic_Bootstrap_Results.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    # Observed data
    data.to_excel(
        writer,
        sheet_name="Observed_Data",
        index=False
    )

    # Descriptive statistics
    continuous_summary.to_excel(
        writer,
        sheet_name="Descriptive_Continuous"
    )

    # Logistic results
    logistic_results.to_excel(
        writer,
        sheet_name="Logistic_Estimates",
        index=False
    )

    bootstrap_logistic_summary.to_excel(
        writer,
        sheet_name="Logistic_Bootstrap",
        index=False
    )

    # Transition results
    count_matrix.to_excel(
        writer,
        sheet_name="Transition_Counts"
    )

    P_sim.to_excel(
        writer,
        sheet_name="Transition_Probabilities"
    )

    transition_uncertainty.to_excel(
        writer,
        sheet_name="Transition_Bootstrap",
        index=False
    )

    # Predictive results
    predictive_results.to_excel(
        writer,
        sheet_name="Prediction_Metrics",
        index=False
    )

    metric_summary.to_excel(
        writer,
        sheet_name="Prediction_Bootstrap",
        index=False
    )

    # Regional results
    regional_observed.to_excel(
        writer,
        sheet_name="Regional_Observed",
        index=False
    )

    regional_predicted.to_excel(
        writer,
        sheet_name="Regional_Predicted",
        index=False
    )

    regional_ranking.to_excel(
        writer,
        sheet_name="Regional_Ranking",
        index=False
    )

    regional_bootstrap.to_excel(
        writer,
        sheet_name="Regional_Bootstrap",
        index=False
    )

    # Simulations
    simulation_distribution[
        "Normal"
    ].to_excel(
        writer,
        sheet_name="Simulation_Normal",
        index=False
    )


print(
    "\nResults successfully exported to:"
)

print(
    output_file
)


# ================================================================
# 40. FINAL AUTOMATED INTERPRETATION
# ================================================================

print("\n============================================================")
print("FINAL ANALYTICAL INTERPRETATION")
print("============================================================")

print(f"""
OBJECTIVE 1
Observed malnutrition prevalence in the supplied 14 row dataset
was {observed_prev*100:.2f}%.
The regularized logistic model was used instead of conventional
unpenalized maximum likelihood logistic regression because the
small sample can produce separation and unstable coefficients.

OBJECTIVE 2
The transition probability matrix was estimated from the supplied
transition counts and used to generate stochastic nutritional state
trajectories.

The simulated trajectories should NOT be described as observed
longitudinal progression. They represent model based probabilistic
scenarios.

OBJECTIVE 3
The apparent predictive AUC was {auc:.3f}.
Because the supplied dataset contains only 14 observations,
this result should be treated as apparent in sample performance.
Bootstrap uncertainty was estimated to show the instability
associated with the small sample.

Objective 3 should preferably be repeated using the complete
appropriate UDHS analytical sample.

OBJECTIVE 4
Regional heterogeneity was examined using observed and
model estimated nutritional risk.

A valid Moran's I analysis requires verified geographic coordinates
or a spatial weights matrix. The present dataset does not contain
this information, so spatial autocorrelation was not artificially
computed.

FINAL METHODOLOGICAL CONCLUSION

The bootstrap analysis improves uncertainty assessment but does
not substitute for reanalysis of the complete UDHS sample.

For the final PhD thesis, the strongest version would therefore
combine this pipeline with the actual UDHS child level dataset,
survey weights, clustering and stratification variables, and
verified geographic information.
""")


print("\n============================================================")
print("PIPELINE COMPLETED")
print("============================================================")
```
