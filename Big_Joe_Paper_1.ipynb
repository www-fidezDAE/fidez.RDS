```python
# ================================================================
# GOVERNMENT DEBT AND ECONOMIC GROWTH IN LIBERIA
# Evidence from Structural Breaks and Dynamic Time Series Analysis
#
# Period: 1994 to 2024
#
# Main objective:
# 1. Examine stationarity properties and structural breaks.
# 2. Estimate short run and long run effects of government debt.
# 3. Assess stability and robustness across structural changes.
#
# IMPORTANT:
# CAPITAL FORMATION HAS BEEN REMOVED FROM THE ANALYSIS.
#
# Main model:
# GDPG = f(DEBT, PE, TAX, LFP)
#
# GDPG = GDP Growth
# DEBT = Public Debt (% GDP)
# PE   = Public Expenditure (% GDP)
# TAX  = Tax Revenue (% GDP)
# LFP  = Labour Force Participation Rate
# ================================================================


# ================================================================
# 0. INSTALL / IMPORT REQUIRED PACKAGES
# ================================================================
# ============================================================
# LIBERIA GOVERNMENT DEBT AND ECONOMIC GROWTH STUDY
# Corrected Google Colab package installation and imports
# ============================================================

# Install or upgrade the required packages
!pip -q install --upgrade statsmodels scipy pandas numpy matplotlib seaborn openpyxl xlsxwriter ruptures

# ------------------------------------------------------------
# Restart recommendation
# ------------------------------------------------------------
# After running the installation cell above:
# Runtime > Restart session
#
# Then run the complete block below.
# ------------------------------------------------------------

# ============================================================
# CORE PYTHON PACKAGES
# ============================================================

import sys
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================
# STATISTICAL PACKAGES
# ============================================================

import scipy
from scipy import stats

import statsmodels
import statsmodels.api as sm

# ============================================================
# TIME SERIES TESTS
# ============================================================

from statsmodels.tsa.stattools import (
    adfuller,
    kpss,
    acf,
    pacf
)

# Phillips Perron test
# arch is used because statsmodels does not provide PP directly
try:
    from arch.unitroot import PhillipsPerron
except ImportError:
    !pip -q install arch
    from arch.unitroot import PhillipsPerron

# ============================================================
# REGRESSION DIAGNOSTICS
# ============================================================

# IMPORTANT:
# These functions are imported from statsmodels.stats.diagnostic
# and NOT from statsmodels.stats.api

from statsmodels.stats.diagnostic import (
    het_breuschpagan,
    het_white,
    acorr_breusch_godfrey,
    acorr_ljungbox,
    linear_reset,
    linear_rainbow
)

# ============================================================
# NORMALITY TEST
# ============================================================

from statsmodels.stats.stattools import (
    jarque_bera
)

# ============================================================
# OTHER STATISTICAL UTILITIES
# ============================================================

from statsmodels.stats.outliers_influence import (
    variance_inflation_factor
)

from statsmodels.stats.stattools import (
    durbin_watson
)

# ============================================================
# OLS REGRESSION
# ============================================================

from statsmodels.regression.linear_model import (
    OLS
)

# ============================================================
# ARDL
# ============================================================

try:
    from statsmodels.tsa.ardl import (
        ARDL,
        ardl_select_order
    )
except ImportError as e:
    print("ARDL import error:", e)

# ============================================================
# STRUCTURAL BREAK ANALYSIS
# ============================================================

import ruptures as rpt

# ============================================================
# ADDITIONAL UTILITIES
# ============================================================

from itertools import product
from pathlib import Path

# ============================================================
# DISPLAY PACKAGE VERSIONS
# ============================================================

print("=" * 70)
print("LIBERIA ECONOMIC GROWTH AND GOVERNMENT DEBT ANALYSIS")
print("=" * 70)

print(f"Python version       : {sys.version.split()[0]}")
print(f"NumPy version        : {np.__version__}")
print(f"Pandas version       : {pd.__version__}")
print(f"SciPy version        : {scipy.__version__}")
print(f"Statsmodels version  : {statsmodels.__version__}")

try:
    import arch
    print(f"ARCH version         : {arch.__version__}")
except:
    print("ARCH version         : Not available")

try:
    print(f"Ruptures version     : {rpt.__version__}")
except:
    print("Ruptures version     : Installed")

print("=" * 70)

# ============================================================
# TEST THAT THE PROBLEMATIC FUNCTION IS AVAILABLE
# ============================================================

print("\nTesting Breusch Pagan import...")

try:
    print("het_breuschpagan imported successfully.")
    print("Location: statsmodels.stats.diagnostic")
except Exception as e:
    print("Breusch Pagan import failed:", e)

# ============================================================
# TEST OTHER IMPORTANT FUNCTIONS
# ============================================================

print("\nTesting required statistical functions...")

functions_to_test = {
    "ADF": adfuller,
    "KPSS": kpss,
    "Breusch Pagan": het_breuschpagan,
    "White": het_white,
    "Breusch Godfrey": acorr_breusch_godfrey,
    "Ljung Box": acorr_ljungbox,
    "Jarque Bera": jarque_bera,
    "Durbin Watson": durbin_watson,
    "RESET": linear_reset,
    "ARDL": ARDL,
    "ARDL Select Order": ardl_select_order
}

for name, function in functions_to_test.items():
    try:
        print(f"✓ {name}")
    except Exception as e:
        print(f"✗ {name}: {e}")

print("\nAll core imports completed.")
print("=" * 70)


# ================================================================
# 1. CREATE THE LIBERIA DATASET
# ================================================================

data = {
    "Year": list(range(1994, 2025)),

    "GDPG": [
        -21.76, -4.26, 12.12, 106.27, 30.20, 21.74, 28.61,
        2.92, 3.76, -30.14, -2.61, 5.28, 8.04, 9.53, 7.14,
        5.30, 6.09, 8.20, 7.99, 8.68, 0.70, -0.01, -1.55,
        2.45, 1.15, -2.46, -2.98, 4.98, 4.80, 4.68, 4.02
    ],

    "PE": [
        15.43, 13.89, 14.68, 14.11, 15.45, 14.76, 14.54,
        12.06, 12.73, 8.98, 12.12, 11.68, 10.64, 16.05,
        21.78, 23.95, 25.06, 29.84, 30.75, 27.41, 33.44,
        37.12, 35.75, 35.07, 32.66, 32.31, 35.28, 29.78,
        26.93, 27.12, 24.30
    ],

    "DEBT": [
        469.8, 449.6, 454.4, 398.9, 465.8, 467.5, 468.1,
        465.2, 463.6, 600.1, 543.4, 488.5, 451.0, 366.8,
        235.3, 132.2, 25.4, 22.5, 20.5, 20.6, 24.3, 24.8,
        28.6, 31.8, 37.1, 48.6, 58.7, 53.3, 54.3, 57.8, 57.2
    ],

    "TAX": [
        14.87, 13.69, 12.98, 11.89, 6.78, 7.29, 7.74,
        10.22, 9.64, 12.13, 13.08, 13.90, 7.64, 7.23,
        10.20, 9.84, 10.45, 11.20, 11.14, 12.97, 12.46,
        15.18, 14.83, 14.96, 15.05, 14.36, 15.10, 14.29,
        13.55, 14.59, 13.99
    ],

    "LFP": [
        59.85, 55.69, 54.34, 59.43, 46.80, 49.65, 48.87,
        48.98, 59.73, 54.11, 54.32, 44.02, 49.81, 67.84,
        72.02, 53.12, 62.34, 57.55, 66.76, 65.43, 56.89,
        69.85, 76.93, 46.71, 55.98, 55.87, 54.78, 54.87,
        56.92, 55.99, 57.32
    ]
}

df = pd.DataFrame(data)

df["Year"] = pd.to_numeric(df["Year"])

for col in ["GDPG", "PE", "DEBT", "TAX", "LFP"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.sort_values("Year").reset_index(drop=True)

df.set_index("Year", inplace=True)

print("\nDataset successfully created.")
display(df.head())
display(df.tail())


# ================================================================
# 2. DATA VALIDATION
# ================================================================

print("\n" + "="*80)
print("DATA VALIDATION")
print("="*80)

print("\nNumber of observations:", len(df))

print("\nVariables:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate years:", df.index.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nBasic validation:")
assert len(df) == 31, "Expected 31 observations."
assert df.index.min() == 1994, "Expected start year 1994."
assert df.index.max() == 2024, "Expected end year 2024."
assert df.isna().sum().sum() == 0, "Missing observations detected."

print("\nValidation completed successfully.")

# Confirm CAP is NOT present
if "CAP" in df.columns:
    raise ValueError("CAP is still present. Remove CAP before continuing.")

print("\nCAPITAL FORMATION HAS BEEN SUCCESSFULLY REMOVED.")


# ================================================================
# 3. VARIABLE DICTIONARY
# ================================================================

variable_dictionary = pd.DataFrame({
    "Variable": ["GDPG", "DEBT", "PE", "TAX", "LFP"],
    "Description": [
        "Economic growth measured by annual GDP growth rate",
        "Public debt measured as percentage of GDP",
        "Public expenditure measured as percentage of GDP",
        "Tax revenue measured as percentage of GDP",
        "Labour force participation rate"
    ],
    "Role": [
        "Dependent variable",
        "Main explanatory variable",
        "Control variable",
        "Control variable",
        "Control variable"
    ],
    "Expected role": [
        "Economic growth",
        "Government debt effect",
        "Fiscal policy control",
        "Fiscal capacity control",
        "Labour market control"
    ]
})

display(variable_dictionary)


# ================================================================
# 4. DESCRIPTIVE STATISTICS
# ================================================================

print("\n" + "="*80)
print("DESCRIPTIVE STATISTICS")
print("="*80)

desc = df.describe().T

desc["Skewness"] = df.skew()
desc["Kurtosis"] = df.kurtosis()

display(desc.round(4))


# ================================================================
# 5. ADDITIONAL DESCRIPTIVE STATISTICS
# ================================================================

additional_desc = pd.DataFrame(index=df.columns)

additional_desc["Mean"] = df.mean()
additional_desc["Median"] = df.median()
additional_desc["Std. Dev."] = df.std()
additional_desc["Minimum"] = df.min()
additional_desc["Maximum"] = df.max()
additional_desc["Range"] = df.max() - df.min()
additional_desc["Skewness"] = df.skew()
additional_desc["Kurtosis"] = df.kurtosis()

display(additional_desc.round(4))


# ================================================================
# 6. IDENTIFY EXTREME OBSERVATIONS
# ================================================================

print("\n" + "="*80)
print("EXTREME OBSERVATION CHECK")
print("="*80)

for col in df.columns:

    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\n{col}")
    print("Lower threshold:", round(lower, 4))
    print("Upper threshold:", round(upper, 4))

    if len(outliers) > 0:
        print("Potential extreme observations:")
        display(outliers[[col]])
    else:
        print("No IQR based extreme observations.")


# ================================================================
# 7. TREND ANALYSIS
# ================================================================

plt.figure(figsize=(12, 6))

plt.plot(
    df.index,
    df["GDPG"],
    marker="o",
    linewidth=2,
    label="GDP Growth"
)

plt.axhline(0, linestyle="--", linewidth=1)

plt.title(
    "Economic Growth in Liberia, 1994 to 2024",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Year")
plt.ylabel("GDP Growth (%)")
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()


plt.figure(figsize=(12, 6))

plt.plot(
    df.index,
    df["DEBT"],
    marker="o",
    linewidth=2,
    label="Public Debt"
)

plt.title(
    "Public Debt in Liberia, 1994 to 2024",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Year")
plt.ylabel("Public Debt (% of GDP)")
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()


# ================================================================
# 8. MULTI SERIES TREND PLOTS
# ================================================================

fig, axes = plt.subplots(
    nrows=4,
    ncols=1,
    figsize=(13, 16),
    sharex=True
)

variables_for_plot = ["DEBT", "PE", "TAX", "LFP"]

titles = {
    "DEBT": "Public Debt (% of GDP)",
    "PE": "Public Expenditure (% of GDP)",
    "TAX": "Tax Revenue (% of GDP)",
    "LFP": "Labour Force Participation Rate"
}

for ax, var in zip(axes, variables_for_plot):

    ax.plot(
        df.index,
        df[var],
        marker="o",
        linewidth=1.8
    )

    ax.set_title(titles[var])
    ax.set_ylabel(var)
    ax.grid(alpha=0.25)

axes[-1].set_xlabel("Year")

plt.suptitle(
    "Major Macroeconomic Variables in Liberia, 1994 to 2024",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()


# ================================================================
# 9. CORRELATION MATRIX
# ================================================================

corr = df.corr()

print("\n" + "="*80)
print("CORRELATION MATRIX")
print("="*80)

display(corr.round(4))

plt.figure(figsize=(9, 7))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True
)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()


# ================================================================
# 10. MULTICOLLINEARITY CHECK
# ================================================================

print("\n" + "="*80)
print("VARIANCE INFLATION FACTOR")
print("="*80)

X_vif = df[["DEBT", "PE", "TAX", "LFP"]].copy()
X_vif = sm.add_constant(X_vif)

vif_table = pd.DataFrame()

vif_table["Variable"] = X_vif.columns

vif_table["VIF"] = [
    variance_inflation_factor(
        X_vif.values,
        i
    )
    for i in range(X_vif.shape[1])
]

display(vif_table.round(4))


# ================================================================
# 11. ADF UNIT ROOT TEST
# ================================================================

def adf_test(series, name):

    series = series.dropna()

    result = adfuller(
        series,
        regression="c",
        autolag="AIC"
    )

    output = {
        "Variable": name,
        "ADF Statistic": result[0],
        "p-value": result[1],
        "Used Lag": result[2],
        "Observations": result[3],
        "1% Critical": result[4]["1%"],
        "5% Critical": result[4]["5%"],
        "10% Critical": result[4]["10%"],
        "Decision": (
            "Stationary I(0)"
            if result[1] < 0.05
            else "Nonstationary at level"
        )
    }

    return output


adf_results = []

for col in df.columns:

    adf_results.append(
        adf_test(df[col], col)
    )

adf_table = pd.DataFrame(adf_results)

print("\n" + "="*80)
print("ADF TESTS AT LEVEL")
print("="*80)

display(adf_table.round(4))


# ================================================================
# 12. ADF TESTS AT FIRST DIFFERENCE
# ================================================================

adf_diff_results = []

for col in df.columns:

    diff_series = df[col].diff().dropna()

    result = adfuller(
        diff_series,
        regression="c",
        autolag="AIC"
    )

    adf_diff_results.append({
        "Variable": col,
        "ADF Statistic": result[0],
        "p-value": result[1],
        "Used Lag": result[2],
        "Observations": result[3],
        "Decision": (
            "Stationary I(1)"
            if result[1] < 0.05
            else "Still nonstationary"
        )
    })

adf_diff_table = pd.DataFrame(adf_diff_results)

print("\n" + "="*80)
print("ADF TESTS AT FIRST DIFFERENCE")
print("="*80)

display(adf_diff_table.round(4))


# ================================================================
# 13. PHILLIPS PERRON TEST
# ================================================================
#
# statsmodels does not consistently expose PP in every Colab version.
# We therefore install arch and use PhillipsPerron.
# ================================================================

!pip -q install arch

from arch.unitroot import PhillipsPerron


def pp_test(series, name):

    series = series.dropna()

    result = PhillipsPerron(
        series,
        trend="c"
    )

    return {
        "Variable": name,
        "PP Statistic": result.stat,
        "p-value": result.pvalue,
        "Lags": result.lags,
        "Decision": (
            "Stationary I(0)"
            if result.pvalue < 0.05
            else "Nonstationary at level"
        )
    }


pp_results = []

for col in df.columns:

    pp_results.append(
        pp_test(df[col], col)
    )

pp_table = pd.DataFrame(pp_results)

print("\n" + "="*80)
print("PHILLIPS PERRON TESTS AT LEVEL")
print("="*80)

display(pp_table.round(4))


# ================================================================
# 14. PHILLIPS PERRON AT FIRST DIFFERENCE
# ================================================================

pp_diff_results = []

for col in df.columns:

    diff_series = df[col].diff().dropna()

    result = PhillipsPerron(
        diff_series,
        trend="c"
    )

    pp_diff_results.append({
        "Variable": col,
        "PP Statistic": result.stat,
        "p-value": result.pvalue,
        "Lags": result.lags,
        "Decision": (
            "Stationary I(1)"
            if result.pvalue < 0.05
            else "Still nonstationary"
        )
    })

pp_diff_table = pd.DataFrame(pp_diff_results)

print("\n" + "="*80)
print("PHILLIPS PERRON TESTS AT FIRST DIFFERENCE")
print("="*80)

display(pp_diff_table.round(4))


# ================================================================
# 15. KPSS TEST
# ================================================================

def kpss_test(series, name):

    series = series.dropna()

    try:

        result = kpss(
            series,
            regression="c",
            nlags="auto"
        )

        return {
            "Variable": name,
            "KPSS Statistic": result[0],
            "p-value": result[1],
            "Lags": result[2],
            "5% Critical": result[3]["5%"],
            "Decision": (
                "Stationary"
                if result[1] > 0.05
                else "Nonstationary"
            )
        }

    except Exception as e:

        return {
            "Variable": name,
            "KPSS Statistic": np.nan,
            "p-value": np.nan,
            "Lags": np.nan,
            "5% Critical": np.nan,
            "Decision": "Test failed"
        }


kpss_results = []

for col in df.columns:

    kpss_results.append(
        kpss_test(df[col], col)
    )

kpss_table = pd.DataFrame(kpss_results)

print("\n" + "="*80)
print("KPSS TESTS AT LEVEL")
print("="*80)

display(kpss_table.round(4))


# ================================================================
# 16. KPSS AT FIRST DIFFERENCE
# ================================================================

kpss_diff_results = []

for col in df.columns:

    diff_series = df[col].diff().dropna()

    kpss_diff_results.append(
        kpss_test(diff_series, col)
    )

kpss_diff_table = pd.DataFrame(kpss_diff_results)

print("\n" + "="*80)
print("KPSS TESTS AT FIRST DIFFERENCE")
print("="*80)

display(kpss_diff_table.round(4))


# ================================================================
# 17. COMBINE UNIT ROOT RESULTS
# ================================================================

unit_root_summary = adf_table[
    ["Variable", "ADF Statistic", "p-value", "Decision"]
].copy()

unit_root_summary = unit_root_summary.merge(
    pp_table[
        ["Variable", "PP Statistic", "p-value", "Decision"]
    ],
    on="Variable",
    suffixes=("_ADF", "_PP")
)

unit_root_summary = unit_root_summary.merge(
    kpss_table[
        ["Variable", "KPSS Statistic", "p-value", "Decision"]
    ],
    on="Variable"
)

print("\n" + "="*80)
print("COMBINED UNIT ROOT SUMMARY")
print("="*80)

display(unit_root_summary.round(4))


# ================================================================
# 18. AUTOMATIC ARDL ADMISSIBILITY CHECK
# ================================================================
#
# ARDL is admissible only when no variable is I(2).
#
# We classify using ADF, PP and KPSS evidence.
# The final classification should also consider economic theory
# and structural break results.
# ================================================================

def determine_order(row):

    adf_p = row["p-value_ADF"]
    pp_p = row["p-value_PP"]

    if adf_p < 0.05 and pp_p < 0.05:
        return "Likely I(0)"

    return "Requires first-difference assessment"


orders = []

for col in df.columns:

    level_adf = adf_table.loc[
        adf_table["Variable"] == col,
        "p-value"
    ].iloc[0]

    level_pp = pp_table.loc[
        pp_table["Variable"] == col,
        "p-value"
    ].iloc[0]

    diff_adf = adf_diff_table.loc[
        adf_diff_table["Variable"] == col,
        "p-value"
    ].iloc[0]

    diff_pp = pp_diff_table.loc[
        pp_diff_table["Variable"] == col,
        "p-value"
    ].iloc[0]

    if level_adf < 0.05 and level_pp < 0.05:

        order = "I(0)"

    elif diff_adf < 0.05 and diff_pp < 0.05:

        order = "I(1)"

    else:

        order = "Uncertain"

    orders.append({
        "Variable": col,
        "ADF level p": level_adf,
        "PP level p": level_pp,
        "ADF difference p": diff_adf,
        "PP difference p": diff_pp,
        "Preliminary integration order": order
    })

integration_table = pd.DataFrame(orders)

print("\n" + "="*80)
print("PRELIMINARY INTEGRATION ORDER")
print("="*80)

display(integration_table.round(4))

if "I(2)" in integration_table[
    "Preliminary integration order"
].values:

    print(
        "\nWARNING: An I(2) variable has been detected."
    )

    print(
        "ARDL Bounds Testing should NOT be used."
    )

else:

    print(
        "\nNo I(2) variable has been automatically identified."
    )

    print(
        "ARDL remains potentially admissible."
    )


# ================================================================
# 19. ZIVOT ANDREWS STRUCTURAL BREAK UNIT ROOT TEST
# ================================================================

def za_test(series, name, regression="c"):

    series = series.dropna()

    try:

        result = zivot_andrews(
            series,
            trim=0.15,
            maxlag=2,
            regression=regression,
            autolag="AIC"
        )

        test_stat = result[0]
        pvalue = result[1]
        critical_values = result[2]
        used_lag = result[3]
        break_index = result[4]

        years = series.index.to_numpy()

        break_year = years[break_index]

        if pvalue < 0.05:
            decision = (
                "Reject unit root with one structural break"
            )
        else:
            decision = (
                "Fail to reject unit root"
            )

        return {
            "Variable": name,
            "ZA Statistic": test_stat,
            "p-value": pvalue,
            "Used Lag": used_lag,
            "Break Year": break_year,
            "1% Critical": critical_values["1%"],
            "5% Critical": critical_values["5%"],
            "10% Critical": critical_values["10%"],
            "Decision": decision
        }

    except Exception as e:

        return {
            "Variable": name,
            "ZA Statistic": np.nan,
            "p-value": np.nan,
            "Used Lag": np.nan,
            "Break Year": np.nan,
            "1% Critical": np.nan,
            "5% Critical": np.nan,
            "10% Critical": np.nan,
            "Decision": "Test failed"
        }


za_results = []

for col in df.columns:

    za_results.append(
        za_test(
            df[col],
            col,
            regression="c"
        )
    )

za_table = pd.DataFrame(za_results)

print("\n" + "="*80)
print("ZIVOT ANDREWS TEST")
print("="*80)

display(za_table.round(4))


# ================================================================
# 20. ZIVOT ANDREWS WITH CONSTANT AND TREND
# ================================================================

za_ct_results = []

for col in df.columns:

    za_ct_results.append(
        za_test(
            df[col],
            col,
            regression="ct"
        )
    )

za_ct_table = pd.DataFrame(za_ct_results)

print("\n" + "="*80)
print("ZIVOT ANDREWS TEST WITH CONSTANT AND TREND")
print("="*80)

display(za_ct_table.round(4))


# ================================================================
# 21. MULTIPLE STRUCTURAL BREAK DETECTION
# ================================================================
#
# The ruptures package implements change point detection.
# Here we use dynamic programming with a linear cost.
#
# This is reported as a Bai Perron type multiple break
# detection procedure, not as a substitute for every
# formal Bai Perron hypothesis test.
# ================================================================

def multiple_break_detection(
    series,
    name,
    max_breaks=3,
    min_size=4
):

    y = series.values.astype(float)

    # Linear model uses a time trend as regressor.
    t = np.arange(len(y)).reshape(-1, 1)

    signal = np.column_stack([
        y,
        t.flatten()
    ])

    detected = []

    try:

        model = rpt.Dynp(
            model="linear",
            min_size=min_size,
            jump=1
        )

        model.fit(signal)

        for n_breaks in range(1, max_breaks + 1):

            try:

                bkps = model.predict(
                    n_bkps=n_breaks
                )

                # Remove final observation marker
                break_indices = [
                    x for x in bkps
                    if x < len(series)
                ]

                break_years = [
                    int(series.index[x])
                    for x in break_indices
                ]

                detected.append({
                    "Variable": name,
                    "Number of Breaks": n_breaks,
                    "Break Years": break_years
                })

            except Exception:

                pass

    except Exception as e:

        detected.append({
            "Variable": name,
            "Number of Breaks": np.nan,
            "Break Years": f"Failed: {e}"
        })

    return detected


bp_results = []

for col in ["GDPG", "DEBT", "PE", "TAX", "LFP"]:

    bp_results.extend(
        multiple_break_detection(
            df[col],
            col,
            max_breaks=3,
            min_size=4
        )
    )

bp_table = pd.DataFrame(bp_results)

print("\n" + "="*80)
print("MULTIPLE STRUCTURAL BREAK DETECTION")
print("="*80)

display(bp_table)


# ================================================================
# 22. PLOT DETECTED BREAKS
# ================================================================

def plot_breaks(series, name):

    y = series.values.astype(float)

    t = np.arange(len(y)).reshape(-1, 1)

    signal = np.column_stack([
        y,
        t.flatten()
    ])

    try:

        model = rpt.Dynp(
            model="linear",
            min_size=4,
            jump=1
        )

        model.fit(signal)

        bkps = model.predict(
            n_bkps=2
        )

        break_indices = [
            x for x in bkps
            if x < len(series)
        ]

        plt.figure(figsize=(13, 5))

        plt.plot(
            series.index,
            series.values,
            marker="o",
            linewidth=1.8
        )

        for idx in break_indices:

            year = series.index[idx]

            plt.axvline(
                year,
                linestyle="--",
                linewidth=1.5
            )

            plt.text(
                year,
                np.nanmax(series.values),
                f"Break: {year}",
                rotation=90,
                verticalalignment="top"
            )

        plt.title(
            f"{name}: Multiple Structural Break Detection"
        )

        plt.xlabel("Year")
        plt.ylabel(name)
        plt.grid(alpha=0.25)
        plt.tight_layout()
        plt.show()

    except Exception as e:

        print(
            f"Break plot failed for {name}: {e}"
        )


for col in ["GDPG", "DEBT"]:

    plot_breaks(
        df[col],
        col
    )


# ================================================================
# 23. DEFINE THE MAIN MODEL
# ================================================================

Y = df["GDPG"]

X = df[
    [
        "DEBT",
        "PE",
        "TAX",
        "LFP"
    ]
]

print("\n" + "="*80)
print("MAIN MODEL")
print("="*80)

print(
    "GDPG = f(DEBT, PE, TAX, LFP)"
)

print(
    "\nDependent variable: GDPG"
)

print(
    "Main explanatory variable: DEBT"
)

print(
    "Controls: PE, TAX, LFP"
)


# ================================================================
# 24. ARDL LAG SELECTION
# ================================================================
#
# Because the sample contains only 31 observations,
# maximum lags are restricted.
#
# Maximum lag for GDP growth = 2
# Maximum lag for explanatory variables = 1
# ================================================================

print("\n" + "="*80)
print("ARDL LAG SELECTION")
print("="*80)

try:

    selected = ardl_select_order(
        endog=Y,
        maxlag=2,
        exog=X,
        maxorder=1,
        trend="c",
        fixed=None,
        causal=False,
        ic="aic",
        glob=False
    )

    print(
        "\nUnrestricted AIC selected ARDL order:"
    )

    print(
        selected.model.ardl_order
    )

    print(
        "\nAIC:",
        selected.model.fit().aic
    )

    print(
        "BIC:",
        selected.model.fit().bic
    )

except Exception as e:

    selected = None

    print(
        "\nAutomatic ARDL selection failed:"
    )

    print(e)


# ================================================================
# 25. MANUAL PARSIMONIOUS ARDL SEARCH
# ================================================================
#
# We also search manually over small models.
# This is important because the sample is small.
# ================================================================

candidate_models = []

for p in [1, 2]:

    for q_debt in [1]:

        for q_pe in [1]:

            for q_tax in [1]:

                for q_lfp in [1]:

                    try:

                        model = ARDL(
                            endog=Y,
                            lags=p,
                            exog=X,
                            order={
                                "DEBT": q_debt,
                                "PE": q_pe,
                                "TAX": q_tax,
                                "LFP": q_lfp
                            },
                            trend="c"
                        )

                        result = model.fit()

                        candidate_models.append({
                            "p": p,
                            "q_DEBT": q_debt,
                            "q_PE": q_pe,
                            "q_TAX": q_tax,
                            "q_LFP": q_lfp,
                            "AIC": result.aic,
                            "BIC": result.bic,
                            "HQIC": result.hqic
                        })

                    except Exception as e:

                        pass


lag_table = pd.DataFrame(candidate_models)

if len(lag_table) > 0:

    lag_table = lag_table.sort_values(
        "AIC"
    ).reset_index(drop=True)

    print(
        "\nParsimonious ARDL candidates ranked by AIC:"
    )

    display(
        lag_table.round(4)
    )

    best_p = int(lag_table.loc[0, "p"])
    best_q_debt = int(lag_table.loc[0, "q_DEBT"])
    best_q_pe = int(lag_table.loc[0, "q_PE"])
    best_q_tax = int(lag_table.loc[0, "q_TAX"])
    best_q_lfp = int(lag_table.loc[0, "q_LFP"])

else:

    raise RuntimeError(
        "No valid ARDL model could be estimated."
    )


# ================================================================
# 26. ESTIMATE SELECTED ARDL
# ================================================================

best_order = {
    "DEBT": best_q_debt,
    "PE": best_q_pe,
    "TAX": best_q_tax,
    "LFP": best_q_lfp
}

best_ardl = ARDL(
    endog=Y,
    lags=best_p,
    exog=X,
    order=best_order,
    trend="c"
)

best_ardl_res = best_ardl.fit()

print("\n" + "="*80)
print("SELECTED ARDL MODEL")
print("="*80)

print(
    f"ARDL({best_p}, "
    f"{best_q_debt}, "
    f"{best_q_pe}, "
    f"{best_q_tax}, "
    f"{best_q_lfp})"
)

print(
    "\nAIC:",
    best_ardl_res.aic
)

print(
    "BIC:",
    best_ardl_res.bic
)

print(
    "HQIC:",
    best_ardl_res.hqic
)

print(
    "\nModel summary:"
)

print(
    best_ardl_res.summary()
)


# ================================================================
# 27. CONVERT ARDL TO UECM
# ================================================================

try:

    uecm_model = UECM.from_ardl(
        best_ardl
    )

    uecm_res = uecm_model.fit()

    print("\n" + "="*80)
    print("UNRESTRICTED ERROR CORRECTION MODEL")
    print("="*80)

    print(
        uecm_res.summary()
    )

except Exception as e:

    print(
        "\nUECM conversion failed."
    )

    print(e)

    uecm_res = None


# ================================================================
# 28. ARDL BOUNDS TEST
# ================================================================
#
# Case 3:
# Unrestricted intercept and no deterministic trend.
#
# This is the conventional specification for a model where
# the level relationship contains an intercept but no trend.
#
# We use finite sample simulation because T is only 31.
# ================================================================

bounds_result = None

if uecm_res is not None:

    try:

        bounds_result = uecm_res.bounds_test(
            case=3,
            cov_type="nonrobust",
            use_t=True,
            asymptotic=False,
            nsim=20000,
            seed=12345
        )

        print("\n" + "="*80)
        print("PESARAN, SHIN AND SMITH ARDL BOUNDS TEST")
        print("="*80)

        print(
            bounds_result
        )

        print(
            "\nF statistic:",
            bounds_result.stat
        )

        print(
            "\nCritical values:"
        )

        display(
            bounds_result.crit_vals
        )

        print(
            "\nP values:"
        )

        print(
            bounds_result.p_values
        )

    except Exception as e:

        print(
            "\nBounds test failed:"
        )

        print(e)


# ================================================================
# 29. AUTOMATIC COINTEGRATION DECISION
# ================================================================

cointegration_decision = "Not established"

if bounds_result is not None:

    try:

        Fstat = bounds_result.stat

        cv = bounds_result.crit_vals

        # Extract 5 percent bounds
        lower_5 = cv.loc[5.0, "lower"]
        upper_5 = cv.loc[5.0, "upper"]

        print("\n" + "="*80)
        print("COINTEGRATION DECISION AT 5 PERCENT")
        print("="*80)

        print(
            "F statistic:",
            round(Fstat, 4)
        )

        print(
            "Lower bound:",
            round(lower_5, 4)
        )

        print(
            "Upper bound:",
            round(upper_5, 4)
        )

        if Fstat > upper_5:

            cointegration_decision = (
                "Cointegration established"
            )

            print(
                "\nDECISION: Cointegration is supported."
            )

        elif Fstat < lower_5:

            cointegration_decision = (
                "No cointegration"
            )

            print(
                "\nDECISION: No cointegration is supported."
            )

        else:

            cointegration_decision = (
                "Inconclusive"
            )

            print(
                "\nDECISION: Bounds test is inconclusive."
            )

    except Exception as e:

        print(
            "\nCould not automatically extract critical values."
        )

        print(e)


# ================================================================
# 30. LONG RUN ESTIMATION
# ================================================================
#
# Only performed if cointegration is established.
# ================================================================

long_run_table = None

if (
    uecm_res is not None
    and cointegration_decision == "Cointegration established"
):

    try:

        long_run = uecm_res.ci_summary()

        print("\n" + "="*80)
        print("LONG RUN COINTEGRATING RELATIONSHIP")
        print("="*80)

        print(long_run)

        # Attempt to obtain long run coefficients
        try:

            long_run_params = uecm_res.ci_params

            long_run_table = pd.DataFrame({
                "Coefficient": long_run_params
            })

            display(
                long_run_table.round(6)
            )

        except Exception:

            pass

    except Exception as e:

        print(
            "\nLong run estimation failed:"
        )

        print(e)

else:

    print(
        "\nLong run estimation was NOT automatically performed."
    )

    print(
        "Reason:",
        cointegration_decision
    )


# ================================================================
# 31. ERROR CORRECTION MODEL
# ================================================================

ecm_result = None

if (
    uecm_res is not None
    and cointegration_decision == "Cointegration established"
):

    ecm_result = uecm_res

    print("\n" + "="*80)
    print("SHORT RUN ERROR CORRECTION MODEL")
    print("="*80)

    print(
        ecm_result.summary()
    )

else:

    print(
        "\nECM not estimated because cointegration was not established."
    )


# ================================================================
# 32. EXTRACT ERROR CORRECTION COEFFICIENT
# ================================================================

if ecm_result is not None:

    params = ecm_result.params

    ecm_candidates = [
        name
        for name in params.index
        if "L1.GDPG" in name
        or "GDPG.L1" in name
        or "GDPG.L1" in str(name)
    ]

    print(
        "\nPotential ECM coefficients:"
    )

    for name in ecm_candidates:

        print(
            name,
            "=",
            params[name]
        )

    print(
        "\nInterpretation requirement:"
    )

    print(
        "The coefficient on lagged GDP growth in the UECM level "
        "representation should generally be negative and statistically "
        "significant if a stable error correction mechanism is present."
    )


# ================================================================
# 33. RESIDUALS
# ================================================================

if best_ardl_res is not None:

    residuals = pd.Series(
        best_ardl_res.resid,
        index=best_ardl_res.model.data.row_labels
    )

    residuals = residuals.dropna()

    fitted = pd.Series(
        best_ardl_res.fittedvalues,
        index=best_ardl_res.model.data.row_labels
    )

    fitted = fitted.dropna()

    print("\nResidual observations:", len(residuals))


# ================================================================
# 34. RESIDUAL PLOT
# ================================================================

plt.figure(figsize=(13, 5))

plt.plot(
    residuals.index,
    residuals.values,
    marker="o",
    linewidth=1.5
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "ARDL Model Residuals"
)

plt.xlabel("Year")
plt.ylabel("Residual")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()


# ================================================================
# 35. RESIDUAL HISTOGRAM
# ================================================================

plt.figure(figsize=(9, 5))

plt.hist(
    residuals,
    bins=8,
    edgecolor="black"
)

plt.title(
    "Distribution of ARDL Residuals"
)

plt.xlabel("Residual")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# ================================================================
# 36. SERIAL CORRELATION TEST
# ================================================================

print("\n" + "="*80)
print("SERIAL CORRELATION DIAGNOSTICS")
print("="*80)

try:

    bg_test = acorr_breusch_godfrey(
        best_ardl_res,
        nlags=1
    )

    bg_table = pd.DataFrame({
        "Statistic": [
            "LM Statistic",
            "LM p-value",
            "F Statistic",
            "F p-value"
        ],
        "Value": [
            bg_test[0],
            bg_test[1],
            bg_test[2],
            bg_test[3]
        ]
    })

    display(
        bg_table.round(6)
    )

except Exception as e:

    print(
        "Breusch Godfrey test failed:",
        e
    )


# ================================================================
# 37. LJUNG BOX TEST
# ================================================================

try:

    lb = acorr_ljungbox(
        residuals,
        lags=[1, 2],
        return_df=True
    )

    print(
        "\nLjung Box test:"
    )

    display(
        lb.round(6)
    )

except Exception as e:

    print(
        "Ljung Box test failed:",
        e
    )


# ================================================================
# 38. HETEROSKEDASTICITY
# ================================================================

print("\n" + "="*80)
print("HETEROSKEDASTICITY TESTS")
print("="*80)

try:

    exog_for_diag = best_ardl_res.model._x

    bp_test = het_breusch_pagan(
        residuals,
        exog_for_diag
    )

    bp_table = pd.DataFrame({
        "Statistic": [
            "LM Statistic",
            "LM p-value",
            "F Statistic",
            "F p-value"
        ],
        "Value": [
            bp_test[0],
            bp_test[1],
            bp_test[2],
            bp_test[3]
        ]
    })

    display(
        bp_table.round(6)
    )

except Exception as e:

    print(
        "Breusch Pagan test failed:",
        e
    )


# ================================================================
# 39. WHITE TEST
# ================================================================

try:

    white_test = het_white(
        residuals,
        exog_for_diag
    )

    white_table = pd.DataFrame({
        "Statistic": [
            "LM Statistic",
            "LM p-value",
            "F Statistic",
            "F p-value"
        ],
        "Value": [
            white_test[0],
            white_test[1],
            white_test[2],
            white_test[3]
        ]
    })

    display(
        white_table.round(6)
    )

except Exception as e:

    print(
        "White test failed:",
        e
    )


# ================================================================
# 40. NORMALITY TEST
# ================================================================

print("\n" + "="*80)
print("NORMALITY TEST")
print("="*80)

try:

    jb_stat, jb_pvalue, skew, kurtosis = jarque_bera(
        residuals
    )

    normality_table = pd.DataFrame({
        "Statistic": [
            "Jarque Bera",
            "p-value",
            "Skewness",
            "Kurtosis"
        ],
        "Value": [
            jb_stat,
            jb_pvalue,
            skew,
            kurtosis
        ]
    })

    display(
        normality_table.round(6)
    )

except Exception as e:

    print(
        "Jarque Bera test failed:",
        e
    )


# ================================================================
# 41. ANDERSON DARLING NORMALITY TEST
# ================================================================

try:

    ad_stat, ad_pvalue = normal_ad(
        residuals
    )

    print(
        "\nAnderson Darling statistic:",
        ad_stat
    )

    print(
        "p-value:",
        ad_pvalue
    )

except Exception as e:

    print(
        "Anderson Darling test failed:",
        e
    )


# ================================================================
# 42. RAMSEY RESET FUNCTIONAL FORM TEST
# ================================================================

print("\n" + "="*80)
print("RAMSEY RESET FUNCTIONAL FORM TEST")
print("="*80)

try:

    # Fit a simple OLS representation using fitted values
    reset_model = sm.OLS(
        Y.loc[fitted.index],
        sm.add_constant(
            df.loc[fitted.index, ["DEBT", "PE", "TAX", "LFP"]]
        )
    ).fit()

    reset_test = linear_reset(
        reset_model,
        power=2,
        use_f=True
    )

    print(
        reset_test
    )

except Exception as e:

    print(
        "RESET test failed:",
        e
    )


# ================================================================
# 43. CUSUM STABILITY TEST
# ================================================================
#
# Conventional CUSUM test based on recursive residuals.
# ================================================================

print("\n" + "="*80)
print("CUSUM STABILITY TEST")
print("="*80)

try:

    # OLS approximation of the static long run relationship
    X_static = sm.add_constant(
        df[
            ["DEBT", "PE", "TAX", "LFP"]
        ]
    )

    y_static = df["GDPG"]

    ols_static = sm.OLS(
        y_static,
        X_static
    ).fit()

    cusum_result = breaks_cusumolsresid(
        ols_static.resid,
        ddof=int(ols_static.df_model)
    )

    print(
        "CUSUM statistic:",
        cusum_result[0]
    )

    print(
        "p-value:",
        cusum_result[1]
    )

    print(
        "Critical values:"
    )

    print(
        cusum_result[2]
    )

except Exception as e:

    print(
        "CUSUM test failed:",
        e
    )


# ================================================================
# 44. CUSUM PLOT
# ================================================================

try:

    recursive_model = AutoReg(
        df["GDPG"],
        lags=1,
        trend="c"
    ).fit()

    rec_resid = recursive_model.resid.dropna()

    standardized = (
        rec_resid - rec_resid.mean()
    ) / rec_resid.std()

    cusum_series = standardized.cumsum()

    plt.figure(figsize=(13, 5))

    plt.plot(
        cusum_series.index,
        cusum_series.values,
        linewidth=2,
        label="CUSUM"
    )

    plt.axhline(
        0,
        linestyle="--",
        linewidth=1
    )

    plt.title(
        "CUSUM Stability Plot"
    )

    plt.xlabel("Year")
    plt.ylabel("Cumulative Recursive Residual")
    plt.grid(alpha=0.25)
    plt.legend()
    plt.tight_layout()
    plt.show()

except Exception as e:

    print(
        "CUSUM plot failed:",
        e
    )


# ================================================================
# 45. CUSUMSQ
# ================================================================
#
# CUSUMSQ is constructed from recursive residuals.
# The cumulative proportion of squared recursive residuals
# is plotted against time.
#
# This is a diagnostic stability plot. For a formal publication,
# the result should be interpreted together with CUSUM and
# parameter stability evidence.
# ================================================================

try:

    recursive_model = AutoReg(
        df["GDPG"],
        lags=1,
        trend="c"
    ).fit()

    rr = recursive_model.resid.dropna()

    squared_rr = rr ** 2

    cusumsq = (
        squared_rr.cumsum()
        /
        squared_rr.sum()
    )

    plt.figure(figsize=(13, 5))

    plt.plot(
        cusumsq.index,
        cusumsq.values,
        linewidth=2,
        label="CUSUMSQ"
    )

    plt.axhline(
        0.5,
        linestyle="--",
        linewidth=1,
        label="Reference 0.5"
    )

    plt.title(
        "CUSUM of Squares Stability Plot"
    )

    plt.xlabel("Year")
    plt.ylabel("Cumulative Sum of Squared Recursive Residuals")
    plt.grid(alpha=0.25)
    plt.legend()
    plt.tight_layout()
    plt.show()

except Exception as e:

    print(
        "CUSUMSQ failed:",
        e
    )


# ================================================================
# 46. 1997 SENSITIVITY ANALYSIS
# ================================================================
#
# The 1997 GDP growth value is 106.27%.
#
# It is NOT deleted automatically.
#
# Instead, the complete analysis is repeated without 1997
# to assess whether the debt coefficient and long run relationship
# depend excessively on that observation.
# ================================================================

print("\n" + "="*80)
print("1997 SENSITIVITY ANALYSIS")
print("="*80)

df_no1997 = df.drop(
    index=1997
)

Y_no1997 = df_no1997["GDPG"]

X_no1997 = df_no1997[
    ["DEBT", "PE", "TAX", "LFP"]
]


# ================================================================
# 47. SELECT ARDL WITHOUT 1997
# ================================================================

sensitivity_results = {}

try:

    candidates_no1997 = []

    for p in [1, 2]:

        try:

            model_no1997 = ARDL(
                endog=Y_no1997,
                lags=p,
                exog=X_no1997,
                order={
                    "DEBT": 1,
                    "PE": 1,
                    "TAX": 1,
                    "LFP": 1
                },
                trend="c"
            )

            result_no1997 = model_no1997.fit()

            candidates_no1997.append(
                (
                    result_no1997.aic,
                    p,
                    result_no1997
                )
            )

        except Exception:

            pass

    candidates_no1997.sort(
        key=lambda x: x[0]
    )

    if len(candidates_no1997) > 0:

        _, p_no1997, result_no1997 = candidates_no1997[0]

        sensitivity_results[
            "1997_excluded"
        ] = result_no1997

        print(
            "Selected ARDL without 1997:",
            p_no1997
        )

        print(
            result_no1997.summary()
        )

except Exception as e:

    print(
        "1997 sensitivity analysis failed:",
        e
    )


# ================================================================
# 48. COMPARE MAIN AND 1997 EXCLUDED MODELS
# ================================================================

def extract_key_coefficients(
    result,
    label
):

    output = []

    for name in result.params.index:

        if (
            "DEBT" in str(name)
            or "PE" in str(name)
            or "TAX" in str(name)
            or "LFP" in str(name)
        ):

            output.append({
                "Model": label,
                "Parameter": name,
                "Coefficient": result.params[name],
                "Std Error": result.bse[name],
                "p-value": result.pvalues[name]
            })

    return pd.DataFrame(output)


main_coef_table = extract_key_coefficients(
    best_ardl_res,
    "Full sample"
)

display(main_coef_table.round(6))


if "1997_excluded" in sensitivity_results:

    sensitivity_coef_table = extract_key_coefficients(
        sensitivity_results["1997_excluded"],
        "1997 excluded"
    )

    display(
        sensitivity_coef_table.round(6)
    )


# ================================================================
# 49. ALTERNATIVE LAG STRUCTURE
# ================================================================

print("\n" + "="*80)
print("ALTERNATIVE LAG STRUCTURE ROBUSTNESS")
print("="*80)

alternative_lag_models = {}

for p in [1, 2]:

    try:

        alt_model = ARDL(
            endog=Y,
            lags=p,
            exog=X,
            order={
                "DEBT": 1,
                "PE": 1,
                "TAX": 1,
                "LFP": 1
            },
            trend="c"
        )

        alt_result = alt_model.fit()

        alternative_lag_models[
            f"ARDL({p},1,1,1,1)"
        ] = alt_result

        print(
            f"\nARDL({p},1,1,1,1)"
        )

        print(
            "AIC:",
            alt_result.aic
        )

        print(
            "BIC:",
            alt_result.bic
        )

    except Exception as e:

        print(
            f"Alternative ARDL({p}) failed:",
            e
        )


# ================================================================
# 50. ALTERNATIVE CONTROL SPECIFICATION
# ================================================================
#
# Specification A:
# GDPG = f(DEBT, PE, TAX, LFP)
#
# Specification B:
# GDPG = f(DEBT, PE, TAX)
#
# Specification C:
# GDPG = f(DEBT, PE, LFP)
#
# CAP IS NOT USED.
# ================================================================

alternative_controls = {}

specifications = {

    "Full controls": [
        "DEBT",
        "PE",
        "TAX",
        "LFP"
    ],

    "Fiscal controls only": [
        "DEBT",
        "PE",
        "TAX"
    ],

    "Debt expenditure labour controls": [
        "DEBT",
        "PE",
        "LFP"
    ]
}


for spec_name, variables in specifications.items():

    try:

        y_spec = df["GDPG"]

        x_spec = df[variables]

        orders_spec = {
            var: 1
            for var in variables
        }

        spec_model = ARDL(
            endog=y_spec,
            lags=1,
            exog=x_spec,
            order=orders_spec,
            trend="c"
        )

        spec_result = spec_model.fit()

        alternative_controls[
            spec_name
        ] = spec_result

        print(
            f"\n{spec_name}"
        )

        print(
            "Variables:",
            variables
        )

        print(
            "AIC:",
            spec_result.aic
        )

        print(
            "BIC:",
            spec_result.bic
        )

    except Exception as e:

        print(
            f"{spec_name} failed:",
            e
        )


# ================================================================
# 51. ROBUSTNESS COEFFICIENT SUMMARY
# ================================================================

robustness_rows = []

for model_name, result in alternative_controls.items():

    for parameter in result.params.index:

        if "DEBT" in str(parameter):

            robustness_rows.append({

                "Model": model_name,

                "Parameter": parameter,

                "Coefficient": result.params[
                    parameter
                ],

                "Std Error": result.bse[
                    parameter
                ],

                "p-value": result.pvalues[
                    parameter
                ],

                "AIC": result.aic,

                "BIC": result.bic
            })


if len(robustness_rows) > 0:

    robustness_table = pd.DataFrame(
        robustness_rows
    )

    print(
        "\n" + "="*80
    )

    print(
        "ROBUSTNESS OF PUBLIC DEBT COEFFICIENT"
    )

    print(
        "="*80
    )

    display(
        robustness_table.round(6)
    )


# ================================================================
# 52. LONG RUN MULTIPLIER FROM ARDL
# ================================================================
#
# For an ARDL model:
#
# y_t = c + sum(alpha_i y_{t-i})
#       + sum(beta_j x_{t-j}) + e_t
#
# The long run coefficient for x is:
#
# sum(beta_j) / (1 - sum(alpha_i))
#
# ================================================================

def calculate_long_run_coefficients(
    result,
    exog_names
):

    params = result.params

    # Find dependent variable name
    endog_name = result.model.endog_names

    ar_params = []

    for name in params.index:

        if (
            str(name).startswith(
                f"{endog_name}.L"
            )
        ):

            ar_params.append(
                params[name]
            )

    denominator = 1 - np.sum(
        ar_params
    )

    rows = []

    for var in exog_names:

        numerator = 0

        for name in params.index:

            if str(name).startswith(
                f"{var}.L"
            ):

                numerator += params[name]

        lr = numerator / denominator

        rows.append({
            "Variable": var,
            "Long Run Coefficient": lr
        })

    return pd.DataFrame(rows)


try:

    long_run_manual = calculate_long_run_coefficients(
        best_ardl_res,
        ["DEBT", "PE", "TAX", "LFP"]
    )

    print("\n" + "="*80)
    print("LONG RUN MULTIPLIERS FROM ARDL")
    print("="*80)

    display(
        long_run_manual.round(6)
    )

except Exception as e:

    print(
        "Long run multiplier calculation failed:",
        e
    )


# ================================================================
# 53. DEBT EFFECT VISUALIZATION
# ================================================================

plt.figure(figsize=(12, 6))

plt.scatter(
    df["DEBT"],
    df["GDPG"],
    s=60,
    alpha=0.8
)

z = np.polyfit(
    df["DEBT"],
    df["GDPG"],
    1
)

p = np.poly1d(z)

x_line = np.linspace(
    df["DEBT"].min(),
    df["DEBT"].max(),
    100
)

plt.plot(
    x_line,
    p(x_line),
    linewidth=2
)

plt.xlabel(
    "Public Debt (% of GDP)"
)

plt.ylabel(
    "GDP Growth (%)"
)

plt.title(
    "Public Debt and Economic Growth in Liberia"
)

plt.grid(alpha=0.25)

plt.tight_layout()

plt.show()


# ================================================================
# 54. HIGHLIGHT 1997
# ================================================================

plt.figure(figsize=(13, 6))

plt.plot(
    df.index,
    df["GDPG"],
    marker="o",
    linewidth=1.8
)

plt.scatter(
    [1997],
    [df.loc[1997, "GDPG"]],
    s=120,
    marker="o"
)

plt.annotate(
    "1997 = 106.27%",
    xy=(
        1997,
        df.loc[1997, "GDPG"]
    ),
    xytext=(
        1999,
        80
    ),
    arrowprops=dict(
        arrowstyle="->"
    )
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "GDP Growth and the 1997 Extreme Observation"
)

plt.xlabel("Year")
plt.ylabel("GDP Growth (%)")

plt.grid(alpha=0.25)

plt.tight_layout()

plt.show()


# ================================================================
# 55. SAVE ALL MAIN RESULTS
# ================================================================

output_folder = "Liberia_Debt_Growth_Analysis"

os.makedirs(
    output_folder,
    exist_ok=True
)

excel_path = os.path.join(
    output_folder,
    "Liberia_Debt_Growth_Complete_Results.xlsx"
)

with pd.ExcelWriter(
    excel_path,
    engine="xlsxwriter"
) as writer:

    df.reset_index().to_excel(
        writer,
        sheet_name="Data",
        index=False
    )

    variable_dictionary.to_excel(
        writer,
        sheet_name="Variable_Dictionary",
        index=False
    )

    desc.to_excel(
        writer,
        sheet_name="Descriptive"
    )

    additional_desc.to_excel(
        writer,
        sheet_name="Descriptive_Extra"
    )

    corr.to_excel(
        writer,
        sheet_name="Correlation"
    )

    vif_table.to_excel(
        writer,
        sheet_name="VIF",
        index=False
    )

    adf_table.to_excel(
        writer,
        sheet_name="ADF_Level",
        index=False
    )

    adf_diff_table.to_excel(
        writer,
        sheet_name="ADF_Difference",
        index=False
    )

    pp_table.to_excel(
        writer,
        sheet_name="PP_Level",
        index=False
    )

    pp_diff_table.to_excel(
        writer,
        sheet_name="PP_Difference",
        index=False
    )

    kpss_table.to_excel(
        writer,
        sheet_name="KPSS_Level",
        index=False
    )

    kpss_diff_table.to_excel(
        writer,
        sheet_name="KPSS_Difference",
        index=False
    )

    unit_root_summary.to_excel(
        writer,
        sheet_name="UnitRoot_Summary",
        index=False
    )

    integration_table.to_excel(
        writer,
        sheet_name="Integration_Order",
        index=False
    )

    za_table.to_excel(
        writer,
        sheet_name="ZA_Level",
        index=False
    )

    za_ct_table.to_excel(
        writer,
        sheet_name="ZA_Trend",
        index=False
    )

    bp_table.to_excel(
        writer,
        sheet_name="Structural_Breaks",
        index=False
    )

    if len(lag_table) > 0:

        lag_table.to_excel(
            writer,
            sheet_name="ARDL_Lag_Selection",
            index=False
        )

    main_coef_table.to_excel(
        writer,
        sheet_name="ARDL_Coefficients",
        index=False
    )

    if long_run_table is not None:

        long_run_table.to_excel(
            writer,
            sheet_name="Long_Run",
            index=True
        )

    try:

        long_run_manual.to_excel(
            writer,
            sheet_name="Long_Run_Multipliers",
            index=False
        )

    except Exception:

        pass

    try:

        robustness_table.to_excel(
            writer,
            sheet_name="Robustness",
            index=False
        )

    except Exception:

        pass


# ================================================================
# 56. SAVE CSV FILES
# ================================================================

df.reset_index().to_csv(
    os.path.join(
        output_folder,
        "Liberia_Clean_Analysis_Data.csv"
    ),
    index=False
)

desc.to_csv(
    os.path.join(
        output_folder,
        "Descriptive_Statistics.csv"
    )
)

unit_root_summary.to_csv(
    os.path.join(
        output_folder,
        "Unit_Root_Summary.csv"
    ),
    index=False
)

za_table.to_csv(
    os.path.join(
        output_folder,
        "Zivot_Andrews_Results.csv"
    ),
    index=False
)

bp_table.to_csv(
    os.path.join(
        output_folder,
        "Structural_Break_Results.csv"
    ),
    index=False
)


# ================================================================
# 57. FINAL AUTOMATIC RESEARCH DECISION
# ================================================================

print("\n")
print("="*90)
print("FINAL AUTOMATED ECONOMETRIC DECISION")
print("="*90)

print(
    "\nStudy:"
)

print(
    "Government Debt and Economic Growth in Liberia: "
    "Evidence from Structural Breaks and Dynamic Time Series Analysis"
)

print(
    "\nSample:",
    df.index.min(),
    "to",
    df.index.max()
)

print(
    "Number of observations:",
    len(df)
)

print(
    "\nVariables:"
)

print(
    "GDPG = GDP Growth"
)

print(
    "DEBT = Public Debt"
)

print(
    "PE = Public Expenditure"
)

print(
    "TAX = Tax Revenue"
)

print(
    "LFP = Labour Force Participation"
)

print(
    "\nCAPITAL FORMATION HAS BEEN EXCLUDED."
)

print(
    "\nPreliminary ARDL admissibility:"
)

if "I(2)" in integration_table[
    "Preliminary integration order"
].values:

    print(
        "ARDL NOT RECOMMENDED because an I(2) variable was detected."
    )

else:

    print(
        "No I(2) variable was automatically detected."
    )

    print(
        "ARDL remains potentially admissible."
    )

print(
    "\nBounds test decision:"
)

print(
    cointegration_decision
)

if cointegration_decision == "Cointegration established":

    print(
        "\nRECOMMENDATION:"
    )

    print(
        "Proceed with long run ARDL and short run ECM interpretation."
    )

elif cointegration_decision == "No cointegration":

    print(
        "\nRECOMMENDATION:"
    )

    print(
        "Do not report an ARDL long run relationship or ECM as the "
        "main result. Consider a stationary dynamic model or "
        "alternative time series specification."
    )

else:

    print(
        "\nRECOMMENDATION:"
    )

    print(
        "The bounds test is inconclusive. Do not force an ECM. "
        "Investigate alternative specifications, structural breaks "
        "and small sample sensitivity."
    )


# ================================================================
# 58. DISPLAY FINAL FILE LOCATIONS
# ================================================================

print("\n" + "="*90)
print("OUTPUT FILES")
print("="*90)

for root, dirs, files in os.walk(
    output_folder
):

    for file in files:

        print(
            os.path.join(
                root,
                file
            )
        )

print(
    "\nAnalysis completed."
)

print(
    "\nIMPORTANT:"
)

print(
    "The empirical results should be interpreted only after "
    "examining the unit root, structural break, bounds test, "
    "diagnostic and stability results."
)
```
