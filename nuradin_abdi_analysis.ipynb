Complete Workflow for Real Dataset

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

from scipy.optimize import curve_fit
from scipy.stats import f_oneway

import matplotlib.pyplot as plt

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

# ============================
# 2. LOAD DATA
# ============================
# The dataset has already been loaded into 'df' in previous cells.
# Uncomment the following lines if you explicitly need to load from Google Drive
# and ensure the file path is correct.
# file_path = "NURADIN ABDI Real_Datasets.xlsx"
# df = pd.read_excel(file_path)

# Assuming 'df' and 'time' are already defined from previous successful executions.
# If not, ensure the cell that loads `df_uploaded` and the cell that sets `df = df_uploaded` and `time = df["Time"].values` run first.

# ============================
# 3. NONLINEAR MODELS
# ============================
def logistic(t,K,r,t0):
    return K/(1+np.exp(-r*(t-t0)))

def gompertz(t,K,r,t0):
    return K*np.exp(-np.exp(-r*(t-t0)))

def exponential(t,y0,r):
    return y0*np.exp(r*t)

def weibull(t,l,b):
    return np.exp(-(t/l)**b)

def fourier(t,a0,a1,b1,a2,b2,T):
    return (
        a0 +
        a1*np.cos(2*np.pi*t/T) +
        b1*np.sin(2*np.pi*t/T) +
        a2*np.cos(4*np.pi*t/T) +
        b2*np.sin(4*np.pi*t/T)
    )

# ============================
# 4. EVALUATION METRICS
# ============================
def evaluate(y, yp, k):
    n = len(y)
    rss = np.sum((y-yp)**2)

    mse = mean_squared_error(y, yp)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y, yp)
    r2 = r2_score(y, yp)

    aic = n*np.log(rss/n) + 2*k
    bic = n*np.log(rss/n) + k*np.log(n)
    hqc = n*np.log(rss/n) + 2*k*np.log(np.log(n))
    rice = n*np.log(rss/n) + 2*k

    return [r2, aic, bic, hqc, rice, mae, mse, rmse]

# ============================
# 5. MODEL SETUP
# ============================
models = {
    "Gompertz": (gompertz, "Gompertz", [100, 0.1, 50]),
    "Weibull": (weibull, "Weibull", [50, 2]),
    "Logistic": (logistic, "Logistic", [100, 0.1, 50]),
    "Exponential": (exponential, "Exponential", [1, 0.05]),
    "Fourier": (fourier, "Fourier", [20, 5, 5, 2, 2, 12])
}

# ============================
# 6. FIT MODELS
# ============================
results = []

print("\n================ MODEL RESULTS ================\n")

for name, (fn, col, p0) in models.items():

    y = df[col].values

    try:
        pars, _ = curve_fit(fn, time, y, p0=p0, maxfev=50000)
        pred = fn(time, *pars)

        metrics = evaluate(y, pred, len(pars))

        # PRINT PARAMETERS
        print(f"\nModel: {name}")
        for i, p in enumerate(pars, 1):
            print(f"Parameter P{i}: {p:.6f}")

        print("Evaluation Metrics:")
        print(f"R2   : {metrics[0]:.6f}")
        print(f"AIC  : {metrics[1]:.6f}")
        print(f"BIC  : {metrics[2]:.6f}")
        print(f"HQC  : {metrics[3]:.6f}")
        print(f"RICE : {metrics[4]:.6f}")
        print(f"MAE  : {metrics[5]:.6f}")
        print(f"MSE  : {metrics[6]:.6f}")
        print(f"RMSE : {metrics[7]:.6f}")

        results.append([name, *pars, *metrics])

        # ============================
        # PLOT FITTED MODELS
        # ============================

        plt.figure(figsize=(10,6))

        # Observed Data
        plt.plot(
            time,
            y,
            linewidth=2,
            label="Observed Data"
        )

        # Fitted Model
        plt.plot(
            time,
            pred,
            linewidth=2,
            label=f"{name} Model"
        )

        # TITLE
        plt.title(
            f"{name} Model Fit",
            fontsize=16,
            fontweight="bold"
        )

        # X AXIS LABEL
        plt.xlabel(
            "Time",
            fontsize=14
        )

        ylabel_map = {
            "Gompertz":"Population Size",
            "Weibull":"Survival Probability",
            "Logistic":"Yeast Growth",
            "Exponential":"Bacterial Growth",
            "Fourier":"Temperature (°C)"
        }

        plt.ylabel(
            ylabel_map[name],
            fontsize=14
        )

        # GRID
        plt.grid(True)

        # LEGEND
        plt.legend(
            fontsize=12
        )

        # TICKS
        plt.xticks(fontsize=11)
        plt.yticks(fontsize=11)

        # LAYOUT
        plt.tight_layout()

        plt.show()

    except Exception as e:
        print(f"{name} failed: {e}")

# ============================
# 7. RESULTS TABLE
# ============================
cols = [
    "Model","P1","P2","P3","P4","P5","P6",
    "R2","AIC","BIC","HQC","RICE","MAE","MSE","RMSE"
]

table = pd.DataFrame(results, columns=cols)
print("\n================ FINAL TABLE ================\n")
print(table)

# ============================
# 8. MODEL RANKING
# ============================
table["Score"] = (
    table["R2"].rank(ascending=False) +
    table["RMSE"].rank(ascending=True) +
    table["AIC"].rank(ascending=True)
)

table = table.sort_values("Score")

print("\n================ RANKED MODELS ================\n")
print(table[["Model","R2","RMSE","AIC","Score"]])
```
