```python
# ================================================================
# DIGITAL CONNECTIVITY AND INTERNET USE AS PREDICTORS OF
# UNIVERSITY STUDENTS' ACADEMIC PERFORMANCE:
# MULTISITE EVIDENCE FROM UGANDA
#
# COMPLETE GOOGLE COLAB ANALYSIS PIPELINE
#
# Objectives
# 1. Assess the level and dimensions of digital connectivity.
# 2. Examine relationships between DCI, AIU, NAIU and CGPA.
# 3. Develop and evaluate a multivariable model of CGPA while
#    controlling for demographic, academic and institutional factors.
#
# IMPORTANT:
# - This script DOES NOT invent a probability sampling procedure.
# - University recruitment is documented as availability/convenience
#   based multisite recruitment because enumerators were allowed to
#   administer questionnaires at universities within their reach.
# - Existing composite variables are NOT trusted. They are recalculated
#   from the questionnaire items.
# - B9 is reverse coded.
# - Invalid Likert values are converted to missing after being reported.
# - Invalid CGPA values outside 0-5 are flagged and converted to missing.
# - Outliers are flagged, not automatically deleted.
# - Institutional clustering is explicitly assessed.
# ================================================================

# IMPORTANT:
# After this cell finishes, restart the Colab runtime:
# Runtime -> Restart session
#
# Then run the analysis from the beginning.

import os
import re
import glob
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import (
    pearsonr,
    spearmanr,
    shapiro,
    chi2_contingency,
    normaltest
)

import statsmodels.api as sm
import statsmodels.formula.api as smf

from statsmodels.stats.diagnostic import (
    het_breuschpagan,
    het_white,
    linear_reset
)

from statsmodels.stats.outliers_influence import (
    variance_inflation_factor,
    OLSInfluence
)

from statsmodels.stats.anova import anova_lm

from sklearn.model_selection import (
    GroupKFold,
    KFold,
    cross_validate
)

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression, Ridge, ElasticNet

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import pingouin as pg

from semopy import Model, calc_stats

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 200)

print("Libraries loaded successfully.")

# ================================================================
# REQUIRED FACTOR ANALYSIS IMPORTS
# ================================================================
from factor_analyzer import FactorAnalyzer
from factor_analyzer.factor_analyzer import (
    calculate_kmo,
    calculate_bartlett_sphericity
)

# ================================================================
# 2. MOUNT GOOGLE DRIVE
# ================================================================

from google.colab import drive

drive.mount("/content/drive")

BASE_DIR = "/content/drive/MyDrive/Digital_Connectivity_Uganda_Study"

os.makedirs(BASE_DIR, exist_ok=True)

DATA_DIR = os.path.join(BASE_DIR, "data")
RESULT_DIR = os.path.join(BASE_DIR, "results")
TABLE_DIR = os.path.join(RESULT_DIR, "tables")
FIG_DIR = os.path.join(RESULT_DIR, "figures")
AUDIT_DIR = os.path.join(RESULT_DIR, "data_audit")

for folder in [DATA_DIR, RESULT_DIR, TABLE_DIR, FIG_DIR, AUDIT_DIR]:
    os.makedirs(folder, exist_ok=True)

print("Project directory:", BASE_DIR)


# ================================================================
# 3. UPLOAD EXCEL DATA FILE
# ================================================================

from google.colab import files

print("Please upload the Excel dataset (.xlsx or .xls).")

uploaded = files.upload()

uploaded_files = list(uploaded.keys())

excel_files = [
    f for f in uploaded_files
    if f.lower().endswith((".xlsx", ".xls", ".csv"))
]

if len(excel_files) == 0:
    raise FileNotFoundError(
        "No Excel/CSV file was uploaded. Please upload the dataset."
    )

INPUT_FILE = excel_files[0]

print("Selected input file:", INPUT_FILE)

# Copy original uploaded dataset into project data folder
import shutil

ORIGINAL_PATH = os.path.join(DATA_DIR, "original_dataset" + os.path.splitext(INPUT_FILE)[1])

shutil.copy(INPUT_FILE, ORIGINAL_PATH)

print("Original dataset saved to:", ORIGINAL_PATH)


# ================================================================
# 4. READ DATA
# ================================================================

if INPUT_FILE.lower().endswith(".csv"):
    df_raw = pd.read_csv(INPUT_FILE)
else:
    excel_obj = pd.ExcelFile(INPUT_FILE)
    print("Available sheets:", excel_obj.sheet_names)

    # Use first worksheet by default
    df_raw = pd.read_excel(INPUT_FILE, sheet_name=0)

print("\nRaw data shape:", df_raw.shape)

print("\nRaw columns:")
print(df_raw.columns.tolist())

display(df_raw.head())


# ================================================================
# 5. STANDARDISE COLUMN NAMES
# ================================================================

df = df_raw.copy()

def clean_column_name(x):
    x = str(x)
    x = x.replace("\xa0", " ")
    x = x.strip()
    x = re.sub(r"\s+", "_", x)
    x = re.sub(r"[^A-Za-z0-9_]", "", x)
    return x

df.columns = [clean_column_name(c) for c in df.columns]

print("\nCleaned columns:")
print(df.columns.tolist())


# ================================================================
# 6. EXPECTED VARIABLES
# ================================================================

expected_columns = [
    "Universities",
    "Age",
    "Gender",
    "Study_Year",
    "Study_Level",
    "Academic_Discipline",
    "Academic_Resisdence",
    "Internet_access_device",
    "Average_daily_internet_use"
]

B_ITEMS = [f"B{i}" for i in range(1, 11)]
C_ITEMS = [f"C{i}" for i in range(1, 11)]
D_ITEMS = [f"D{i}" for i in range(1, 7)]

ALL_ITEMS = B_ITEMS + C_ITEMS + D_ITEMS

expected_columns_full = (
    expected_columns +
    B_ITEMS +
    ["Digital_Connectivity"] +
    C_ITEMS +
    ["Academic_Internet_Use"] +
    D_ITEMS +
    ["Non-Academic_Internet_Use"] +
    ["CGPA"]
)

missing_columns = [
    c for c in expected_columns_full
    if c not in df.columns
]

if missing_columns:
    print("\nWARNING: The following expected variables are missing:")
    print(missing_columns)
else:
    print("\nAll expected variables are present.")


# ================================================================
# 7. DATASET SIZE AND INITIAL AUDIT
# ================================================================

initial_n = len(df)

print("\n================================================")
print("INITIAL DATASET AUDIT")
print("================================================")
print("Number of rows:", initial_n)
print("Number of columns:", len(df.columns))

duplicate_rows = df.duplicated().sum()

print("Exact duplicate rows:", duplicate_rows)

missing_summary = (
    df.isna()
      .sum()
      .to_frame("Missing_n")
)

missing_summary["Missing_percent"] = (
    missing_summary["Missing_n"] / initial_n * 100
)

display(
    missing_summary[
        missing_summary["Missing_n"] > 0
    ].sort_values("Missing_n", ascending=False)
)


# ================================================================
# 8. PRESERVE ORIGINAL COMPOSITE VARIABLES FOR AUDIT ONLY
# ================================================================

existing_composites = [
    "Digital_Connectivity",
    "Academic_Internet_Use",
    "Non-Academic_Internet_Use"
]

existing_composite_backup = df[
    [c for c in existing_composites if c in df.columns]
].copy()

existing_composite_backup.to_csv(
    os.path.join(AUDIT_DIR, "original_composite_scores.csv"),
    index=False
)

print(
    "\nExisting composite scores have been preserved for audit "
    "but will NOT be used as primary analytical variables."
)


# ================================================================
# 9. CONVERT LIKERT ITEMS TO NUMERIC
# ================================================================

for c in ALL_ITEMS:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors="coerce")

df["CGPA"] = pd.to_numeric(
    df["CGPA"],
    errors="coerce"
)


# ================================================================
# 10. CHECK INVALID LIKERT RESPONSES
# ================================================================

invalid_likert_records = []

for c in ALL_ITEMS:

    if c not in df.columns:
        continue

    invalid_mask = (
        df[c].notna() &
        ~df[c].between(1, 5)
    )

    if invalid_mask.any():

        temp = df.loc[
            invalid_mask,
            ["Universities", c]
        ].copy()

        temp["Variable"] = c
        temp["Invalid_Value"] = df.loc[
            invalid_mask,
            c
        ].values

        invalid_likert_records.append(temp)

        # Invalid values are treated as missing
        df.loc[invalid_mask, c] = np.nan


if invalid_likert_records:

    invalid_likert_df = pd.concat(
        invalid_likert_records,
        ignore_index=True
    )

else:

    invalid_likert_df = pd.DataFrame(
        columns=[
            "Universities",
            "Variable",
            "Invalid_Value"
        ]
    )


invalid_likert_df.to_csv(
    os.path.join(
        AUDIT_DIR,
        "invalid_likert_responses.csv"
    ),
    index=False
)

print("\nInvalid Likert observations detected:")
print(len(invalid_likert_df))

if len(invalid_likert_df) > 0:
    display(invalid_likert_df)


# ================================================================
# 11. SPECIFIC AUDIT OF B3 AND B5
# ================================================================

print("\n================================================")
print("SPECIFIC AUDIT OF B3 AND B5")
print("================================================")

for c in ["B3", "B5"]:

    print(f"\n{c} value frequencies before cleaning:")
    print(
        df[c]
        .value_counts(dropna=False)
        .sort_index()
    )

print(
    "\nAny values outside 1-5 in B3 or B5 were treated as "
    "invalid and converted to missing."
)


# ================================================================
# 12. CGPA VALIDITY CHECK
# ================================================================

invalid_cgpa_mask = (
    df["CGPA"].notna() &
    ~df["CGPA"].between(0, 5)
)

invalid_cgpa = df.loc[
    invalid_cgpa_mask,
    ["Universities", "CGPA"]
].copy()

invalid_cgpa.to_csv(
    os.path.join(
        AUDIT_DIR,
        "invalid_CGPA_records.csv"
    ),
    index=False
)

print("\nInvalid CGPA observations:", len(invalid_cgpa))

if len(invalid_cgpa) > 0:
    display(invalid_cgpa)

# Do not allow impossible CGPA values
df.loc[invalid_cgpa_mask, "CGPA"] = np.nan


# ================================================================
# 13. CLEAN TEXT VARIABLES
# ================================================================

text_columns = [
    "Universities",
    "Age",
    "Gender",
    "Study_Year",
    "Study_Level",
    "Academic_Discipline",
    "Academic_Resisdence",
    "Internet_access_device",
    "Average_daily_internet_use"
]

for c in text_columns:

    if c in df.columns:

        df[c] = (
            df[c]
            .astype(str)
            .replace("nan", np.nan)
            .str.replace("\xa0", " ", regex=False)
            .str.replace("<br>", " ", regex=False)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )


# ================================================================
# 14. HARMONISE UNIVERSITY NAMES
# ================================================================

def normalise_university_name(x):

    if pd.isna(x):
        return np.nan

    x = str(x)
    x = x.replace("\xa0", " ")
    x = re.sub(r"\s+", " ", x).strip().upper()

    university_map = {

        # Kampala International University
        "KIU":
            "KAMPALA INTERNATIONAL UNIVERSITY",

        "KAMPALA INTERNATIONAL UNIVERSITY":
            "KAMPALA INTERNATIONAL UNIVERSITY",

        "KAMPALA INTERNATIONAL UNIVERSITY":
            "KAMPALA INTERNATIONAL UNIVERSITY",

        # Makerere University Business School
        "MUBS":
            "MAKERERE UNIVERSITY BUSINESS SCHOOL",

        "MAKERERE UNIVERSITY BUSINESS SCHOOL":
            "MAKERERE UNIVERSITY BUSINESS SCHOOL",

        "MUBS STRESS":
            "MAKERERE UNIVERSITY BUSINESS SCHOOL",

        # International University of East Africa
        "IUEA":
            "INTERNATIONAL UNIVERSITY OF EAST AFRICA",

        # Victoria University
        "VICTORIA":
            "VICTORIA UNIVERSITY",

        "VICTORIA UNIVERSITY":
            "VICTORIA UNIVERSITY",

        "VU":
            "VICTORIA UNIVERSITY",

        # Mbarara University of Science and Technology
        "MUST":
            "MBARARA UNIVERSITY OF SCIENCE AND TECHNOLOGY",

        "MBARARA UNIVERSITY OF SCIENCE AND TECHNOLOGY":
            "MBARARA UNIVERSITY OF SCIENCE AND TECHNOLOGY",

        # Kyambogo
        "KYU":
            "KYAMBOGO UNIVERSITY",

        "KYAMBOGO":
            "KYAMBOGO UNIVERSITY",

        "KYAMBOGO UNIVERSITY":
            "KYAMBOGO UNIVERSITY",

        # Muni
        "MUNI":
            "MUNI UNIVERSITY",

        "MUNI UNIVERSITY":
            "MUNI UNIVERSITY",

        # Kabale
        "KABALE":
            "KABALE UNIVERSITY",

        "KABALE UNIVERSITY":
            "KABALE UNIVERSITY",

        # Makerere
        "MAKERERE":
            "MAKERERE UNIVERSITY",

        "MAKERERE UNIVERSITY":
            "MAKERERE UNIVERSITY",

        "MUK":
            "MAKERERE UNIVERSITY",

        # Soroti
        "SOROTI":
            "SOROTI UNIVERSITY",

        "SOROTI UNIVERSITY":
            "SOROTI UNIVERSITY",

        "SORITI":
            "SOROTI UNIVERSITY",

        # Other known universities
        "BISHOP STUART":
            "BISHOP STUART UNIVERSITY",

        "BISHOP STUART UNIVERSITY":
            "BISHOP STUART UNIVERSITY",

        "KING CEASOR":
            "KING CEASOR UNIVERSITY",

        "KING CEASOR UNIVERSITY":
            "KING CEASOR UNIVERSITY",

        "METROPOLITAN":
            "METROPOLITAN INTERNATIONAL UNIVERSITY",

        "METROPOLITAN INTERNATIONAL UNIVERSITY":
            "METROPOLITAN INTERNATIONAL UNIVERSITY",

        "KAMPALA UNIVERSITY":
            "KAMPALA UNIVERSITY",

        "NDEJJE UNIVERSITY":
            "NDEJJE UNIVERSITY",

        "BUSOGA UNIVERSITY":
            "BUSOGA UNIVERSITY",

        "BUSITEMA UNIVERSITY":
            "BUSITEMA UNIVERSITY",

        "UMU":
            "UGANDA MARTYRS UNIVERSITY",

        "UGANDA MARTYRS UNIVERSITY":
            "UGANDA MARTYRS UNIVERSITY",

        "ISLAMIC UNIVERSITY IN UGANDA":
            "ISLAMIC UNIVERSITY IN UGANDA"
    }

    return university_map.get(x, x)


df["University"] = df["Universities"].apply(
    normalise_university_name
)

university_audit = pd.DataFrame({
    "Original":
        df["Universities"].astype(str),
    "Harmonised":
        df["University"].astype(str)
})

university_audit = (
    university_audit
    .drop_duplicates()
    .sort_values(["Harmonised", "Original"])
)

university_audit.to_csv(
    os.path.join(
        AUDIT_DIR,
        "university_name_harmonisation.csv"
    ),
    index=False
)

print("\nUniversity harmonisation:")
display(university_audit)

print("\nFinal university frequencies:")
display(
    df["University"]
    .value_counts()
    .to_frame("n")
)


# ================================================================
# 15. HARMONISE AGE
# ================================================================

def clean_age(x):

    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    numeric_map = {
        "1": "Below 20 years",
        "2": "20 to 24 years",
        "3": "25 to 29 years",
        "4": "30 years and above"
    }

    if x in numeric_map:
        return numeric_map[x]

    if x == "5":
        return np.nan

    if "below 20" in x:
        return "Below 20 years"

    if "20" in x and "24" in x:
        return "20 to 24 years"

    if "21" in x and "24" in x:
        return "20 to 24 years"

    if "22" in x and "24" in x:
        return "20 to 24 years"

    if "23" in x and "24" in x:
        return "20 to 24 years"

    if "25" in x and "29" in x:
        return "25 to 29 years"

    if "26" in x and "29" in x:
        return "25 to 29 years"

    if "27" in x and "29" in x:
        return "25 to 29 years"

    if "30" in x or "31" in x or "32" in x:
        return "30 years and above"

    return np.nan


df["Age_clean"] = df["Age"].apply(clean_age)


# ================================================================
# 16. HARMONISE GENDER
# ================================================================

def clean_gender(x):

    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    mapping = {
        "1": "Male",
        "2": "Female",
        "male": "Male",
        "female": "Female"
    }

    return mapping.get(x, np.nan)


df["Gender_clean"] = df["Gender"].apply(clean_gender)


# ================================================================
# 17. HARMONISE STUDY YEAR
# ================================================================

def clean_study_year(x):

    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    mapping = {
        "1": "First year",
        "2": "Second year",
        "3": "Third year",
        "4": "Fourth year",
        "5": "Fifth year or above",

        "first year": "First year",
        "second year": "Second year",
        "third year": "Third year",
        "fourth year": "Fourth year",
        "fifth year or above": "Fifth year or above"
    }

    return mapping.get(x, np.nan)


df["Study_Year_clean"] = df["Study_Year"].apply(
    clean_study_year
)


# ================================================================
# 18. HARMONISE STUDY LEVEL
# ================================================================

def clean_study_level(x):

    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    mapping = {
        "1": "Diploma",
        "2": "Bachelor's degree",
        "3": "Postgraduate",

        "diploma": "Diploma",
        "bachelor's degree": "Bachelor's degree",
        "bachelor's degree ": "Bachelor's degree",
        "postgraduate": "Postgraduate"
    }

    return mapping.get(x, np.nan)


df["Study_Level_clean"] = df["Study_Level"].apply(
    clean_study_level
)


# ================================================================
# 19. HARMONISE ACADEMIC DISCIPLINE
# ================================================================

def clean_discipline(x):

    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    mapping = {

        "1": "Education",
        "education": "Education",

        "2": "Business and Management",
        "business and management": "Business and Management",

        "3": "Computing and Information Technology",
        "computing and information technology":
            "Computing and Information Technology",
        "computing and it":
            "Computing and Information Technology",
        "computer and it":
            "Computing and Information Technology",

        "4": "Science and Mathematics",
        "science and mathematics":
            "Science and Mathematics",

        "5": "Health Sciences",
        "health sciences":
            "Health Sciences",
        "health science":
            "Health Sciences",

        "6": "Social Sciences",
        "social sciences":
            "Social Sciences",

        "7": "Engineering",
        "engineering":
            "Engineering",

        "8": "Humanities",
        "humanities":
            "Humanities",
        "arts and humanities":
            "Humanities",

        "9": "Other",

        "other": "Other",
        "other:": "Other",

        "law": "Other",
        "agriculture": "Other",
        "bstat": "Other",
        "economics and applied statistics": "Other"
    }

    return mapping.get(x, np.nan)


df["Academic_Discipline_clean"] = (
    df["Academic_Discipline"]
    .apply(clean_discipline)
)


# ================================================================
# 20. HARMONISE RESIDENCE
# ================================================================

def clean_residence(x):

    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    mapping = {

        "1": "University residence",
        "2": "Private hostel",
        "3": "Family home",
        "4": "Rented accommodation",
        "5": "Other",

        "university residence":
            "University residence",

        "private hostel":
            "Private hostel",

        "family home":
            "Family home",

        "rented accommodation":
            "Rented accommodation",

        "rented accomodation":
            "Rented accommodation",

        "other":
            "Other",

        "private hostel family home":
            "Other"
    }

    return mapping.get(x, np.nan)


df["Residence_clean"] = (
    df["Academic_Resisdence"]
    .apply(clean_residence)
)


# ================================================================
# 21. HARMONISE PRIMARY INTERNET DEVICE
# ================================================================

def clean_device(x):

    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    mapping = {

        "1": "Smartphone",
        "2": "Laptop",
        "3": "Tablet",
        "4": "Desktop computer",
        "5": "More than one device",

        "smartphone":
            "Smartphone",

        "laptop":
            "Laptop",

        "tablet":
            "Tablet",

        "desktop computer":
            "Desktop computer",

        "more than one device":
            "More than one device",

        "smartphone laptop":
            "More than one device"
    }

    return mapping.get(x, np.nan)


df["Device_clean"] = (
    df["Internet_access_device"]
    .apply(clean_device)
)


# ================================================================
# 22. HARMONISE DAILY INTERNET USE
# ================================================================

def clean_daily_use(x):

    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    mapping = {

        "1": "Less than 1 hour",
        "2": "1 to 2 hours",
        "3": "3 to 4 hours",
        "4": "5 to 6 hours",
        "5": "More than 6 hours",

        "less than 1 hour":
            "Less than 1 hour",

        "1 to 2 hours":
            "1 to 2 hours",

        "3 to 4 hours":
            "3 to 4 hours",

        "5 to 6 hours":
            "5 to 6 hours",

        "more than 6 hours":
            "More than 6 hours"
    }

    return mapping.get(x, np.nan)


df["Daily_Internet_Use_clean"] = (
    df["Average_daily_internet_use"]
    .apply(clean_daily_use)
)


# ================================================================
# 23. REPORT CLEANING RESULTS
# ================================================================

cleaning_summary = pd.DataFrame({

    "Variable": [
        "Age",
        "Gender",
        "Study year",
        "Study level",
        "Academic discipline",
        "Residence",
        "Primary device",
        "Daily internet use",
        "University"
    ],

    "Missing_after_cleaning": [
        df["Age_clean"].isna().sum(),
        df["Gender_clean"].isna().sum(),
        df["Study_Year_clean"].isna().sum(),
        df["Study_Level_clean"].isna().sum(),
        df["Academic_Discipline_clean"].isna().sum(),
        df["Residence_clean"].isna().sum(),
        df["Device_clean"].isna().sum(),
        df["Daily_Internet_Use_clean"].isna().sum(),
        df["University"].isna().sum()
    ]
})

cleaning_summary["Missing_percent"] = (
    cleaning_summary["Missing_after_cleaning"] /
    len(df) * 100
)

display(cleaning_summary)


# ================================================================
# 24. CREATE ANALYTICAL ITEM DATA
# ================================================================

# Reverse coded B9
df["B9_R"] = 6 - df["B9"]


# ================================================================
# 25. RECOMPUTE DIGITAL CONNECTIVITY INDEX
# ================================================================

DCI_ITEMS = [
    "B1",
    "B2",
    "B3",
    "B4",
    "B5",
    "B6",
    "B7",
    "B8",
    "B9_R",
    "B10"
]

df["DCI_n_answered"] = (
    df[DCI_ITEMS].notna().sum(axis=1)
)

df["DCI"] = (
    df[DCI_ITEMS]
    .mean(axis=1)
    .where(df["DCI_n_answered"] >= 8)
)


# ================================================================
# 26. RECOMPUTE ACADEMIC INTERNET USE
# ================================================================

df["AIU_n_answered"] = (
    df[C_ITEMS].notna().sum(axis=1)
)

df["AIU"] = (
    df[C_ITEMS]
    .mean(axis=1)
    .where(df["AIU_n_answered"] >= 8)
)


# ================================================================
# 27. RECOMPUTE NON ACADEMIC INTERNET USE
# ================================================================

df["NAIU_n_answered"] = (
    df[D_ITEMS].notna().sum(axis=1)
)

df["NAIU"] = (
    df[D_ITEMS]
    .mean(axis=1)
    .where(df["NAIU_n_answered"] >= 5)
)


# ================================================================
# 28. DIGITAL CONNECTIVITY DIMENSIONS
# ================================================================

# These dimensions follow the questionnaire content.

ACCESS_ITEMS = [
    "B1",
    "B2",
    "B8",
    "B10"
]

AFFORDABILITY_ITEMS = [
    "B3"
]

RELIABILITY_ITEMS = [
    "B5",
    "B9_R"
]

SPEED_ITEMS = [
    "B4"
]

CAMPUS_WIFI_ITEMS = [
    "B6",
    "B7"
]

dimension_specs = {

    "Access": ACCESS_ITEMS,

    "Affordability": AFFORDABILITY_ITEMS,

    "Reliability": RELIABILITY_ITEMS,

    "Speed": SPEED_ITEMS,

    "Campus_WiFi": CAMPUS_WIFI_ITEMS
}


for dimension, items in dimension_specs.items():

    df[dimension] = df[items].mean(axis=1)


# ================================================================
# 29. DIGITAL CONNECTIVITY LEVEL
# ================================================================

def classify_likert_mean(x):

    if pd.isna(x):
        return np.nan

    if x < 2.34:
        return "Low"

    elif x < 3.67:
        return "Moderate"

    else:
        return "High"


df["DCI_Level"] = (
    df["DCI"]
    .apply(classify_likert_mean)
)


# ================================================================
# 30. DATA QUALITY CHECK FOR COMPOSITE VARIABLES
# ================================================================

composite_summary = pd.DataFrame({

    "Construct": [
        "Digital Connectivity Index",
        "Academic Internet Use",
        "Non Academic Internet Use"
    ],

    "Items": [
        10,
        10,
        6
    ],

    "Missing_score": [
        df["DCI"].isna().sum(),
        df["AIU"].isna().sum(),
        df["NAIU"].isna().sum()
    ],

    "Mean": [
        df["DCI"].mean(),
        df["AIU"].mean(),
        df["NAIU"].mean()
    ],

    "SD": [
        df["DCI"].std(),
        df["AIU"].std(),
        df["NAIU"].std()
    ]
})

display(composite_summary)


# ================================================================
# 31. FINAL ANALYTICAL DATASET
# ================================================================

analysis_columns = [
    "University",
    "Age_clean",
    "Gender_clean",
    "Study_Year_clean",
    "Study_Level_clean",
    "Academic_Discipline_clean",
    "Residence_clean",
    "Device_clean",
    "Daily_Internet_Use_clean",

    *B_ITEMS,
    "B9_R",

    *C_ITEMS,
    *D_ITEMS,

    "DCI",
    "DCI_Level",   # <-- ADD THIS LINE
    "AIU",
    "NAIU",

    "Access",
    "Affordability",
    "Reliability",
    "Speed",
    "Campus_WiFi",

    "CGPA"
]

analysis_df = df[
    [c for c in analysis_columns if c in df.columns]
].copy()

print("\nInitial N:", initial_n)
print("Analytical dataset N:", len(analysis_df))


# ================================================================
# 32. IDENTIFY COMPLETE CASES FOR PRIMARY REGRESSION
# ================================================================

model_variables = [
    "CGPA",
    "DCI",
    "AIU",
    "NAIU",
    "University",
    "Age_clean",
    "Gender_clean",
    "Study_Year_clean",
    "Study_Level_clean",
    "Academic_Discipline_clean",
    "Residence_clean",
    "Device_clean",
    "Daily_Internet_Use_clean"
]

model_df = analysis_df[
    model_variables
].dropna().copy()

print(
    "\nComplete cases for primary multivariable analysis:",
    len(model_df)
)


# ================================================================
# 33. SAVE CLEANED DATA
# ================================================================

analysis_df.to_csv(
    os.path.join(
        DATA_DIR,
        "cleaned_analysis_dataset.csv"
    ),
    index=False
)

model_df.to_csv(
    os.path.join(
        DATA_DIR,
        "complete_case_regression_dataset.csv"
    ),
    index=False
)

print("Cleaned datasets saved.")


# ================================================================
# 34. OBJECTIVE 1:
# RESPONDENT CHARACTERISTICS
# ================================================================

categorical_variables = {
    "Age": "Age_clean",
    "Gender": "Gender_clean",
    "Study year": "Study_Year_clean",
    "Study level": "Study_Level_clean",
    "Academic discipline": "Academic_Discipline_clean",
    "Residence": "Residence_clean",
    "Primary internet device": "Device_clean",
    "Daily internet use": "Daily_Internet_Use_clean",
    "University": "University"
}


def frequency_table(data, variable, label):

    tab = (
        data[variable]
        .value_counts(dropna=False)
        .rename_axis("Category")
        .reset_index(name="n")
    )

    tab["Percent"] = (
        tab["n"] / len(data) * 100
    ).round(2)

    tab.insert(
        0,
        "Variable",
        label
    )

    return tab


demographic_tables = []

for label, variable in categorical_variables.items():

    demographic_tables.append(
        frequency_table(
            analysis_df,
            variable,
            label
        )
    )

demographic_table = pd.concat(
    demographic_tables,
    ignore_index=True
)

display(demographic_table)

demographic_table.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_1_respondent_characteristics.csv"
    ),
    index=False
)


# ================================================================
# 35. CONTINUOUS DESCRIPTIVE STATISTICS
# ================================================================

continuous_variables = [
    "DCI",
    "AIU",
    "NAIU",
    "Access",
    "Affordability",
    "Reliability",
    "Speed",
    "Campus_WiFi",
    "CGPA"
]

descriptive_rows = []

for c in continuous_variables:

    s = pd.to_numeric(
        analysis_df[c],
        errors="coerce"
    ).dropna()

    descriptive_rows.append({

        "Variable": c,
        "N": s.count(),
        "Mean": s.mean(),
        "SD": s.std(),
        "Median": s.median(),
        "Minimum": s.min(),
        "Maximum": s.max(),
        "IQR": s.quantile(.75) - s.quantile(.25)
    })

descriptive_table = pd.DataFrame(
    descriptive_rows
)

display(descriptive_table)

descriptive_table.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_2_descriptive_statistics.csv"
    ),
    index=False
)


# ================================================================
# 36. OBJECTIVE 1:
# DIGITAL CONNECTIVITY DIMENSIONS
# ================================================================

dimension_order = [
    "Access",
    "Affordability",
    "Reliability",
    "Speed",
    "Campus_WiFi",
    "DCI"
]

dimension_table = descriptive_table[
    descriptive_table["Variable"].isin(
        dimension_order
    )
].copy()

dimension_table["Order"] = (
    dimension_table["Variable"]
    .map({
        v: i
        for i, v in enumerate(
            dimension_order
        )
    })
)

dimension_table = (
    dimension_table
    .sort_values("Order")
    .drop(columns="Order")
)

display(dimension_table)

dimension_table.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_3_digital_connectivity_dimensions.csv"
    ),
    index=False
)


# ================================================================
# 37. DIGITAL CONNECTIVITY LEVEL
# ================================================================

dci_level_table = (
    analysis_df["DCI_Level"]
    .value_counts()
    .rename_axis("Level")
    .reset_index(name="n")
)

dci_level_table["Percent"] = (
    dci_level_table["n"] /
    dci_level_table["n"].sum() * 100
).round(2)

display(dci_level_table)

dci_level_table.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_4_digital_connectivity_level.csv"
    ),
    index=False
)


# ================================================================
# 38. VISUALISATION:
# DIGITAL CONNECTIVITY DIMENSIONS
# ================================================================

plot_df = dimension_table.copy()

plt.figure(figsize=(10, 6))

sns.barplot(
    data=plot_df,
    x="Variable",
    y="Mean"
)

plt.axhline(
    3,
    linestyle="--",
    linewidth=1
)

plt.ylim(1, 5)

plt.ylabel("Mean score")
plt.xlabel("Digital connectivity dimension")
plt.title(
    "Digital connectivity dimensions among university students"
)

plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIG_DIR,
        "Figure_1_digital_connectivity_dimensions.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 39. CGPA DISTRIBUTION
# ================================================================

plt.figure(figsize=(9, 6))

sns.histplot(
    analysis_df["CGPA"].dropna(),
    kde=True,
    bins=20
)

plt.xlabel("CGPA")
plt.ylabel("Frequency")
plt.title("Distribution of students' CGPA")

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIG_DIR,
        "Figure_2_CGPA_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 40. BOX PLOTS FOR MAIN CONSTRUCTS
# ================================================================

long_construct = analysis_df[
    ["DCI", "AIU", "NAIU"]
].melt(
    var_name="Construct",
    value_name="Score"
)

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=long_construct,
    x="Construct",
    y="Score"
)

plt.ylim(1, 5)

plt.title(
    "Distribution of the principal study constructs"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIG_DIR,
        "Figure_3_construct_distributions.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 41. OBJECTIVE 2:
# CORRELATION ANALYSIS
# ================================================================

correlation_variables = [
    "DCI",
    "AIU",
    "NAIU",
    "CGPA"
]

corr_data = analysis_df[
    correlation_variables
].dropna()

pearson_r = pd.DataFrame(
    index=correlation_variables,
    columns=correlation_variables,
    dtype=float
)

pearson_p = pd.DataFrame(
    index=correlation_variables,
    columns=correlation_variables,
    dtype=float
)

spearman_r = pd.DataFrame(
    index=correlation_variables,
    columns=correlation_variables,
    dtype=float
)

spearman_p = pd.DataFrame(
    index=correlation_variables,
    columns=correlation_variables,
    dtype=float
)

for x in correlation_variables:

    for y in correlation_variables:

        if x == y:

            pearson_r.loc[x, y] = 1
            pearson_p.loc[x, y] = 0

            spearman_r.loc[x, y] = 1
            spearman_p.loc[x, y] = 0

        else:

            r, p = scipy.stats.pearsonr(
                corr_data[x],
                corr_data[y]
            )

            sr, sp = scipy.stats.spearmanr(
                corr_data[x],
                corr_data[y]
            )

            pearson_r.loc[x, y] = r
            pearson_p.loc[x, y] = p

            spearman_r.loc[x, y] = sr
            spearman_p.loc[x, y] = sp


print("\nPearson correlations")
display(pearson_r.round(3))

print("\nPearson p-values")
display(pearson_p.round(4))

print("\nSpearman correlations")
display(spearman_r.round(3))


pearson_r.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_5_Pearson_correlations.csv"
    )
)


# ================================================================
# 42. CORRELATION HEATMAP
# ================================================================

plt.figure(figsize=(8, 6))

sns.heatmap(
    pearson_r.astype(float),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1
)

plt.title(
    "Pearson correlations among principal study variables"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIG_DIR,
        "Figure_4_correlation_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 43. RELIABILITY FUNCTIONS
# ================================================================

def cronbach_alpha(data):

    x = data.dropna()

    k = x.shape[1]

    if k < 2:
        return np.nan

    item_variances = x.var(axis=0, ddof=1)

    total_variance = x.sum(axis=1).var(ddof=1)

    if total_variance == 0:
        return np.nan

    alpha = (
        k / (k - 1)
    ) * (
        1 -
        item_variances.sum() /
        total_variance
    )

    return alpha


def corrected_item_total_correlations(data):

    x = data.dropna()

    results = []

    for item in x.columns:

        other_items = [
            c for c in x.columns
            if c != item
        ]

        total_without_item = (
            x[other_items]
            .sum(axis=1)
        )

        r = x[item].corr(
            total_without_item
        )

        results.append({
            "Item": item,
            "Corrected_Item_Total_r": r
        })

    return pd.DataFrame(results)


def alpha_if_item_deleted(data):

    results = []

    for item in data.columns:

        reduced = data.drop(
            columns=[item]
        )

        results.append({

            "Item": item,

            "Alpha_if_deleted":
                cronbach_alpha(reduced)
        })

    return pd.DataFrame(results)


# ================================================================
# 44. McDONALD'S OMEGA
# ================================================================

def omega_total(data):

    x = data.dropna()

    if x.shape[0] < 20:
        return np.nan

    try:

        fa = FactorAnalyzer(
            n_factors=1,
            rotation=None,
            method="minres"
        )

        fa.fit(x)

        loadings = (
            fa.loadings_
            .flatten()
        )

        uniqueness = (
            fa.get_uniquenesses()
        )

        numerator = (
            np.sum(loadings)
        ) ** 2

        denominator = (
            numerator +
            np.sum(uniqueness)
        )

        return numerator / denominator

    except Exception as e:

        print(
            "Omega calculation failed:",
            e
        )

        return np.nan


# ================================================================
# 45. RELIABILITY ANALYSIS
# ================================================================

reliability_specs = {

    "Digital Connectivity":
        B_ITEMS[:-1] + ["B9_R", "B10"],

    "Academic Internet Use":
        C_ITEMS,

    "Non Academic Internet Use":
        D_ITEMS
}

# Remove duplicate B9/B10 issue safely
reliability_specs["Digital Connectivity"] = [
    "B1",
    "B2",
    "B3",
    "B4",
    "B5",
    "B6",
    "B7",
    "B8",
    "B9_R",
    "B10"
]

reliability_summary = []

item_reliability_tables = []

for construct, items in reliability_specs.items():

    x = df[items].copy()

    alpha = cronbach_alpha(x)

    omega = omega_total(x)

    citc = corrected_item_total_correlations(x)

    aid = alpha_if_item_deleted(x)

    item_table = citc.merge(
        aid,
        on="Item"
    )

    item_table.insert(
        0,
        "Construct",
        construct
    )

    item_reliability_tables.append(
        item_table
    )

    reliability_summary.append({

        "Construct": construct,
        "Items": len(items),
        "Complete_cases": x.dropna().shape[0],
        "Cronbach_alpha": alpha,
        "McDonalds_omega": omega
    })


reliability_summary = pd.DataFrame(
    reliability_summary
)

item_reliability = pd.concat(
    item_reliability_tables,
    ignore_index=True
)

print("\nRELIABILITY SUMMARY")
display(
    reliability_summary.round(3)
)

print("\nITEM LEVEL RELIABILITY")
display(
    item_reliability.round(3)
)

reliability_summary.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_6_reliability_summary.csv"
    ),
    index=False
)

item_reliability.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_7_item_reliability.csv"
    ),
    index=False
)


# ================================================================
# 46. KMO AND BARTLETT'S TEST
# ================================================================

VALIDATION_ITEMS = [
    "B1",
    "B2",
    "B3",
    "B4",
    "B5",
    "B6",
    "B7",
    "B8",
    "B9_R",
    "B10",
    "C1",
    "C2",
    "C3",
    "C4",
    "C5",
    "C6",
    "C7",
    "C8",
    "C9",
    "C10",
    "D1",
    "D2",
    "D3",
    "D4",
    "D5",
    "D6"
]

efa_data = df[
    VALIDATION_ITEMS
].dropna()

print(
    "\nComplete cases for factor analysis:",
    len(efa_data)
)

kmo_all, kmo_model = calculate_kmo(
    efa_data
)

bartlett_chi2, bartlett_p = (
    calculate_bartlett_sphericity(
        efa_data
    )
)

print("\nKMO:", round(kmo_model, 4))
print(
    "Bartlett chi-square:",
    round(bartlett_chi2, 3)
)
print(
    "Bartlett p-value:",
    bartlett_p
)


# ================================================================
# 47. PARALLEL ANALYSIS FOR NUMBER OF FACTORS
# ================================================================

def parallel_analysis(
    data,
    n_iter=500,
    random_state=42
):

    rng = np.random.default_rng(
        random_state
    )

    data = data.dropna()

    n, p = data.shape

    corr = np.corrcoef(
        data,
        rowvar=False
    )

    observed_eigenvalues = np.linalg.eigvalsh(
        corr
    )[::-1]

    random_eigenvalues = np.zeros(
        (n_iter, p)
    )

    for i in range(n_iter):

        random_data = rng.normal(
            size=(n, p)
        )

        random_corr = np.corrcoef(
            random_data,
            rowvar=False
        )

        random_eigenvalues[i, :] = (
            np.linalg.eigvalsh(
                random_corr
            )[::-1]
        )

    mean_random = (
        random_eigenvalues.mean(axis=0)
    )

    factors = int(
        np.sum(
            observed_eigenvalues >
            mean_random
        )
    )

    return (
        observed_eigenvalues,
        mean_random,
        factors
    )


observed_eigenvalues, random_eigenvalues, suggested_factors = (
    parallel_analysis(
        efa_data
    )
)

print(
    "\nParallel analysis suggests approximately",
    suggested_factors,
    "factor(s)."
)

parallel_table = pd.DataFrame({

    "Factor": np.arange(
        1,
        len(observed_eigenvalues) + 1
    ),

    "Observed_Eigenvalue":
        observed_eigenvalues,

    "Mean_Random_Eigenvalue":
        random_eigenvalues
})

display(
    parallel_table.head(10).round(3)
)


# ================================================================
# 48. EIGENVALUE PLOT
# ================================================================

plt.figure(figsize=(9, 6))

plt.plot(
    parallel_table["Factor"],
    parallel_table["Observed_Eigenvalue"],
    marker="o",
    label="Observed"
)

plt.plot(
    parallel_table["Factor"],
    parallel_table["Mean_Random_Eigenvalue"],
    marker="o",
    label="Random parallel analysis"
)

plt.axhline(
    1,
    linestyle="--"
)

plt.xlabel("Factor")
plt.ylabel("Eigenvalue")
plt.title(
    "Parallel analysis for exploratory factor analysis"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIG_DIR,
        "Figure_5_parallel_analysis.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 49. EXPLORATORY FACTOR ANALYSIS
# ================================================================

# The conceptual framework specifies:
# 1. Digital Connectivity
# 2. Academic Internet Use
# 3. Non Academic Internet Use
#
# Therefore, a 3-factor EFA is also estimated for theoretical
# comparability, regardless of the parallel-analysis result.

efa_3 = FactorAnalyzer(
    n_factors=3,
    rotation="oblimin",
    method="minres"
)

efa_3.fit(efa_data)

loadings = pd.DataFrame(
    efa_3.loadings_,
    index=VALIDATION_ITEMS,
    columns=[
        "Factor_1",
        "Factor_2",
        "Factor_3"
    ]
)

communalities = pd.DataFrame({
    "Item": VALIDATION_ITEMS,
    "Communality":
        efa_3.get_communalities()
})

print("\n3-factor EFA loadings")
display(
    loadings.round(3)
)

print("\nCommunalities")
display(
    communalities.round(3)
)

loadings.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_8_EFA_factor_loadings.csv"
    )
)

communalities.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_9_EFA_communalities.csv"
    ),
    index=False
)


# ================================================================
# 50. TOTAL VARIANCE EXPLAINED
# ================================================================

variance = efa_3.get_factor_variance()

variance_table = pd.DataFrame({
    "Factor": [
        "Factor 1",
        "Factor 2",
        "Factor 3"
    ],

    "SS_Loadings":
        variance[0],

    "Proportion_Variance":
        variance[1],

    "Cumulative_Variance":
        variance[2]
})

display(
    variance_table.round(3)
)

variance_table.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_10_EFA_variance_explained.csv"
    ),
    index=False
)


# ================================================================
# 51. CFA
# ================================================================

RUN_CFA = True

if RUN_CFA:

    cfa_data = df[
        VALIDATION_ITEMS
    ].dropna().copy()

    cfa_model_description = """

    DCI =~ B1 + B2 + B3 + B4 + B5 + B6 + B7 + B8 + B9_R + B10

    AIU =~ C1 + C2 + C3 + C4 + C5 + C6 + C7 + C8 + C9 + C10

    NAIU =~ D1 + D2 + D3 + D4 + D5 + D6

    DCI ~~ AIU
    DCI ~~ NAIU
    AIU ~~ NAIU

    """

    try:

        cfa_model = Model(
            cfa_model_description
        )

        cfa_model.fit(
            cfa_data
        )

        cfa_stats = calc_stats(
            cfa_model
        )

        print("\nCFA FIT STATISTICS")

        display(
            cfa_stats.T
        )

        cfa_parameters = cfa_model.inspect(
            std_est=True
        )

        print("\nCFA parameter estimates")

        display(
            cfa_parameters
        )

        cfa_parameters.to_csv(
            os.path.join(
                TABLE_DIR,
                "Table_11_CFA_parameters.csv"
            ),
            index=False
        )

        cfa_stats.T.to_csv(
            os.path.join(
                TABLE_DIR,
                "Table_12_CFA_fit_statistics.csv"
            )
        )

    except Exception as e:

        print(
            "\nCFA could not be completed automatically."
        )

        print(
            "Reason:",
            e
        )

        print(
            "Do not report CFA results unless the model "
            "successfully converges."
        )


# ================================================================
# 52. PREPARE REGRESSION DATA
# ================================================================

reg = model_df.copy()

reg = reg.rename(
    columns={
        "Age_clean": "Age",
        "Gender_clean": "Gender",
        "Study_Year_clean": "StudyYear",
        "Study_Level_clean": "StudyLevel",
        "Academic_Discipline_clean": "Discipline",
        "Residence_clean": "Residence",
        "Device_clean": "Device",
        "Daily_Internet_Use_clean": "DailyUse"
    }
)

print("\nRegression dataset:")
print(reg.shape)

display(reg.head())


# ================================================================
# 53. OBJECTIVE 2:
# MAIN RELATIONSHIPS
# ================================================================

formula_bivariate = """
CGPA ~ DCI + AIU + NAIU
"""

ols_bivariate = smf.ols(
    formula_bivariate,
    data=reg
).fit(
    cov_type="HC3"
)

print(
    ols_bivariate.summary()
)


# ================================================================
# 54. SENSITIVITY MODEL 1:
# DEMOGRAPHIC AND ACADEMIC CONTROLS
# ================================================================

formula_m1 = """
CGPA ~
C(Age) +
C(Gender) +
C(StudyYear) +
C(StudyLevel) +
C(Discipline) +
C(Residence) +
C(University) +
C(Device) +
C(DailyUse)
"""


# ================================================================
# 55. MODEL 2:
# CONTROLS + DIGITAL CONNECTIVITY
# ================================================================

formula_m2 = formula_m1 + " + DCI"


# ================================================================
# 56. MODEL 3:
# CONTROLS + DIGITAL CONNECTIVITY + ACADEMIC INTERNET USE
# ================================================================

formula_m3 = formula_m2 + " + AIU"


# ================================================================
# 57. MODEL 4:
# CONTROLS + DCI + AIU + NAIU
# ================================================================

formula_m4 = formula_m3 + " + NAIU"


# ================================================================
# 58. MODEL 5:
# FINAL MODEL
# ================================================================

formula_final = formula_m4


# ================================================================
# 59. FIT NESTED OLS MODELS
# ================================================================

model_formulas = {

    "Model 1 Controls only":
        formula_m1,

    "Model 2 + DCI":
        formula_m2,

    "Model 3 + AIU":
        formula_m3,

    "Model 4 + NAIU":
        formula_m4,

    "Model 5 Final":
        formula_final
}

ols_models = {}

for name, formula in model_formulas.items():

    model = smf.ols(
        formula,
        data=reg
    ).fit(
        cov_type="HC3"
    )

    ols_models[name] = model

    print("\n================================================")
    print(name)
    print("================================================")

    print(
        "N:",
        int(model.nobs)
    )

    print(
        "R-squared:",
        round(model.rsquared, 4)
    )

    print(
        "Adjusted R-squared:",
        round(model.rsquared_adj, 4)
    )

    print(
        "AIC:",
        round(model.aic, 3)
    )

    print(
        "BIC:",
        round(model.bic, 3)
    )


# ================================================================
# 60. MODEL COMPARISON TABLE
# ================================================================

model_comparison = []

previous_r2 = 0

for name, model in ols_models.items():

    model_comparison.append({

        "Model": name,

        "N": int(model.nobs),

        "R2": model.rsquared,

        "Adjusted_R2":
            model.rsquared_adj,

        "Delta_R2":
            model.rsquared - previous_r2,

        "AIC": model.aic,

        "BIC": model.bic,

        "RMSE":
            np.sqrt(
                np.mean(
                    model.resid ** 2
                )
            ),

        "MAE":
            np.mean(
                np.abs(
                    model.resid
                )
            )
    })

    previous_r2 = model.rsquared


model_comparison = pd.DataFrame(
    model_comparison
)

display(
    model_comparison.round(4)
)

model_comparison.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_13_nested_model_comparison.csv"
    ),
    index=False
)


# ================================================================
# 61. FINAL OLS COEFFICIENT TABLE
# ================================================================

final_ols = ols_models["Model 5 Final"]

final_coefficients = pd.DataFrame({

    "Coefficient":
        final_ols.params,

    "SE":
        final_ols.bse,

    "t":
        final_ols.tvalues,

    "p":
        final_ols.pvalues,

    "CI_lower":
        final_ols.conf_int()[0],

    "CI_upper":
        final_ols.conf_int()[1]
})

display(
    final_coefficients.round(4)
)

final_coefficients.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_14_final_OLS_coefficients.csv"
    )
)


# ================================================================
# 62. CHECK MULTICOLLINEARITY
# ================================================================

# Build design matrix from the final formula
y, X = smf.ols(
    formula_final,
    data=reg
).endog, smf.ols(
    formula_final,
    data=reg
).exog

design_info = smf.ols(
    formula_final,
    data=reg
).data.design_info

X_df = pd.DataFrame(
    X,
    columns=design_info.column_names
)

vif_table = []

for i, col in enumerate(
    X_df.columns
):

    if col == "Intercept":
        continue

    vif_table.append({

        "Variable": col,

        "VIF":
            variance_inflation_factor(
                X_df.values,
                i
            )
    })

vif_table = pd.DataFrame(
    vif_table
)

display(
    vif_table.sort_values(
        "VIF",
        ascending=False
    ).round(3)
)

vif_table.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_15_VIF.csv"
    ),
    index=False
)


# ================================================================
# 63. REGRESSION DIAGNOSTICS
# ================================================================
# Import the required diagnostic tests
from statsmodels.stats.diagnostic import (
    het_breuschpagan,
    het_white
)

residuals = final_ols.resid
fitted = final_ols.fittedvalues

# Breusch-Pagan
bp = het_breuschpagan(
    residuals,
    final_ols.model.exog
)

bp_table = pd.DataFrame({
    "Test": [
        "Breusch-Pagan LM",
        "Breusch-Pagan p-value",
        "F-statistic",
        "F p-value"
    ],

    "Value": [
        bp[0],
        bp[1],
        bp[2],
        bp[3]
    ]
})

display(bp_table)


# White test
white = het_white(
    residuals,
    final_ols.model.exog
)

white_table = pd.DataFrame({
    "Test": [
        "White LM",
        "White p-value",
        "White F",
        "White F p-value"
    ],

    "Value": [
        white[0],
        white[1],
        white[2],
        white[3]
    ]
})

display(white_table)


# Jarque-Bera
jb = stats.jarque_bera(
    residuals
)

print(
    "\nJarque-Bera statistic:",
    jb.statistic
)

print(
    "Jarque-Bera p-value:",
    jb.pvalue
)

bp_table.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_16_Breusch_Pagan.csv"
    ),
    index=False
)

white_table.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_17_White_test.csv"
    ),
    index=False
)


# ================================================================
# 64. RESIDUAL DIAGNOSTIC PLOTS
# ================================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x=fitted,
    y=residuals
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("Fitted values")
plt.ylabel("Residuals")
plt.title(
    "Residuals versus fitted values"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIG_DIR,
        "Figure_6_residuals_fitted.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# Q-Q plot
plt.figure(figsize=(8, 6))

sm.qqplot(
    residuals,
    line="45",
    fit=True
)

plt.title(
    "Normal Q-Q plot of regression residuals"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIG_DIR,
        "Figure_7_QQ_plot.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 65. INFLUENTIAL OBSERVATIONS
# ================================================================

influence = OLSInfluence(
    final_ols
)

cooks_d = influence.cooks_distance[0]
leverage = influence.hat_matrix_diag
studentized = influence.resid_studentized_external

influence_df = reg.copy()

influence_df["Cooks_D"] = cooks_d
influence_df["Leverage"] = leverage
influence_df["Studentized_Residual"] = studentized

cook_threshold = 4 / len(reg)

influence_df["Influential"] = (
    influence_df["Cooks_D"] >
    cook_threshold
)

print(
    "\nCook's distance threshold:",
    cook_threshold
)

print(
    "Potentially influential observations:",
    influence_df["Influential"].sum()
)

display(
    influence_df.loc[
        influence_df["Influential"]
    ].sort_values(
        "Cooks_D",
        ascending=False
    ).head(20)
)

influence_df.to_csv(
    os.path.join(
        AUDIT_DIR,
        "influential_observations.csv"
    ),
    index=False
)


# ================================================================
# 66. INFLUENCE PLOT
# ================================================================

plt.figure(figsize=(10, 6))

plt.stem(
    np.arange(len(cooks_d)),
    cooks_d,
    markerfmt=","
)

plt.axhline(
    cook_threshold,
    linestyle="--"
)

plt.xlabel("Observation")
plt.ylabel("Cook's distance")
plt.title(
    "Influence diagnostics using Cook's distance"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIG_DIR,
        "Figure_8_cooks_distance.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 67. SENSITIVITY ANALYSIS:
# REMOVE POTENTIALLY INFLUENTIAL CASES
# ================================================================

sensitivity_df = reg.loc[
    ~influence_df["Influential"]
].copy()

print(
    "\nOriginal regression N:",
    len(reg)
)

print(
    "N after excluding Cook's-D influential cases:",
    len(sensitivity_df)
)

sensitivity_model = smf.ols(
    formula_final,
    data=sensitivity_df
).fit(
    cov_type="HC3"
)

sensitivity_coefficients = pd.DataFrame({

    "Original_beta":
        final_ols.params,

    "Sensitivity_beta":
        sensitivity_model.params,

    "Original_p":
        final_ols.pvalues,

    "Sensitivity_p":
        sensitivity_model.pvalues
})

display(
    sensitivity_coefficients.round(4)
)

sensitivity_coefficients.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_18_influence_sensitivity.csv"
    )
)


# ================================================================
# 68. OBJECTIVE 3:
# ASSESS INSTITUTIONAL CLUSTERING
# ================================================================

# Null random-intercept model
null_mixed_formula = """
CGPA ~ 1
"""

try:

    null_mixed = smf.mixedlm(
        null_mixed_formula,
        reg,
        groups=reg["University"]
    ).fit(
        reml=True,
        method="lbfgs"
    )

    var_between = float(
        null_mixed.cov_re.iloc[0, 0]
    )

    var_within = float(
        null_mixed.scale
    )

    ICC = (
        var_between /
        (var_between + var_within)
    )

    print(
        "\nBetween-university variance:",
        var_between
    )

    print(
        "Within-university variance:",
        var_within
    )

    print(
        "Intraclass correlation coefficient (ICC):",
        round(ICC, 4)
    )

except Exception as e:

    print(
        "\nNull mixed model failed:",
        e
    )

    ICC = np.nan


# ================================================================
# 69. FINAL MULTILEVEL MODEL
# ================================================================

mixed_formula = """
CGPA ~
DCI +
AIU +
NAIU +
C(Age) +
C(Gender) +
C(StudyYear) +
C(StudyLevel) +
C(Discipline) +
C(Residence) +
C(Device) +
C(DailyUse)
"""


try:

    mixed_model_ml = smf.mixedlm(
        mixed_formula,
        reg,
        groups=reg["University"]
    ).fit(
        reml=False,
        method="lbfgs"
    )

    print(
        "\n================================================"
    )
    print(
        "MULTILEVEL RANDOM-INTERCEPT MODEL"
    )
    print(
        "================================================"
    )

    print(
        mixed_model_ml.summary()
    )

except Exception as e:

    mixed_model_ml = None

    print(
        "\nMultilevel model failed:"
    )

    print(e)


# ================================================================
# 70. REML FINAL MULTILEVEL MODEL
# ================================================================

if mixed_model_ml is not None:

    try:

        mixed_model_reml = smf.mixedlm(
            mixed_formula,
            reg,
            groups=reg["University"]
        ).fit(
            reml=True,
            method="lbfgs"
        )

        print(
            "\nREML estimates:"
        )

        print(
            mixed_model_reml.summary()
        )

    except Exception as e:

        mixed_model_reml = None

        print(
            "\nREML estimation failed:"
        )

        print(e)


# ================================================================
# 71. MULTILEVEL MODEL COEFFICIENT TABLE
# ================================================================

if mixed_model_ml is not None:

    mixed_coef = pd.DataFrame({

        "Coefficient":
            mixed_model_ml.params,

        "SE":
            mixed_model_ml.bse,

        "z":
            mixed_model_ml.tvalues,

        "p":
            mixed_model_ml.pvalues,

        "CI_lower":
            mixed_model_ml.conf_int()[0],

        "CI_upper":
            mixed_model_ml.conf_int()[1]
    })

    display(
        mixed_coef.round(4)
    )

    mixed_coef.to_csv(
        os.path.join(
            TABLE_DIR,
            "Table_19_multilevel_model.csv"
        )
    )


# ================================================================
# 72. COMPARE OLS AND MULTILEVEL RESULTS
# ================================================================

main_predictors = [
    "DCI",
    "AIU",
    "NAIU"
]

comparison_rows = []

for variable in main_predictors:

    row = {

        "Variable": variable,

        "OLS_beta":
            final_ols.params.get(
                variable,
                np.nan
            ),

        "OLS_p":
            final_ols.pvalues.get(
                variable,
                np.nan
            )
    }

    if mixed_model_ml is not None:

        row["Mixed_beta"] = (
            mixed_model_ml.params.get(
                variable,
                np.nan
            )
        )

        row["Mixed_p"] = (
            mixed_model_ml.pvalues.get(
                variable,
                np.nan
            )
        )

    else:

        row["Mixed_beta"] = np.nan
        row["Mixed_p"] = np.nan

    comparison_rows.append(row)


ols_mixed_comparison = pd.DataFrame(
    comparison_rows
)

display(
    ols_mixed_comparison.round(4)
)

ols_mixed_comparison.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_20_OLS_vs_multilevel.csv"
    ),
    index=False
)


# ================================================================
# 73. CLUSTER ROBUST OLS
# ================================================================

cluster_robust_ols = smf.ols(
    formula_final,
    data=reg
).fit(
    cov_type="cluster",
    cov_kwds={
        "groups":
            reg["University"]
    }
)

cluster_coef = pd.DataFrame({

    "Coefficient":
        cluster_robust_ols.params,

    "Cluster_Robust_SE":
        cluster_robust_ols.bse,

    "t":
        cluster_robust_ols.tvalues,

    "p":
        cluster_robust_ols.pvalues
})

print(
    "\nCluster-robust OLS results:"
)

display(
    cluster_coef.round(4)
)

cluster_coef.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_21_cluster_robust_OLS.csv"
    )
)


# ================================================================
# 74. GROUP LEVEL DESCRIPTIVES
# ================================================================

university_summary = (
    analysis_df
    .groupby("University")
    .agg(
        N=("CGPA", "count"),
        Mean_CGPA=("CGPA", "mean"),
        SD_CGPA=("CGPA", "std"),
        Mean_DCI=("DCI", "mean"),
        Mean_AIU=("AIU", "mean"),
        Mean_NAIU=("NAIU", "mean")
    )
    .reset_index()
    .sort_values(
        "N",
        ascending=False
    )
)

display(
    university_summary.round(3)
)

university_summary.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_22_university_level_summary.csv"
    ),
    index=False
)


# ================================================================
# 75. MODEL PREDICTION PERFORMANCE ON FULL SAMPLE
# ================================================================

reg["OLS_Predicted_CGPA"] = (
    final_ols.fittedvalues
)

reg["OLS_Residual"] = (
    final_ols.resid
)

full_sample_metrics = pd.DataFrame({

    "Metric": [
        "R2",
        "Adjusted R2",
        "RMSE",
        "MAE"
    ],

    "Value": [

        r2_score(
            reg["CGPA"],
            reg["OLS_Predicted_CGPA"]
        ),

        final_ols.rsquared_adj,

        np.sqrt(
            mean_squared_error(
                reg["CGPA"],
                reg["OLS_Predicted_CGPA"]
            )
        ),

        mean_absolute_error(
            reg["CGPA"],
            reg["OLS_Predicted_CGPA"]
        )
    ]
})

display(
    full_sample_metrics.round(4)
)


# ================================================================
# 76. GROUP-AWARE CROSS-VALIDATED PREDICTION
# ================================================================

# This is important because students from the same university
# should not simply appear in both training and test folds.

predictor_columns = [
    "DCI",
    "AIU",
    "NAIU",
    "Age",
    "Gender",
    "StudyYear",
    "StudyLevel",
    "Discipline",
    "Residence",
    "University",
    "Device",
    "DailyUse"
]

X_pred = reg[predictor_columns].copy()
y_pred = reg["CGPA"].copy()
groups = reg["University"].copy()

numeric_features = [
    "DCI",
    "AIU",
    "NAIU"
]

categorical_features = [
    "Age",
    "Gender",
    "StudyYear",
    "StudyLevel",
    "Discipline",
    "Residence",
    "University",
    "Device",
    "DailyUse"
]

preprocessor = ColumnTransformer(

    transformers=[

        (
            "numeric",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="median"
                    )
                )
            ]),
            numeric_features
        ),

        (
            "categorical",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="most_frequent"
                    )
                ),

                (
                    "onehot",
                    OneHotEncoder(
                        handle_unknown="ignore"
                    )
                )
            ]),
            categorical_features
        )
    ]
)


prediction_pipeline = Pipeline([

    (
        "preprocessor",
        preprocessor
    ),

    (
        "model",
        LinearRegression()
    )
])


n_groups = groups.nunique()

n_splits = min(
    5,
    n_groups
)

group_kfold = GroupKFold(
    n_splits=n_splits
)


cv_results = []

for fold, (train_idx, test_idx) in enumerate(

    group_kfold.split(
        X_pred,
        y_pred,
        groups=groups
    ),

    start=1
):

    X_train = X_pred.iloc[
        train_idx
    ]

    X_test = X_pred.iloc[
        test_idx
    ]

    y_train = y_pred.iloc[
        train_idx
    ]

    y_test = y_pred.iloc[
        test_idx
    ]

    prediction_pipeline.fit(
        X_train,
        y_train
    )

    y_hat = prediction_pipeline.predict(
        X_test
    )

    cv_results.append({

        "Fold": fold,

        "N_test": len(test_idx),

        "R2":
            r2_score(
                y_test,
                y_hat
            ),

        "RMSE":
            np.sqrt(
                mean_squared_error(
                    y_test,
                    y_hat
                )
            ),

        "MAE":
            mean_absolute_error(
                y_test,
                y_hat
            )
    })


cv_results = pd.DataFrame(
    cv_results
)

print(
    "\nGroup-aware cross-validation results:"
)

display(
    cv_results.round(4)
)

cv_summary = pd.DataFrame({

    "Metric": [
        "Mean R2",
        "SD R2",
        "Mean RMSE",
        "SD RMSE",
        "Mean MAE",
        "SD MAE"
    ],

    "Value": [

        cv_results["R2"].mean(),
        cv_results["R2"].std(),

        cv_results["RMSE"].mean(),
        cv_results["RMSE"].std(),

        cv_results["MAE"].mean(),
        cv_results["MAE"].std()
    ]
})

display(
    cv_summary.round(4)
)

cv_results.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_23_group_aware_cross_validation.csv"
    ),
    index=False
)

cv_summary.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_24_cross_validation_summary.csv"
    ),
    index=False
)


# ================================================================
# 77. OPTIONAL RIDGE PREDICTIVE SENSITIVITY MODEL
# ================================================================

ridge_pipeline = Pipeline([

    (
        "preprocessor",
        preprocessor
    ),

    (
        "model",
        Ridge(alpha=1.0)
    )
])


ridge_cv_results = []

for fold, (train_idx, test_idx) in enumerate(

    group_kfold.split(
        X_pred,
        y_pred,
        groups=groups
    ),

    start=1
):

    ridge_pipeline.fit(
        X_pred.iloc[train_idx],
        y_pred.iloc[train_idx]
    )

    y_hat = ridge_pipeline.predict(
        X_pred.iloc[test_idx]
    )

    ridge_cv_results.append({

        "Fold": fold,

        "R2":
            r2_score(
                y_pred.iloc[test_idx],
                y_hat
            ),

        "RMSE":
            np.sqrt(
                mean_squared_error(
                    y_pred.iloc[test_idx],
                    y_hat
                )
            ),

        "MAE":
            mean_absolute_error(
                y_pred.iloc[test_idx],
                y_hat
            )
    })


ridge_cv_results = pd.DataFrame(
    ridge_cv_results
)

print(
    "\nRidge group-aware cross-validation:"
)

display(
    ridge_cv_results.round(4)
)


# ================================================================
# 78. MODEL PERFORMANCE SUMMARY
# ================================================================

predictive_comparison = pd.DataFrame({

    "Model": [
        "OLS",
        "Ridge"
    ],

    "Mean_CV_R2": [
        cv_results["R2"].mean(),
        ridge_cv_results["R2"].mean()
    ],

    "Mean_CV_RMSE": [
        cv_results["RMSE"].mean(),
        ridge_cv_results["RMSE"].mean()
    ],

    "Mean_CV_MAE": [
        cv_results["MAE"].mean(),
        ridge_cv_results["MAE"].mean()
    ]
})

display(
    predictive_comparison.round(4)
)

predictive_comparison.to_csv(
    os.path.join(
        TABLE_DIR,
        "Table_25_predictive_model_comparison.csv"
    ),
    index=False
)


# ================================================================
# 79. EXPORT ALL IMPORTANT TABLES TO ONE EXCEL WORKBOOK
# ================================================================

excel_output = os.path.join(
    RESULT_DIR,
    "Complete_Statistical_Analysis_Results.xlsx"
)

with pd.ExcelWriter(
    excel_output,
    engine="xlsxwriter"
) as writer:

    demographic_table.to_excel(
        writer,
        sheet_name="Demographics",
        index=False
    )

    descriptive_table.to_excel(
        writer,
        sheet_name="Descriptives",
        index=False
    )

    dimension_table.to_excel(
        writer,
        sheet_name="DCI_Dimensions",
        index=False
    )

    dci_level_table.to_excel(
        writer,
        sheet_name="DCI_Level",
        index=False
    )

    reliability_summary.to_excel(
        writer,
        sheet_name="Reliability",
        index=False
    )

    item_reliability.to_excel(
        writer,
        sheet_name="Item_Reliability",
        index=False
    )

    loadings.to_excel(
        writer,
        sheet_name="EFA_Loadings"
    )

    communalities.to_excel(
        writer,
        sheet_name="EFA_Communalities",
        index=False
    )

    variance_table.to_excel(
        writer,
        sheet_name="EFA_Variance",
        index=False
    )

    model_comparison.to_excel(
        writer,
        sheet_name="Model_Comparison",
        index=False
    )

    final_coefficients.to_excel(
        writer,
        sheet_name="Final_OLS"
    )

    vif_table.to_excel(
        writer,
        sheet_name="VIF",
        index=False
    )

    bp_table.to_excel(
        writer,
        sheet_name="Breusch_Pagan",
        index=False
    )

    white_table.to_excel(
        writer,
        sheet_name="White_Test",
        index=False
    )

    influence_df.to_excel(
        writer,
        sheet_name="Influence",
        index=False
    )

    sensitivity_coefficients.to_excel(
        writer,
        sheet_name="Sensitivity",
        index=False
    )

    university_summary.to_excel(
        writer,
        sheet_name="University_Summary",
        index=False
    )

    ols_mixed_comparison.to_excel(
        writer,
        sheet_name="OLS_Mixed",
        index=False
    )

    cluster_coef.to_excel(
        writer,
        sheet_name="Cluster_Robust",
        index=False
    )

    cv_results.to_excel(
        writer,
        sheet_name="CV_Folds",
        index=False
    )

    cv_summary.to_excel(
        writer,
        sheet_name="CV_Summary",
        index=False
    )

    predictive_comparison.to_excel(
        writer,
        sheet_name="Prediction_Comparison",
        index=False
    )

    university_audit.to_excel(
        writer,
        sheet_name="University_Audit",
        index=False
    )

    cleaning_summary.to_excel(
        writer,
        sheet_name="Cleaning_Summary",
        index=False
    )


print(
    "\nExcel results workbook created:"
)

print(excel_output)


# ================================================================
# 80. CREATE ANALYSIS LOG
# ================================================================

analysis_log = {

    "Study":
        "Digital Connectivity and Internet Use as Predictors of University Students' Academic Performance: Multisite Evidence from Uganda",

    "Initial_N":
        int(initial_n),

    "Final_complete_case_N":
        int(len(reg)),

    "Number_of_universities":
        int(reg["University"].nunique()),

    "Invalid_Likert_observations":
        int(len(invalid_likert_df)),

    "Invalid_CGPA_observations":
        int(len(invalid_cgpa)),

    "Exact_duplicate_rows":
        int(duplicate_rows),

    "B9_reverse_coded":
        True,

    "DCI_items":
        DCI_ITEMS,

    "AIU_items":
        C_ITEMS,

    "NAIU_items":
        D_ITEMS,

    "Regression_model":
        "CGPA ~ DCI + AIU + NAIU + demographic, academic and institutional controls",

    "Institutional_clustering_assessed":
        True,

    "ICC":
        None if pd.isna(ICC) else float(ICC),

    "Sampling_note":
        "Non-probability availability/convenience-based multisite recruitment. Enumerators were permitted to administer questionnaires at universities within their geographical reach because the study was self-sponsored and transport costs were substantial. The dataset does not establish probability sampling.",

    "Causal_claims":
        "Not appropriate because the design is cross-sectional."
}

with open(
    os.path.join(
        AUDIT_DIR,
        "analysis_log.json"
    ),
    "w"
) as f:

    json.dump(
        analysis_log,
        f,
        indent=4
    )


# ================================================================
# 81. FINAL AUTOMATED REPORT
# ================================================================

print("\n")
print("=" * 80)
print("FINAL ANALYSIS SUMMARY")
print("=" * 80)

print(
    f"Initial sample: N = {initial_n}"
)

print(
    f"Complete cases for final regression: N = {len(reg)}"
)

print(
    f"Universities represented after harmonisation: "
    f"{reg['University'].nunique()}"
)

print(
    f"Invalid Likert observations corrected: "
    f"{len(invalid_likert_df)}"
)

print(
    f"Invalid CGPA observations corrected: "
    f"{len(invalid_cgpa)}"
)

print(
    f"Digital Connectivity mean: "
    f"{reg['DCI'].mean():.3f}"
)

print(
    f"Academic Internet Use mean: "
    f"{reg['AIU'].mean():.3f}"
)

print(
    f"Non Academic Internet Use mean: "
    f"{reg['NAIU'].mean():.3f}"
)

print(
    f"CGPA mean: "
    f"{reg['CGPA'].mean():.3f}"
)

if not pd.isna(ICC):

    print(
        f"University ICC: {ICC:.4f}"
    )

print(
    f"Final OLS R2: "
    f"{final_ols.rsquared:.4f}"
)

print(
    f"Final OLS adjusted R2: "
    f"{final_ols.rsquared_adj:.4f}"
)

print(
    f"Final OLS RMSE: "
    f"{np.sqrt(mean_squared_error(reg['CGPA'], final_ols.fittedvalues)):.4f}"
)

print(
    f"Group-aware CV R2: "
    f"{cv_results['R2'].mean():.4f}"
)

print(
    f"Group-aware CV RMSE: "
    f"{cv_results['RMSE'].mean():.4f}"
)

print(
    f"Group-aware CV MAE: "
    f"{cv_results['MAE'].mean():.4f}"
)

print("\nResults directory:")
print(RESULT_DIR)

print("\nExcel workbook:")
print(excel_output)

print("\nAnalysis completed.")
print("=" * 80)
```
