```python
# ================================================================
# STUDY:
# ANALYZING THE IMPACT OF TRANSACTION TAXES ON SAVINGS AND
# INVESTMENT IN NIGERIA: A BOX-JENKINS APPROACH
#
# COMPLETE GOOGLE COLAB ANALYTICAL PIPELINE
# DATA PERIOD: 2000-2024
# FORECAST PERIOD: 2025-2030
# ================================================================


# ================================================================
# STEP 1: INSTALL REQUIRED PACKAGES
# ================================================================

!pip -q install statsmodels openpyxl scipy scikit-learn


# ================================================================
# STEP 2: IMPORT LIBRARIES
# ================================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from scipy.stats import pearsonr

from statsmodels.tsa.stattools import adfuller, kpss, acf, pacf
from statsmodels.tsa.ar_model import AutoReg
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import ccf

from statsmodels.stats.diagnostic import acorr_ljungbox, het_arch
from statsmodels.stats.stattools import jarque_bera

from sklearn.metrics import mean_absolute_error, mean_squared_error

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)

np.random.seed(123)


# ================================================================
# STEP 3: UPLOAD EXCEL DATASET
# ================================================================

print("Upload your Excel dataset containing annual transaction taxes,")
print("savings, investment, exchange rate, interest rate and inflation.")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

data = pd.read_excel(file_name)

print("\nOriginal dataset:")
display(data.head())

print("\nDataset dimensions:", data.shape)

print("\nOriginal variable names:")
print(data.columns.tolist())


# ================================================================
# STEP 4: DATA PREPROCESSING
# ================================================================

# Remove unwanted spaces from column names

data.columns = data.columns.str.strip()

# Standardize column names

data.columns = [
    col.replace(" ", "_").replace("-", "_")
    for col in data.columns
]

# Remove duplicate observations

data = data.drop_duplicates()

# Expected variables

required_columns = [
    "Year",
    "Investment",
    "Exchange_Rate",
    "VAT",
    "Interest_Rate",
    "Inflation",
    "Savings"
]

# Check whether all required variables are available

missing_columns = [
    col for col in required_columns
    if col not in data.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}. "
        f"Available columns: {data.columns.tolist()}"
    )

# Retain relevant variables

data = data[required_columns].copy()

# Convert variables to numeric

for col in required_columns:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )

# Remove missing observations

data = data.dropna()

# Sort chronologically

data = data.sort_values("Year")

# Ensure year is integer

data["Year"] = data["Year"].astype(int)

# Check duplicate years

if data["Year"].duplicated().any():
    raise ValueError(
        "Duplicate years detected. Resolve duplicate annual observations."
    )

# Set year as index

data = data.set_index("Year")

# Check annual frequency

expected_years = np.arange(
    data.index.min(),
    data.index.max() + 1
)

if not np.array_equal(
    data.index.to_numpy(),
    expected_years
):
    raise ValueError(
        "Missing years detected. The annual time series must "
        "be checked before estimation."
    )

print("\nCleaned dataset:")
display(data.head())

print("\nSample period:", data.index.min(), "-", data.index.max())

print("\nNumber of observations:", len(data))

print("\nMissing observations:")
print(data.isnull().sum())


# ================================================================
# STEP 5: DESCRIPTIVE STATISTICS
# OBJECTIVE 1
# ================================================================

print("\nDESCRIPTIVE STATISTICS")

descriptive = data.describe().T

descriptive["Skewness"] = data.skew()

descriptive["Kurtosis"] = data.kurtosis()

display(descriptive.round(4))

descriptive.to_excel("Descriptive_Statistics.xlsx")


# ================================================================
# STEP 6: TREND AND PATTERN ANALYSIS
# OBJECTIVE 1
# ================================================================

variables = [
    "VAT",
    "Savings",
    "Investment",
    "Exchange_Rate",
    "Interest_Rate",
    "Inflation"
]

# Individual time series plots

for variable in variables:

    plt.figure(figsize=(12, 5))

    plt.plot(
        data.index,
        data[variable],
        marker="o",
        linewidth=2
    )

    plt.title(
        f"Annual Trend of {variable} in Nigeria",
        fontsize=14
    )

    plt.xlabel("Year")

    plt.ylabel(variable)

    plt.grid(alpha=0.3)

    plt.tight_layout()

    plt.show()


# ================================================================
# STEP 7: PEARSON CORRELATION ANALYSIS
# ================================================================

print("\nPEARSON CORRELATION MATRIX")

correlation = data.corr(method="pearson")

display(correlation.round(4))

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".3f"
)

plt.title("Correlation Matrix of Study Variables")

plt.tight_layout()

plt.show()

correlation.to_excel("Correlation_Matrix.xlsx")


# ================================================================
# STEP 8: STATIONARITY TESTS
# ADF AND KPSS
# ================================================================

def stationarity_tests(series, name):

    series = series.dropna()

    # ADF test

    adf_result = adfuller(
        series,
        regression="c",
        autolag="AIC"
    )

    # KPSS test

    try:

        kpss_result = kpss(
            series,
            regression="c",
            nlags="auto"
        )

        kpss_stat = kpss_result[0]

        kpss_p = kpss_result[1]

    except Exception:

        kpss_stat = np.nan

        kpss_p = np.nan

    return {

         "Variable": name,

         "ADF Statistic": adf_result[0],

         "ADF p-value": adf_result[1],

         "ADF Lags": adf_result[2],

         "KPSS Statistic": kpss_stat,

         "KPSS p-value": kpss_p

    }


print("\nSTATIONARITY TESTS AT LEVEL")

stationarity_level = pd.DataFrame([
    stationarity_tests(data[col], col)
    for col in variables
])

display(stationarity_level.round(4))


# First differences

diff_data = data[variables].diff().dropna()

print("\nSTATIONARITY TESTS AT FIRST DIFFERENCE")

stationarity_diff = pd.DataFrame([
    stationarity_tests(diff_data[col], col)
    for col in variables
])

display(stationarity_diff.round(4))


# Export stationarity results

with pd.ExcelWriter("Stationarity_Tests.xlsx") as writer:

    stationarity_level.to_excel(
        writer,
        sheet_name="Level",
        index=False
    )

    stationarity_diff.to_excel(
        writer,
        sheet_name="First_Difference",
        index=False
    )


# ================================================================
# STEP 9: ACF AND PACF ANALYSIS
# ================================================================

for variable in variables:

    series = data[variable].dropna()

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(14, 4)
    )

    acf(
        series,
        nlags=8,
        fft=False
    )

    from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

    plot_acf(
        series,
        lags=8,
        ax=axes[0]
    )

    plot_pacf(
        series,
        lags=8,
        ax=axes[1],
        method="ywm"
    )

    axes[0].set_title(f"ACF: {variable}")

    axes[1].set_title(f"PACF: {variable}")

    plt.tight_layout()

    plt.show()


# ================================================================
# STEP 10: CROSS CORRELATION ANALYSIS
# VAT, SAVINGS AND INVESTMENT
# ================================================================

# First difference to reduce spurious correlation
# arising from common trends.

vat_diff = data["VAT"].diff()

investment_diff = data["Investment"].diff()

savings_diff = data["Savings"].diff()


def cross_correlation_analysis(x, y, title):

    x = x.dropna()

    y = y.dropna()

    joined = pd.concat(
        [x, y],
        axis=1
    ).dropna()

    x = joined.iloc[:, 0]

    y = joined.iloc[:, 1]

    # Standardize both variables

    x = (x - x.mean()) / x.std()

    y = (y - y.mean()) / y.std()

    # Positive lag means VAT precedes the outcome

    lags = range(0, 5)

    correlations = []

    for lag in lags:

        if lag == 0:

            r = np.corrcoef(
                x,
                y
            )[0, 1]

        else:

            r = np.corrcoef(
                x.iloc[:-lag],
                y.iloc[lag:]
            )[0, 1]

        correlations.append(r)

    plt.figure(figsize=(10, 5))

    plt.bar(
        list(lags),
        correlations
    )

    plt.axhline(
        0,
        color="black"
    )

    plt.title(title)

    plt.xlabel("VAT Lead Lag")

    plt.ylabel("Correlation")

    plt.xticks(list(lags))

    plt.grid(alpha=0.2)

    plt.show()

    return pd.DataFrame({

        "Lag": list(lags),

        "Cross_Correlation": correlations

    })


ccf_investment = cross_correlation_analysis(
    vat_diff,
    investment_diff,
    "Cross Correlation: VAT and Investment"
)

ccf_savings = cross_correlation_analysis(
    vat_diff,
    savings_diff,
    "Cross Correlation: VAT and Savings"
)

print("\nVAT AND INVESTMENT CROSS CORRELATION")

display(ccf_investment)

print("\nVAT AND SAVINGS CROSS CORRELATION")

display(ccf_savings)


# ================================================================
# STEP 11: ARIMAX MODEL ESTIMATION
# OBJECTIVES 2 AND 3
# ================================================================

# Model structure:
#
# Savings_t = ARIMA errors + VAT + Exchange Rate
#             + Interest Rate + Inflation
#
# Investment_t = ARIMA errors + VAT + Exchange Rate
#               + Interest Rate + Inflation
#
# ARIMAX is estimated using SARIMAX with seasonal order zero.


exog_variables = [
    "VAT",
    "Exchange_Rate",
    "Interest_Rate",
    "Inflation"
]

exog = data[exog_variables].copy()

# Standardize explanatory variables to improve numerical stability.
# Coefficients will be interpreted per one standard deviation
# increase in each explanatory variable.

exog_mean = exog.mean()

exog_std = exog.std(ddof=0)

exog_std[exog_std == 0] = 1

exog_scaled = (
    exog - exog_mean
) / exog_std


# ================================================================
# STEP 12: AUTOMATIC PARSIMONIOUS ARIMAX SELECTION
# ================================================================

def select_arimax(y, X, max_p=2, max_q=2, max_d=1):

    results = []

    best_model = None

    best_aic = np.inf

    for d in range(max_d + 1):

        for p in range(max_p + 1):

            for q in range(max_q + 1):

                # Avoid the unnecessary ARIMA(0,0,0)
                # with no autoregressive or moving average term.

                try:

                    model = SARIMAX(

                        y,

                        exog=X,

                        order=(p, d, q),

                        trend="c",

                        enforce_stationarity=False,

                        enforce_invertibility=False

                    )

                    fitted = model.fit(
                        disp=False,
                        maxiter=500
                    )

                    results.append({

                        "p": p,

                        "d": d,

                        "q": q,

                        "AIC": fitted.aic,

                        "BIC": fitted.bic,

                        "Converged": fitted.mle_retvals.get(
                            "converged",
                            True
                        )

                    })

                    # Only converged models are considered.

                    if fitted.mle_retvals.get(
                        "converged",
                        True
                    ) and np.isfinite(fitted.aic):

                        if fitted.aic < best_aic:

                            best_aic = fitted.aic

                            best_model = fitted

                except Exception:

                    continue

    results_df = pd.DataFrame(results)

    if best_model is None:

        raise RuntimeError(
            "No ARIMAX model converged. "
            "Inspect the data and reduce model complexity."
        )

    return best_model, results_df


# ================================================================
# STEP 13: ESTIMATE SAVINGS MODEL
# ================================================================

print("\nESTIMATING ARIMAX SAVINGS MODEL")

savings_model, savings_selection = select_arimax(

    y=data["Savings"],

    X=exog_scaled,

    max_p=2,

    max_q=2,

    max_d=1

)

print("\nSelected Savings ARIMAX Model")

print(savings_model.summary())

print("\nSavings Model Selection Results")

display(
    savings_selection.sort_values("AIC").head(10)
)


# ================================================================
# STEP 14: ESTIMATE INVESTMENT MODEL
# ================================================================

print("\nESTIMATING ARIMAX INVESTMENT MODEL")

investment_model, investment_selection = select_arimax(

    y=data["Investment"],

    X=exog_scaled,

    max_p=2,

    max_q=2,

    max_d=1

)

print("\nSelected Investment ARIMAX Model")

print(investment_model.summary())

print("\nInvestment Model Selection Results")

display(
    investment_selection.sort_values("AIC").head(10)
)


# ================================================================
# STEP 15: VAT EFFECT AND STATISTICAL SIGNIFICANCE
# OBJECTIVES 2 AND 3
# ================================================================

def vat_effect(model, variable_name, scaling_factor):

    params = model.params

    pvalues = model.pvalues

    vat_coef = params["VAT"]

    vat_se = model.bse["VAT"]

    vat_p = pvalues["VAT"]

    # Convert standardized VAT coefficient into original scale.

    original_coef = vat_coef / scaling_factor

    original_se = vat_se / scaling_factor

    z_value = vat_coef / vat_se

    return {

        "Dependent Variable": variable_name,

        "VAT Coefficient": original_coef,

        "Standard Error": original_se,

        "Z Statistic": z_value,

        "P Value": vat_p,

        "Significant at 5%": vat_p < 0.05

    }


vat_savings = vat_effect(

    savings_model,

    "Savings",

    exog_std["VAT"]

)

vat_investment = vat_effect(

    investment_model,

    "Investment",

    exog_std["VAT"]

)

vat_results = pd.DataFrame([
    vat_savings,
    vat_investment
])

print("\nVAT EFFECTS ON SAVINGS AND INVESTMENT")

display(vat_results.round(6))


# ================================================================
# STEP 16: DYNAMIC AND LONG-RUN EFFECTS
# OBJECTIVE 4
# ================================================================

def dynamic_effects(model, dependent_name, vat_scale):

    params = model.params

    # Obtain the coefficient on the dependent variable's
    # first autoregressive lag, if present.

    ar_terms = [
        name for name in params.index
        if name.startswith("ar.L")
    ]

    if len(ar_terms) > 0:

        phi = params[ar_terms[0]]

    else:

        phi = 0.0

    beta_vat = params["VAT"] / vat_scale

    # For an AR(1) level model with |phi| < 1,
    # long-run multiplier = beta / (1 - phi).

    if abs(phi) < 1:

        long_run = beta_vat / (1 - phi)

    else:

        long_run = np.nan

    return {

        "Dependent Variable": dependent_name,

        "Short Run VAT Effect": beta_vat,

        "AR Coefficient": phi,

        "Long Run VAT Multiplier": long_run,

        "Stable AR Process": abs(phi) < 1

    }


dynamic_results = pd.DataFrame([

    dynamic_effects(
        savings_model,
        "Savings",
        exog_std["VAT"]
    ),

    dynamic_effects(
        investment_model,
        "Investment",
        exog_std["VAT"]
    )

])

print("\nSHORT RUN AND LONG RUN VAT EFFECTS")

display(dynamic_results.round(6))


# ================================================================
# STEP 17: RESIDUAL DIAGNOSTICS
# ================================================================

def residual_diagnostics(model, name):

    residuals = model.resid.dropna()

    lb_lag = min(
        5,
        max(1, len(residuals) // 4)
    )

    lb = acorr_ljungbox(

        residuals,

        lags=[lb_lag],

        return_df=True

    )

    jb_stat, jb_p, skew, kurt = jarque_bera(
        residuals
    )

    arch_stat, arch_p, _, _ = het_arch(
        residuals,
        nlags=lb_lag
    )

    result = {

        "Model": name,

        "Ljung Box Statistic": lb["lb_stat"].iloc[0],

        "Ljung Box P Value": lb["lb_pvalue"].iloc[0],

        "Jarque Bera Statistic": jb_stat,

        "Jarque Bera P Value": jb_p,

        "ARCH Statistic": arch_stat,

        "ARCH P Value": arch_p

    }

    # Residual time plot

    plt.figure(figsize=(12, 4))

    plt.plot(residuals)

    plt.axhline(
        0,
        color="red"
    )

    plt.title(f"Residual Plot: {name}")

    plt.show()

    # Residual ACF

    from statsmodels.graphics.tsaplots import plot_acf

    plot_acf(
        residuals,
        lags=min(8, len(residuals) - 1)
    )

    plt.title(f"Residual ACF: {name}")

    plt.show()

    return result


savings_diagnostics = residual_diagnostics(

    savings_model,

    "Savings"

)

investment_diagnostics = residual_diagnostics(

    investment_model,

    "Investment"

)

diagnostics_results = pd.DataFrame([

    savings_diagnostics,

    investment_diagnostics

])

print("\nMODEL DIAGNOSTICS")

display(diagnostics_results.round(5))


# ================================================================
# STEP 18: TRANSFER FUNCTION MODEL
# ================================================================

# Transfer function models allow VAT to affect savings and
# investment with distributed and delayed responses.
#
# A parsimonious transfer function is estimated by adding
# current VAT and one lagged VAT to the ARIMAX specification.
#
# IMPORTANT:
# This is a finite distributed-lag transfer function.
# It is not an unrestricted infinite-lag transfer function.


transfer_data = data.copy()

transfer_data["VAT_L1"] = transfer_data["VAT"].shift(1)

transfer_data = transfer_data.dropna()

transfer_exog = transfer_data[[
    "VAT",
    "VAT_L1",
    "Exchange_Rate",
    "Interest_Rate",
    "Inflation"
]]

# Standardize transfer function explanatory variables

transfer_mean = transfer_exog.mean()

transfer_std = transfer_exog.std(ddof=0)

transfer_std[transfer_std == 0] = 1

transfer_exog_scaled = (
    transfer_exog - transfer_mean
) / transfer_std


# ================================================================
# STEP 19: TRANSFER FUNCTION FOR SAVINGS
# ================================================================

transfer_savings = SARIMAX(

    transfer_data["Savings"],

    exog=transfer_exog_scaled,

    order=savings_model.model.order,

    trend="c",

    enforce_stationarity=False,

    enforce_invertibility=False

).fit(
    disp=False,
    maxiter=500
)

print("\nSAVINGS TRANSFER FUNCTION MODEL")

print(transfer_savings.summary())


# ================================================================
# STEP 20: TRANSFER FUNCTION FOR INVESTMENT
# ================================================================

transfer_investment = SARIMAX(

    transfer_data["Investment"],

    exog=transfer_exog_scaled,

    order=investment_model.model.order,

    trend="c",

    enforce_stationarity=False,

    enforce_invertibility=False

).fit(
    disp=False,
    maxiter=500
)

print("\nINVESTMENT TRANSFER FUNCTION MODEL")

print(transfer_investment.summary())


# ================================================================
# STEP 21: TRANSFER FUNCTION VAT EFFECTS
# ================================================================

def transfer_effect_table(model, scaling, name):

    result = pd.DataFrame({

        "Coefficient": model.params,

        "Standard_Error": model.bse,

        "P_Value": model.pvalues

    })

    # Original-scale VAT coefficients

    result.loc["VAT", "Coefficient"] /= scaling["VAT"]

    result.loc["VAT", "Standard_Error"] /= scaling["VAT"]

    result.loc["VAT_L1", "Coefficient"] /= scaling["VAT_L1"]

    result.loc["VAT_L1", "Standard_Error"] /= scaling["VAT_L1"]

    result["Model"] = name

    return result


transfer_savings_effects = transfer_effect_table(

    transfer_savings,

    transfer_std,

    "Savings Transfer Function"

)

transfer_investment_effects = transfer_effect_table(

    transfer_investment,

    transfer_std,

    "Investment Transfer Function"

)

print("\nTRANSFER FUNCTION EFFECTS: SAVINGS")

display(transfer_savings_effects)

print("\nTRANSFER FUNCTION EFFECTS: INVESTMENT")

display(transfer_investment_effects)


# ================================================================
# CORRECTED STEPS 22 TO 24
# FIX: ValueError: No supported index is available
# ================================================================

import numpy as np
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX

# ------------------------------------------------
# STEP 22: CREATE A VALID ANNUAL DATETIME INDEX
# ------------------------------------------------

# Reconstruct annual datetime index from original Year values
data.index = pd.DatetimeIndex(
    pd.to_datetime(data.index.astype(str) + "-12-31"),
    freq="YE"
)

data.index.name = "Year"

print("Annual datetime index:")
print(data.index)

print("Index frequency:", data.index.freq)

# ------------------------------------------------
# STEP 23: PREPARE FUTURE EXOGENOUS VARIABLES
# ------------------------------------------------

forecast_years = np.arange(
    data.index.year.max() + 1,
    data.index.year.max() + 7
)

forecast_dates = pd.date_range(
    start=f"{forecast_years[0]}-12-31",
    periods=6,
    freq="YE"
)

# Ensure future exogenous variables have a proper datetime index
future_exog_scaled = future_exog_scaled.copy()

future_exog_scaled.index = forecast_dates

future_exog_scaled.index.name = "Year"

# Ensure correct variable order
future_exog_scaled = future_exog_scaled[
    exog_variables
]

# Confirm that future explanatory variables are valid
assert future_exog_scaled.shape == (6, 4)

assert future_exog_scaled.notna().all().all()

print("\nFuture exogenous variables:")
display(future_exog_scaled.round(4))

# ------------------------------------------------
# STEP 24: REESTIMATE ARIMAX MODELS WITH VALID INDEX
# ------------------------------------------------

print("\nREESTIMATING SAVINGS ARIMAX MODEL")

# Reestimate using the previously selected ARIMA order
savings_model = SARIMAX(
    data["Savings"],
    exog=exog_scaled.set_axis(data.index),
    order=savings_model.model.order,
    trend="c",
    enforce_stationarity=False,
    enforce_invertibility=False
).fit(
    disp=False,
    maxiter=500
)

print("\nCorrected Savings Model:")
print(savings_model.summary())


print("\nREESTIMATING INVESTMENT ARIMAX MODEL")

investment_model = SARIMAX(
    data["Investment"],
    exog=exog_scaled.set_axis(data.index),
    order=investment_model.model.order,
    trend="c",
    enforce_stationarity=False,
    enforce_invertibility=False
).fit(
    disp=False,
    maxiter=500
)

print("\nCorrected Investment Model:")
print(investment_model.summary())

# ------------------------------------------------
# FORECAST SAVINGS
# ------------------------------------------------

print("\nFORECASTING SAVINGS: 2025 TO 2030")

savings_forecast = savings_model.get_forecast(
    steps=6,
    exog=future_exog_scaled
)

savings_mean = np.asarray(
    savings_forecast.predicted_mean
).reshape(-1)

savings_ci = np.asarray(
    savings_forecast.conf_int(alpha=0.05)
)

# ------------------------------------------------
# FORECAST INVESTMENT
# ------------------------------------------------

print("\nFORECASTING INVESTMENT: 2025 TO 2030")

investment_forecast = investment_model.get_forecast(
    steps=6,
    exog=future_exog_scaled
)

investment_mean = np.asarray(
    investment_forecast.predicted_mean
).reshape(-1)

investment_ci = np.asarray(
    investment_forecast.conf_int(alpha=0.05)
)

# ------------------------------------------------
# FINAL FORECAST RESULTS
# ------------------------------------------------

forecast_results = pd.DataFrame({

    "Year": forecast_years,

    "Savings_Forecast": savings_mean,

    "Savings_Lower_95": savings_ci[:, 0],

    "Savings_Upper_95": savings_ci[:, 1],

    "Investment_Forecast": investment_mean,

    "Investment_Lower_95": investment_ci[:, 0],

    "Investment_Upper_95": investment_ci[:, 1]

})

print("\nFINAL SAVINGS AND INVESTMENT FORECASTS")

display(forecast_results.round(4))

# ------------------------------------------------
# VALIDATION
# ------------------------------------------------

assert len(savings_mean) == 6

assert len(investment_mean) == 6

assert np.isfinite(savings_mean).all()

assert np.isfinite(investment_mean).all()

print("\nSUCCESS: ARIMAX forecasts generated successfully.")

print("Forecast period: 2025 to 2030")

# ================================================================
# CORRECTED STEP 25: FORECAST RESULTS TABLE
# FIX: AttributeError: 'numpy.ndarray' object has no attribute 'values'
# ================================================================

forecast_results = pd.DataFrame({

    "Year": forecast_years,

    "Savings_Forecast": np.asarray(savings_mean).reshape(-1),

    "Savings_Lower_95": np.asarray(savings_ci)[:, 0],

    "Savings_Upper_95": np.asarray(savings_ci)[:, 1],

    "Investment_Forecast": np.asarray(investment_mean).reshape(-1),

    "Investment_Lower_95": np.asarray(investment_ci)[:, 0],

    "Investment_Upper_95": np.asarray(investment_ci)[:, 1]

})

# Display final forecast results

print("\nSAVINGS AND INVESTMENT FORECASTS: 2025 TO 2030")

display(forecast_results.round(4))

# Check forecast dimensions

assert forecast_results.shape == (6, 7)

assert forecast_results.notna().all().all()

print("\nSUCCESS: Forecast results table generated successfully.")

# ================================================================
# CORRECTED STEP 26: FORECAST VISUALIZATION
# FIX: AttributeError: 'numpy.ndarray' object has no attribute 'values'
# ================================================================

# Convert forecast arrays into NumPy arrays safely
savings_mean = np.asarray(savings_mean).reshape(-1)

investment_mean = np.asarray(investment_mean).reshape(-1)

savings_lower = np.asarray(savings_ci)[:, 0]

savings_upper = np.asarray(savings_ci)[:, 1]

investment_lower = np.asarray(investment_ci)[:, 0]

investment_upper = np.asarray(investment_ci)[:, 1]


# ------------------------------------------------
# SAVINGS FORECAST PLOT
# ------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    data.index.year,
    data["Savings"].values,
    label="Observed Savings",
    marker="o"
)

plt.plot(
    forecast_years,
    savings_mean,
    label="Forecast Savings",
    linestyle="--",
    marker="o"
)

plt.fill_between(
    forecast_years,
    savings_lower,
    savings_upper,
    alpha=0.2
)

plt.title("Savings Forecast for Nigeria, 2025 to 2030")

plt.xlabel("Year")

plt.ylabel("Savings")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ------------------------------------------------
# INVESTMENT FORECAST PLOT
# ------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    data.index.year,
    data["Investment"].values,
    label="Observed Investment",
    marker="o"
)

plt.plot(
    forecast_years,
    investment_mean,
    label="Forecast Investment",
    linestyle="--",
    marker="o"
)

plt.fill_between(
    forecast_years,
    investment_lower,
    investment_upper,
    alpha=0.2
)

plt.title("Investment Forecast for Nigeria, 2025 to 2030")

plt.xlabel("Year")

plt.ylabel("Investment")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()

print("SUCCESS: Savings and investment forecast plots generated.")

# ================================================================
# STEP 27: OUT-OF-SAMPLE FORECAST ACCURACY
# ================================================================

# Use the final 5 observations as a validation sample.

def evaluate_forecast(actual, predicted):

    actual = np.asarray(actual)

    predicted = np.asarray(predicted)

    mae = mean_absolute_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    nonzero = actual != 0

    if np.any(nonzero):

        mape = np.mean(

            np.abs(

                (actual[nonzero] - predicted[nonzero])

                / actual[nonzero]

            )

        ) * 100

    else:

        mape = np.nan

    return {

        "MAE": mae,

        "RMSE": rmse,

        "MAPE": mape

    }


# Forecast validation

train = data.iloc[:-5]

test = data.iloc[-5:]

train_exog = (

    train[exog_variables] - exog_mean

) / exog_std


test_exog = (

    test[exog_variables] - exog_mean

) / exog_std


validation_savings_model = SARIMAX(

    train["Savings"],

    exog=train_exog,

    order=savings_model.model.order,

    trend="c",

    enforce_stationarity=False,

    enforce_invertibility=False

).fit(
    disp=False,
    maxiter=500
)


validation_investment_model = SARIMAX(

    train["Investment"],

    exog=train_exog,

    order=investment_model.model.order,

    trend="c",

    enforce_stationarity=False,

    enforce_invertibility=False

).fit(
    disp=False,
    maxiter=500
)


validation_savings_forecast = validation_savings_model.forecast(

    steps=5,

    exog=test_exog

)


validation_investment_forecast = validation_investment_model.forecast(

    steps=5,

    exog=test_exog

)


accuracy_results = pd.DataFrame({

    "Savings": evaluate_forecast(

        test["Savings"],

        validation_savings_forecast

    ),

    "Investment": evaluate_forecast(

        test["Investment"],

        validation_investment_forecast

    )

}).T


print("\nOUT-OF-SAMPLE FORECAST ACCURACY")

display(accuracy_results.round(4))


# ================================================================
# STEP 28: EXPORT ALL RESULTS
# ================================================================

output_file = "Transaction_Taxes_Box_Jenkins_Results.xlsx"

with pd.ExcelWriter(output_file) as writer:

    data.to_excel(

        writer,

        sheet_name="Cleaned_Data"

    )

    descriptive.to_excel(

        writer,

        sheet_name="Descriptive_Statistics"

    )

    correlation.to_excel(

        writer,

        sheet_name="Correlation"

    )

    stationarity_level.to_excel(

        writer,

        sheet_name="ADF_KPSS_Level",

        index=False

    )

    stationarity_diff.to_excel(

        writer,

        sheet_name="ADF_KPSS_Difference",

        index=False

    )

    savings_selection.to_excel(

        writer,

        sheet_name="Savings_Model_Selection",

        index=False

    )

    investment_selection.to_excel(

        writer,

        sheet_name="Investment_Model_Selection",

        index=False

    )

    vat_results.to_excel(

        writer,

        sheet_name="VAT_Effects",

        index=False

    )

    dynamic_results.to_excel(

        writer,

        sheet_name="Dynamic_Effects",

        index=False

    )

    diagnostics_results.to_excel(

        writer,

        sheet_name="Diagnostics",

        index=False

    )

    transfer_savings_effects.to_excel(

        writer,

        sheet_name="Savings_Transfer"

    )

    transfer_investment_effects.to_excel(

        writer,

        sheet_name="Investment_Transfer"

    )

    forecast_exog.to_excel(

        writer,

        sheet_name="Exogenous_Forecasts"

    )

    forecast_results.to_excel(

        writer,

        sheet_name="Final_Forecasts",

        index=False

    )

    accuracy_results.to_excel(

        writer,

        sheet_name="Forecast_Accuracy"

    )


# ================================================================
# STEP 29: DOWNLOAD FINAL RESULTS
# ================================================================

print("\nANALYSIS COMPLETED SUCCESSFULLY")

print("\nFinal Excel results file:", output_file)

files.download(output_file)


# ================================================================
# END OF ANALYTICAL PIPELINE
# ================================================================
```
