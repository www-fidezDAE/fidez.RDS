```python
# ================================================================
# MODELING NONLINEAR REGIME SHIFTS IN FOOD PRICE INFLATION IN
# UGANDA USING A SELF EXCITING THRESHOLD AUTOREGRESSIVE (SETAR)
# MODEL
#
# Study period: July 2017 to July 2026
# Frequency: Monthly
# Observations: 109
#
# Main dependent variable:
#     Food_Inflation
#
# Additional series:
#     Maize_Inflation
#     Bean_Inflation
#     Rice_Inflation
#
# Objectives:
# 1. Examine time series properties and volatility patterns.
# 2. Estimate SETAR models and identify inflation regimes.
# 3. Estimate statistically significant threshold values.
# 4. Compare SETAR and linear AR forecasting performance.
#
# ================================================================


# ================================================================
# 0. INSTALL REQUIRED PACKAGES IF NECESSARY
# ================================================================
#
# Run these commands once in Anaconda Prompt or terminal if needed:
#
# pip install pandas numpy scipy matplotlib seaborn openpyxl
# pip install statsmodels scikit-learn
#
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import skew, kurtosis

from sklearn.metrics import mean_squared_error, mean_absolute_error

import statsmodels.api as sm

from statsmodels.tsa.stattools import (
    adfuller,
    kpss,
    acf
)

from statsmodels.tsa.ar_model import AutoReg

from statsmodels.stats.diagnostic import (
    acorr_ljungbox,
    het_breuschpagan,
    het_arch,
    acorr_breusch_godfrey
)

from statsmodels.stats.stattools import jarque_bera

try:
    from statsmodels.tsa.stattools import bds
except ImportError:
    bds = None

from statsmodels.tsa.ar_model import ar_select_order


# ================================================================
# 2. USER SETTINGS
# ================================================================

# ----------------------------------------------------------------
# Excel file created from the UBOS data
# ----------------------------------------------------------------
DATA_FILE = "SETAR_Uganda_UBOS_CPI_2017_2026.xlsx"

# ----------------------------------------------------------------
# Sheet containing the cleaned data
# ----------------------------------------------------------------
DATA_SHEET = "SETAR_Data"

# ----------------------------------------------------------------
# Main series
# ----------------------------------------------------------------
MAIN_SERIES = "Food_Inflation"

# ----------------------------------------------------------------
# Other series to analyse
# ----------------------------------------------------------------
SERIES_TO_ANALYSE = [
    "Maize_Inflation",
    "Bean_Inflation",
    "Rice_Inflation",
    "Food_Inflation"
]

# ----------------------------------------------------------------
# Maximum autoregressive order
# ----------------------------------------------------------------
MAX_AR_LAG = 6

# ----------------------------------------------------------------
# Maximum delay parameter
# ----------------------------------------------------------------
MAX_DELAY = 6

# ----------------------------------------------------------------
# Minimum observations required in each regime
# ----------------------------------------------------------------
MIN_REGIME_OBS = 15

# ----------------------------------------------------------------
# Threshold search quantiles
# Avoid extreme thresholds that produce tiny regimes.
# ----------------------------------------------------------------
LOW_THRESHOLD_QUANTILE = 0.15
HIGH_THRESHOLD_QUANTILE = 0.85

# ----------------------------------------------------------------
# Significance level
# ----------------------------------------------------------------
ALPHA = 0.05

# ----------------------------------------------------------------
# Bootstrap replications for SETAR threshold significance
#
# 200 is suitable for initial analysis.
# For the final thesis, 999 or 1999 replications are preferable
# if computational resources permit.
# ----------------------------------------------------------------
BOOTSTRAP_REPS = 200

# ----------------------------------------------------------------
# Forecast split
#
# 80 percent training
# 20 percent testing
# ----------------------------------------------------------------
TRAIN_RATIO = 0.80

# ----------------------------------------------------------------
# Forecast horizon
# ----------------------------------------------------------------
FORECAST_HORIZON = None

# ----------------------------------------------------------------
# Random seed for reproducibility
# ----------------------------------------------------------------
RANDOM_SEED = 12345

np.random.seed(RANDOM_SEED)


# ================================================================
# 3. CREATE OUTPUT FOLDER
# ================================================================

OUTPUT_FOLDER = "SETAR_Uganda_Results"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

print("=" * 80)
print("SETAR ANALYSIS OF FOOD PRICE INFLATION IN UGANDA")
print("=" * 80)


# ================================================================
# 4. LOAD DATA
# ================================================================

if not os.path.exists(DATA_FILE):
    raise FileNotFoundError(
        f"\nThe file '{DATA_FILE}' was not found.\n"
        "Place the Excel file in the same folder as this Python script."
    )

df = pd.read_excel(
    DATA_FILE,
    sheet_name=DATA_SHEET
)

print("\nData successfully loaded.")
print("Number of observations:", len(df))
print("Number of variables:", len(df.columns))


# ================================================================
# 5. STANDARDISE COLUMN NAMES
# ================================================================

df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("%", "pct", regex=False)
)

print("\nAvailable variables:")
print(df.columns.tolist())


# ================================================================
# 6. CHECK REQUIRED VARIABLES
# ================================================================

required_columns = [
    "Date",
    "Year",
    "Month",
    "Maize_Inflation",
    "Bean_Inflation",
    "Rice_Inflation",
    "Food_Inflation"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    print("\nThe following expected variables were not found:")
    print(missing_columns)

    print("\nAvailable columns are:")
    print(df.columns.tolist())

    raise ValueError(
        "\nPlease check the column names in the Excel file."
    )


# ================================================================
# 7. CREATE DATE VARIABLE
# ================================================================

df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)

df = df.sort_values("Date").reset_index(drop=True)


# ================================================================
# 8. NUMERIC CONVERSION
# ================================================================

for variable in SERIES_TO_ANALYSE:

    df[variable] = pd.to_numeric(
        df[variable],
        errors="coerce"
    )


# ================================================================
# 9. DATA QUALITY CHECK
# ================================================================

print("\n" + "=" * 80)
print("DATA QUALITY CHECK")
print("=" * 80)

print("\nMissing observations:")

print(
    df[
        SERIES_TO_ANALYSE
    ].isnull().sum()
)

print("\nDuplicate dates:")

print(
    df["Date"].duplicated().sum()
)

print("\nDate range:")

print(
    df["Date"].min(),
    "to",
    df["Date"].max()
)


# ================================================================
# 10. CHECK FOR DUPLICATE OBSERVATIONS
# ================================================================

if df["Date"].duplicated().any():

    print(
        "\nWarning: duplicate dates detected."
    )

    df = (
        df
        .drop_duplicates(
            subset=["Date"],
            keep="first"
        )
        .reset_index(drop=True)
    )


# ================================================================
# 11. CHECK MONTHLY FREQUENCY
# ================================================================

date_difference = (
    df["Date"]
    .diff()
    .dropna()
)

print("\nDate interval summary:")

print(
    date_difference.value_counts().head()
)


# ================================================================
# 12. DESCRIPTIVE STATISTICS
# ================================================================

print("\n" + "=" * 80)
print("DESCRIPTIVE STATISTICS")
print("=" * 80)

descriptive = pd.DataFrame()

for variable in SERIES_TO_ANALYSE:

    series = (
        df[variable]
        .dropna()
        .astype(float)
    )

    descriptive.loc[
        variable,
        "N"
    ] = len(series)

    descriptive.loc[
        variable,
        "Mean"
    ] = series.mean()

    descriptive.loc[
        variable,
        "Median"
    ] = series.median()

    descriptive.loc[
        variable,
        "Minimum"
    ] = series.min()

    descriptive.loc[
        variable,
        "Maximum"
    ] = series.max()

    descriptive.loc[
        variable,
        "Std_Deviation"
    ] = series.std()

    descriptive.loc[
        variable,
        "Variance"
    ] = series.var()

    descriptive.loc[
        variable,
        "Skewness"
    ] = skew(series)

    descriptive.loc[
        variable,
        "Kurtosis"
    ] = kurtosis(
        series,
        fisher=False
    )

print(
    descriptive.round(4)
)


# ================================================================
# 13. SAVE DESCRIPTIVE STATISTICS
# ================================================================

descriptive.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "01_Descriptive_Statistics.xlsx"
    )
)


# ================================================================
# 14. PLOT FOOD INFLATION SERIES
# ================================================================

plt.figure(
    figsize=(14, 6)
)

plt.plot(
    df["Date"],
    df[MAIN_SERIES],
    linewidth=1.5
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title(
    "Monthly Food Price Inflation in Uganda"
)

plt.xlabel("Date")

plt.ylabel(
    "Monthly Food Inflation (%)"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "02_Food_Inflation_Time_Series.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 15. PLOT ALL INFLATION SERIES
# ================================================================

plt.figure(
    figsize=(14, 8)
)

for variable in SERIES_TO_ANALYSE:

    plt.plot(
        df["Date"],
        df[variable],
        label=variable
    )

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title(
    "Monthly Inflation Rates of Selected Food Commodities"
)

plt.xlabel("Date")

plt.ylabel(
    "Monthly Inflation (%)"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "03_All_Food_Inflation_Series.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 16. HISTOGRAMS
# ================================================================

for variable in SERIES_TO_ANALYSE:

    plt.figure(
        figsize=(9, 6)
    )

    sns.histplot(
        df[variable].dropna(),
        kde=True
    )

    plt.title(
        f"Distribution of {variable}"
    )

    plt.xlabel(
        "Monthly Inflation (%)"
    )

    plt.ylabel(
        "Frequency"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_FOLDER,
            f"04_Histogram_{variable}.png"
        ),
        dpi=300
    )

    plt.show()


# ================================================================
# 17. BOX PLOTS
# ================================================================

plt.figure(
    figsize=(11, 7)
)

sns.boxplot(
    data=df[
        SERIES_TO_ANALYSE
    ]
)

plt.title(
    "Box Plot of Monthly Food Inflation Rates"
)

plt.ylabel(
    "Inflation (%)"
)

plt.xticks(
    rotation=30
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "05_Boxplots_Food_Inflation.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 18. ROLLING VOLATILITY
# ================================================================

# Twelve month rolling standard deviation

rolling_volatility = pd.DataFrame()

rolling_volatility["Date"] = df["Date"]

for variable in SERIES_TO_ANALYSE:

    rolling_volatility[
        variable
    ] = (
        df[variable]
        .rolling(
            window=12
        )
        .std()
    )

plt.figure(
    figsize=(14, 7)
)

for variable in SERIES_TO_ANALYSE:

    plt.plot(
        rolling_volatility["Date"],
        rolling_volatility[variable],
        label=variable
    )

plt.title(
    "12 Month Rolling Volatility of Food Inflation"
)

plt.xlabel("Date")

plt.ylabel(
    "Rolling Standard Deviation"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "06_Rolling_Volatility.png"
    ),
    dpi=300
)

plt.show()


rolling_volatility.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "02_Rolling_Volatility.xlsx"
    ),
    index=False
)


# ================================================================
# 19. ADF TEST
# ================================================================

def adf_test(series):

    series = (
        pd.Series(series)
        .dropna()
        .astype(float)
    )

    result = adfuller(
        series,
        autolag="AIC"
    )

    return {
        "ADF_Statistic": result[0],
        "p_value": result[1],
        "Used_Lag": result[2],
        "Observations": result[3],
        "Critical_1pct": result[4]["1%"],
        "Critical_5pct": result[4]["5%"],
        "Critical_10pct": result[4]["10%"],
        "Stationary_5pct": (
            result[1] < ALPHA
        )
    }


adf_results = pd.DataFrame()

for variable in SERIES_TO_ANALYSE:

    result = adf_test(
        df[variable]
    )

    adf_results.loc[
        variable,
        "ADF_Statistic"
    ] = result["ADF_Statistic"]

    adf_results.loc[
        variable,
        "p_value"
    ] = result["p_value"]

    adf_results.loc[
        variable,
        "Used_Lag"
    ] = result["Used_Lag"]

    adf_results.loc[
        variable,
        "Observations"
    ] = result["Observations"]

    adf_results.loc[
        variable,
        "Critical_5pct"
    ] = result["Critical_5pct"]

    adf_results.loc[
        variable,
        "Stationary_5pct"
    ] = result["Stationary_5pct"]

print("\n" + "=" * 80)
print("AUGMENTED DICKEY FULLER TEST")
print("=" * 80)

print(
    adf_results.round(4)
)


# ================================================================
# 20. KPSS TEST
# ================================================================

def kpss_test(series):

    series = (
        pd.Series(series)
        .dropna()
        .astype(float)
    )

    result = kpss(
        series,
        regression="c",
        nlags="auto"
    )

    return {
        "KPSS_Statistic": result[0],
        "p_value": result[1],
        "Used_Lags": result[2],
        "Critical_5pct": result[3]["5%"],
        "Stationary_5pct": (
            result[1] > ALPHA
        )
    }


kpss_results = pd.DataFrame()

for variable in SERIES_TO_ANALYSE:

    try:

        result = kpss_test(
            df[variable]
        )

        kpss_results.loc[
            variable,
            "KPSS_Statistic"
        ] = result["KPSS_Statistic"]

        kpss_results.loc[
            variable,
            "p_value"
        ] = result["p_value"]

        kpss_results.loc[
            variable,
            "Used_Lags"
        ] = result["Used_Lags"]

        kpss_results.loc[
            variable,
            "Critical_5pct"
        ] = result["Critical_5pct"]

        kpss_results.loc[
            variable,
            "Stationary_5pct"
        ] = result["Stationary_5pct"]

    except Exception as e:

        print(
            f"KPSS failed for {variable}: {e}"
        )


print("\n" + "=" * 80)
print("KPSS TEST")
print("=" * 80)

print(
    kpss_results.round(4)
)


stationarity_results = adf_results.join(
    kpss_results,
    lsuffix="_ADF",
    rsuffix="_KPSS"
)

stationarity_results.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "03_Stationarity_Tests.xlsx"
    )
)


# ================================================================
# 21. BDS NONLINEARITY TEST
# ================================================================

print("\n" + "=" * 80)
print("BDS NONLINEARITY TEST")
print("=" * 80)

bds_results = pd.DataFrame()

if bds is not None:

    for variable in SERIES_TO_ANALYSE:

        series = (
            df[variable]
            .dropna()
            .astype(float)
            .values
        )

        try:

            # Standardise series before BDS
            series_standardised = (
                series - np.mean(series)
            ) / np.std(series)

            bds_stat, bds_pvalue = bds(
                series_standardised,
                max_dim=3
            )

            if np.ndim(bds_stat) == 0:

                bds_results.loc[
                    variable,
                    "BDS_Statistic"
                ] = float(bds_stat)

                bds_results.loc[
                    variable,
                    "p_value"
                ] = float(bds_pvalue)

            else:

                bds_results.loc[
                    variable,
                    "BDS_Statistic"
                ] = float(
                    np.asarray(bds_stat).flat[-1]
                )

                bds_results.loc[
                    variable,
                    "p_value"
                ] = float(
                    np.asarray(bds_pvalue).flat[-1]
                )

            bds_results.loc[
                variable,
                "Nonlinear_at_5pct"
            ] = (
                bds_results.loc[
                    variable,
                    "p_value"
                ] < ALPHA
            )

        except Exception as e:

            print(
                f"BDS failed for {variable}: {e}"
            )

else:

    print(
        "BDS function is unavailable in this version "
        "of Statsmodels."
    )


print(
    bds_results.round(4)
)

bds_results.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "04_BDS_Nonlinearity_Test.xlsx"
    )
)


# ================================================================
# 22. ACF OF INFLATION
# ================================================================

plt.figure(
    figsize=(12, 7)
)

from statsmodels.graphics.tsaplots import plot_acf

plot_acf(
    df[MAIN_SERIES].dropna(),
    lags=24
)

plt.title(
    "Autocorrelation Function of Food Inflation"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "07_ACF_Food_Inflation.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 23. ACF OF SQUARED INFLATION
# ================================================================

plt.figure(
    figsize=(12, 7)
)

plot_acf(
    df[MAIN_SERIES].dropna() ** 2,
    lags=24
)

plt.title(
    "ACF of Squared Food Inflation"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "08_ACF_Squared_Food_Inflation.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 24. AR MODEL SELECTION
# ================================================================

def select_linear_ar(series, max_lag=6):

    series = (
        pd.Series(series)
        .dropna()
        .astype(float)
    )

    candidate_results = []

    for p in range(1, max_lag + 1):

        try:

            model = AutoReg(
                series,
                lags=p,
                trend="c",
                old_names=False
            ).fit()

            candidate_results.append(
                {
                    "p": p,
                    "AIC": model.aic,
                    "BIC": model.bic,
                    "HQIC": model.hqic
                }
            )

        except Exception:
            continue

    result = pd.DataFrame(
        candidate_results
    )

    best_aic = int(
        result.loc[
            result["AIC"].idxmin(),
            "p"
        ]
    )

    best_bic = int(
        result.loc[
            result["BIC"].idxmin(),
            "p"
        ]
    )

    return result, best_aic, best_bic


ar_selection_all = {}

for variable in SERIES_TO_ANALYSE:

    results, best_aic, best_bic = (
        select_linear_ar(
            df[variable],
            MAX_AR_LAG
        )
    )

    ar_selection_all[
        variable
    ] = results

    results.to_excel(
        os.path.join(
            OUTPUT_FOLDER,
            f"05_AR_Lag_Selection_{variable}.xlsx"
        ),
        index=False
    )

    print("\n")
    print(variable)
    print(results.round(4))

    print(
        "Best AR lag by AIC:",
        best_aic
    )

    print(
        "Best AR lag by BIC:",
        best_bic
    )


# ================================================================
# 25. SETAR HELPER FUNCTIONS
# ================================================================

def create_lag_matrix(y, p):

    """
    Creates lagged variables for SETAR estimation.
    """

    y = np.asarray(y, dtype=float)

    rows = []

    for t in range(p, len(y)):

        lags = [
            y[t - j]
            for j in range(1, p + 1)
        ]

        rows.append(
            [y[t]] + lags
        )

    return np.asarray(rows)


def fit_ols(y, X):

    """
    Ordinary least squares estimation.
    """

    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float)

    X_constant = sm.add_constant(
        X,
        has_constant="add"
    )

    model = sm.OLS(
        y,
        X_constant
    ).fit()

    return model


def calculate_aic_bic(
    rss,
    n,
    k
):

    """
    Calculate AIC and BIC from residual sum of squares.
    """

    rss = max(
        float(rss),
        1e-12
    )

    sigma2 = rss / n

    loglik = (
        -0.5 * n *
        (
            np.log(
                2 * np.pi
            )
            +
            1
            +
            np.log(
                sigma2
            )
        )
    )

    aic = (
        -2 * loglik
        +
        2 * k
    )

    bic = (
        -2 * loglik
        +
        k * np.log(n)
    )

    return aic, bic


def threshold_candidates(
    y,
    delay,
    lower_q=0.15,
    upper_q=0.85
):

    """
    Generates candidate thresholds from the
    empirical distribution of the threshold variable.
    """

    y = np.asarray(y)

    threshold_variable = y[
        delay:
    ]

    low = np.quantile(
        threshold_variable,
        lower_q
    )

    high = np.quantile(
        threshold_variable,
        upper_q
    )

    candidates = np.unique(
        threshold_variable[
            (
                threshold_variable >= low
            )
            &
            (
                threshold_variable <= high
            )
        ]
    )

    return candidates


def fit_setar_candidate(
    y,
    p,
    delay,
    threshold,
    min_regime_obs=15
):

    """
    Estimate a two regime SETAR model.

    Regime 1:
        y(t-d) <= threshold

    Regime 2:
        y(t-d) > threshold
    """

    y = np.asarray(
        y,
        dtype=float
    )

    n = len(y)

    if n <= p + delay + 5:
        return None

    dependent = y[p:]

    X_lags = np.column_stack(
        [
            y[p-j:n-j]
            for j in range(1, p + 1)
        ]
    )

    threshold_variable = y[
        p - delay:
        n - delay
    ]

    regime_low = (
        threshold_variable
        <= threshold
    )

    regime_high = (
        threshold_variable
        > threshold
    )

    n_low = regime_low.sum()
    n_high = regime_high.sum()

    if (
        n_low < min_regime_obs
        or
        n_high < min_regime_obs
    ):
        return None

    X_low = X_lags[
        regime_low
    ]

    y_low = dependent[
        regime_low
    ]

    X_high = X_lags[
        regime_high
    ]

    y_high = dependent[
        regime_high
    ]

    model_low = fit_ols(
        y_low,
        X_low
    )

    model_high = fit_ols(
        y_high,
        X_high
    )

    fitted = np.empty(
        len(dependent)
    )

    fitted[
        regime_low
    ] = model_low.fittedvalues

    fitted[
        regime_high
    ] = model_high.fittedvalues

    residuals = (
        dependent - fitted
    )

    rss = np.sum(
        residuals ** 2
    )

    n_total = len(dependent)

    k = (
        2 * (p + 1)
    )

    aic, bic = calculate_aic_bic(
        rss,
        n_total,
        k
    )

    # Wald statistic for difference between
    # regime coefficient vectors.
    #
    # This is used as a descriptive threshold
    # effect statistic. The bootstrap below is
    # used for formal significance testing.

    beta_low = model_low.params
    beta_high = model_high.params

    covariance_low = model_low.cov_params()
    covariance_high = model_high.cov_params()

    beta_difference = (
        beta_low - beta_high
    )

    covariance_difference = (
        covariance_low
        +
        covariance_high
    )

    try:

        inv_cov = np.linalg.pinv(
            covariance_difference
        )

        wald = float(
            beta_difference.T
            @ inv_cov
            @ beta_difference
        )

        wald_df = len(
            beta_difference
        )

        wald_pvalue = 1 - stats.chi2.cdf(
            wald,
            wald_df
        )

    except Exception:

        wald = np.nan
        wald_pvalue = np.nan

    return {
        "p": p,
        "delay": delay,
        "threshold": threshold,
        "n_low": int(n_low),
        "n_high": int(n_high),
        "RSS": rss,
        "AIC": aic,
        "BIC": bic,
        "wald": wald,
        "wald_pvalue": wald_pvalue,
        "model_low": model_low,
        "model_high": model_high,
        "fitted": fitted,
        "residuals": residuals,
        "regime_low": regime_low,
        "regime_high": regime_high
    }


# ================================================================
# 26. SEARCH FOR OPTIMAL SETAR MODEL
# ================================================================

def search_setar(
    y,
    max_p=6,
    max_delay=6,
    min_regime_obs=15,
    lower_q=0.15,
    upper_q=0.85
):

    """
    Exhaustive grid search over:

        AR order p
        delay parameter d
        threshold gamma

    The selected model minimises BIC.
    """

    y = (
        pd.Series(y)
        .dropna()
        .astype(float)
        .values
    )

    candidates = []

    for p in range(
        1,
        max_p + 1
    ):

        for delay in range(
            1,
            max_delay + 1
        ):

            thresholds = threshold_candidates(
                y,
                delay,
                lower_q,
                upper_q
            )

            for threshold in thresholds:

                result = fit_setar_candidate(
                    y,
                    p,
                    delay,
                    threshold,
                    min_regime_obs
                )

                if result is not None:

                    candidates.append(
                        {
                            "p": result["p"],
                            "delay": result["delay"],
                            "threshold": result["threshold"],
                            "n_low": result["n_low"],
                            "n_high": result["n_high"],
                            "RSS": result["RSS"],
                            "AIC": result["AIC"],
                            "BIC": result["BIC"],
                            "Wald": result["wald"],
                            "Wald_pvalue": result["wald_pvalue"]
                        }
                    )

    results = pd.DataFrame(
        candidates
    )

    if results.empty:

        raise RuntimeError(
            "No valid SETAR models were found."
        )

    results = results.sort_values(
        "BIC"
    ).reset_index(
        drop=True
    )

    best_row = results.iloc[0]

    best_model = fit_setar_candidate(
        y,
        int(best_row["p"]),
        int(best_row["delay"]),
        float(best_row["threshold"]),
        min_regime_obs
    )

    return (
        results,
        best_model
    )


# ================================================================
# 27. ESTIMATE SETAR FOR MAIN FOOD INFLATION
# ================================================================

main_series = (
    df[MAIN_SERIES]
    .dropna()
    .astype(float)
    .values
)

print("\n" + "=" * 80)
print("SETAR MODEL SEARCH FOR MAIN FOOD INFLATION SERIES")
print("=" * 80)

setar_grid, main_setar = search_setar(
    main_series,
    max_p=MAX_AR_LAG,
    max_delay=MAX_DELAY,
    min_regime_obs=MIN_REGIME_OBS,
    lower_q=LOW_THRESHOLD_QUANTILE,
    upper_q=HIGH_THRESHOLD_QUANTILE
)

print(
    "\nTop 20 SETAR candidates:"
)

print(
    setar_grid.head(20).round(5)
)

print(
    "\nSelected SETAR model:"
)

print(
    "AR order p =",
    main_setar["p"]
)

print(
    "Delay d =",
    main_setar["delay"]
)

print(
    "Threshold gamma =",
    round(
        main_setar["threshold"],
        6
    )
)

print(
    "Low regime observations =",
    main_setar["n_low"]
)

print(
    "High regime observations =",
    main_setar["n_high"]
)

print(
    "AIC =",
    round(
        main_setar["AIC"],
        4
    )
)

print(
    "BIC =",
    round(
        main_setar["BIC"],
        4
    )
)

print(
    "Wald statistic =",
    round(
        main_setar["wald"],
        4
    )
)

print(
    "Wald p value =",
    round(
        main_setar["wald_pvalue"],
        6
    )
)


setar_grid.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "06_SETAR_Grid_Search.xlsx"
    ),
    index=False
)


# ================================================================
# 28. SETAR REGIME INTERPRETATION
# ================================================================

print("\n" + "=" * 80)
print("SETAR REGIME INTERPRETATION")
print("=" * 80)

gamma = main_setar[
    "threshold"
]

d = main_setar[
    "delay"
]

print(
    f"\nEstimated threshold: {gamma:.6f}%"
)

print(
    f"Delay parameter: {d} month(s)"
)

print(
    "\nRegime 1:"
)

print(
    f"Threshold variable <= {gamma:.6f}%"
)

print(
    "\nRegime 2:"
)

print(
    f"Threshold variable > {gamma:.6f}%"
)

print(
    "\nImportant:"
)

print(
    "The threshold is a statistically estimated regime boundary."
)

print(
    "It should not automatically be described as a government"
)

print(
    "inflation policy threshold unless supported by additional"
)

print(
    "economic interpretation and robustness evidence."
)


# ================================================================
# 29. REGIME SPECIFIC MODEL SUMMARIES
# ================================================================

print("\n" + "=" * 80)
print("LOW REGIME MODEL")
print("=" * 80)

print(
    main_setar[
        "model_low"
    ].summary()
)


print("\n" + "=" * 80)
print("HIGH REGIME MODEL")
print("=" * 80)

print(
    main_setar[
        "model_high"
    ].summary()
)


# ================================================================
# 30 SETAR REGIME COEFFICIENT TABLE
# ================================================================

p = main_setar["p"]

variable_names = (
    ["Constant"]
    +
    [f"Lag_{i}" for i in range(1, p + 1)]
)

coef_table = pd.DataFrame({
    "Variable": variable_names,
    "Low_Regime_Coefficient": np.asarray(
        low_model.params
    ),
    "Low_Regime_pvalue": np.asarray(
        low_model.pvalues
    ),
    "High_Regime_Coefficient": np.asarray(
        high_model.params
    ),
    "High_Regime_pvalue": np.asarray(
        high_model.pvalues
    )
})

print("\n" + "=" * 80)
print("SETAR REGIME SPECIFIC COEFFICIENTS")
print("=" * 80)

print(
    coef_table.round(6)
)

coef_table.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "07_SETAR_Regime_Coefficients.xlsx"
    ),
    index=False
)

# ================================================================
# 31. CREATE REGIME CLASSIFICATION
# ================================================================

analysis_series = (
    df[
        [
            "Date",
            MAIN_SERIES
        ]
    ]
    .dropna()
    .copy()
)

analysis_series[
    "Threshold_Lag"
] = (
    analysis_series[
        MAIN_SERIES
    ]
    .shift(d)
)

analysis_series[
    "Regime"
] = np.where(
    analysis_series[
        "Threshold_Lag"
    ] <= gamma,
    "Low Inflation Regime",
    "High Inflation Regime"
)

analysis_series[
    "Threshold_Lag"
].iloc[:d] = np.nan

analysis_series.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "08_SETAR_Regime_Classification.xlsx"
    ),
    index=False
)


# ================================================================
# 32. REGIME FREQUENCY
# ================================================================

regime_frequency = (
    analysis_series[
        "Regime"
    ]
    .value_counts()
    .to_frame(
        "Observations"
    )
)

regime_frequency[
    "Percentage"
] = (
    regime_frequency[
        "Observations"
    ]
    /
    regime_frequency[
        "Observations"
    ].sum()
    *
    100
)

print("\nRegime distribution:")

print(
    regime_frequency.round(2)
)

regime_frequency.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "09_Regime_Frequency.xlsx"
    )
)


# ================================================================
# 33. REGIME PLOT
# ================================================================

plot_data = analysis_series.copy()

plt.figure(
    figsize=(15, 7)
)

for regime_name, group in plot_data.groupby(
    "Regime",
    dropna=True
):

    plt.scatter(
        group["Date"],
        group[MAIN_SERIES],
        label=regime_name,
        s=25
    )

plt.axhline(
    y=gamma,
    linestyle="--",
    label="Estimated Threshold"
)

plt.title(
    "Estimated Inflation Regimes from SETAR Model"
)

plt.xlabel("Date")

plt.ylabel(
    "Food Inflation (%)"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "09_SETAR_Regimes.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 34. SETAR RESIDUALS
# ================================================================

setar_residuals = pd.Series(
    main_setar[
        "residuals"
    ],
    name="SETAR_Residuals"
)

print("\nResidual observations:")

print(
    len(setar_residuals)
)


# ================================================================
# 35. LJUNG BOX TEST
# ================================================================

lb_lags = [
    lag
    for lag in [6, 12, 18, 24]
    if lag < len(setar_residuals)
]

ljung_box = acorr_ljungbox(
    setar_residuals,
    lags=lb_lags,
    return_df=True
)

print("\n" + "=" * 80)
print("LJUNG BOX TEST FOR SETAR RESIDUALS")
print("=" * 80)

print(
    ljung_box
)

ljung_box.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "10_SETAR_Ljung_Box_Test.xlsx"
    )
)


# ================================================================
# 36. JARQUE BERA NORMALITY TEST
# ================================================================

jb_stat, jb_pvalue, jb_skew, jb_kurtosis = (
    jarque_bera(
        setar_residuals
    )
)

jb_result = pd.DataFrame({
    "Statistic": [jb_stat],
    "p_value": [jb_pvalue],
    "Skewness": [jb_skew],
    "Kurtosis": [jb_kurtosis]
})

print("\n" + "=" * 80)
print("JARQUE BERA TEST")
print("=" * 80)

print(
    jb_result.round(5)
)

jb_result.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "11_SETAR_Jarque_Bera_Test.xlsx"
    ),
    index=False
)


# ================================================================
# 37. BREUSCH PAGAN HETEROSCEDASTICITY TEST
# ================================================================

y_fit = (
    main_setar[
        "fitted"
    ]
)

X_bp = sm.add_constant(
    np.asarray(
        y_fit
    )
)

try:

    bp_lm, bp_lm_pvalue, bp_f, bp_f_pvalue = (
        het_breuschpagan(
            setar_residuals,
            X_bp
        )
    )

    bp_result = pd.DataFrame({
        "LM_Statistic": [bp_lm],
        "LM_pvalue": [bp_lm_pvalue],
        "F_Statistic": [bp_f],
        "F_pvalue": [bp_f_pvalue]
    })

    print("\n" + "=" * 80)
    print("BREUSCH PAGAN HETEROSCEDASTICITY TEST")
    print("=" * 80)

    print(
        bp_result.round(5)
    )

    bp_result.to_excel(
        os.path.join(
            OUTPUT_FOLDER,
            "12_SETAR_Breusch_Pagan_Test.xlsx"
        ),
        index=False
    )

except Exception as e:

    print(
        "Breusch Pagan test failed:",
        e
    )


# ================================================================
# 38. ARCH LM TEST
# ================================================================

try:

    arch_result = het_arch(
        setar_residuals,
        nlags=12
    )

    arch_table = pd.DataFrame({
        "LM_Statistic": [
            arch_result[0]
        ],
        "LM_pvalue": [
            arch_result[1]
        ],
        "F_Statistic": [
            arch_result[2]
        ],
        "F_pvalue": [
            arch_result[3]
        ]
    })

    print("\n" + "=" * 80)
    print("ARCH LM TEST")
    print("=" * 80)

    print(
        arch_table.round(5)
    )

    arch_table.to_excel(
        os.path.join(
            OUTPUT_FOLDER,
            "13_SETAR_ARCH_LM_Test.xlsx"
        ),
        index=False
    )

except Exception as e:

    print(
        "ARCH LM test failed:",
        e
    )


# ================================================================
# 39. RESIDUAL PLOT
# ================================================================

plt.figure(
    figsize=(14, 6)
)

plt.plot(
    setar_residuals
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title(
    "SETAR Model Residuals"
)

plt.xlabel("Observation")

plt.ylabel("Residual")

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "10_SETAR_Residuals.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 40. BOOTSTRAP SUP W ALD THRESHOLD TEST
# ================================================================

def fit_linear_ar(
    y,
    p
):

    model = AutoReg(
        y,
        lags=p,
        trend="c",
        old_names=False
    ).fit()

    return model


def calculate_sup_wald(
    y,
    max_p=6,
    max_delay=6,
    min_regime_obs=15,
    lower_q=0.15,
    upper_q=0.85
):

    """
    Searches for the maximum regime difference
    Wald statistic over candidate SETAR models.

    This provides the basis for a bootstrap test of
    threshold nonlinearity.
    """

    results, best_model = search_setar(
        y,
        max_p=max_p,
        max_delay=max_delay,
        min_regime_obs=min_regime_obs,
        lower_q=lower_q,
        upper_q=upper_q
    )

    valid_wald = results[
        "Wald"
    ].replace(
        [np.inf, -np.inf],
        np.nan
    ).dropna()

    if len(valid_wald) == 0:
        return np.nan

    return float(
        valid_wald.max()
    )


# ---------------------------------------------------------------
# Observed sup Wald statistic
# ---------------------------------------------------------------

observed_sup_wald = calculate_sup_wald(
    main_series,
    max_p=MAX_AR_LAG,
    max_delay=MAX_DELAY,
    min_regime_obs=MIN_REGIME_OBS,
    lower_q=LOW_THRESHOLD_QUANTILE,
    upper_q=HIGH_THRESHOLD_QUANTILE
)

print("\n" + "=" * 80)
print("BOOTSTRAP THRESHOLD SIGNIFICANCE TEST")
print("=" * 80)

print(
    "Observed sup Wald statistic:",
    round(
        observed_sup_wald,
        5
    )
)


# ================================================================
# 41. GENERATE BOOTSTRAP SERIES UNDER LINEAR AR NULL
# ================================================================

# Use the BIC selected linear AR model

ar_selection_main, best_ar_aic, best_ar_bic = (
    select_linear_ar(
        main_series,
        MAX_AR_LAG
    )
)

linear_ar_order = best_ar_bic

linear_model_full = AutoReg(
    main_series,
    lags=linear_ar_order,
    trend="c",
    old_names=False
).fit()

linear_residuals = np.asarray(
    linear_model_full.resid,
    dtype=float
)

linear_fitted = np.asarray(
    linear_model_full.fittedvalues,
    dtype=float
)


def simulate_ar_bootstrap(
    fitted_model,
    residuals,
    n,
    burn=100
):

    """
    Simulate a bootstrap sample under the
    linear AR null hypothesis using resampled
    residuals.
    """

    params = fitted_model.params

    p = fitted_model.model.ar_lags[-1]

    simulated = np.zeros(
        n + burn
    )

    # Start with observed mean
    simulated[:p] = np.mean(
        fitted_model.model.endog
    )

    for t in range(
        p,
        n + burn
    ):

        intercept = params[0]

        ar_parameters = params[1:]

        lag_values = simulated[
            t-p:t
        ][::-1]

        innovation = np.random.choice(
            residuals
        )

        simulated[t] = (
            intercept
            +
            np.dot(
                ar_parameters,
                lag_values
            )
            +
            innovation
        )

    return simulated[
        burn:
    ]


# ================================================================
# 42. BOOTSTRAP LOOP
# ================================================================

bootstrap_statistics = []

print(
    f"\nRunning {BOOTSTRAP_REPS} bootstrap replications..."
)

for b in range(
    BOOTSTRAP_REPS
):

    try:

        bootstrap_series = simulate_ar_bootstrap(
            linear_model_full,
            linear_residuals,
            len(main_series)
        )

        bootstrap_sup_wald = calculate_sup_wald(
            bootstrap_series,
            max_p=MAX_AR_LAG,
            max_delay=MAX_DELAY,
            min_regime_obs=MIN_REGIME_OBS,
            lower_q=LOW_THRESHOLD_QUANTILE,
            upper_q=HIGH_THRESHOLD_QUANTILE
        )

        bootstrap_statistics.append(
            bootstrap_sup_wald
        )

    except Exception:

        continue

    if (
        (b + 1) % 25 == 0
        or
        b == 0
    ):

        print(
            f"Completed {b + 1} / "
            f"{BOOTSTRAP_REPS}"
        )


bootstrap_statistics = np.asarray(
    bootstrap_statistics
)

bootstrap_statistics = (
    bootstrap_statistics[
        np.isfinite(
            bootstrap_statistics
        )
    ]
)


# ================================================================
# 43. BOOTSTRAP P VALUE
# ================================================================

if len(
    bootstrap_statistics
) > 0:

    bootstrap_pvalue = (
        np.sum(
            bootstrap_statistics
            >= observed_sup_wald
        )
        +
        1
    ) / (
        len(
            bootstrap_statistics
        )
        +
        1
    )

    bootstrap_critical_95 = np.quantile(
        bootstrap_statistics,
        0.95
    )

else:

    bootstrap_pvalue = np.nan
    bootstrap_critical_95 = np.nan


threshold_test_result = pd.DataFrame({
    "Observed_Sup_Wald": [
        observed_sup_wald
    ],
    "Bootstrap_95pct_Critical_Value": [
        bootstrap_critical_95
    ],
    "Bootstrap_p_value": [
        bootstrap_pvalue
    ],
    "Significant_at_5pct": [
        (
            bootstrap_pvalue
            < ALPHA
        )
        if not np.isnan(
            bootstrap_pvalue
        )
        else np.nan
    ],
    "Bootstrap_Replications": [
        len(
            bootstrap_statistics
        )
    ]
})

print("\n" + "=" * 80)
print("BOOTSTRAP SUP WALD RESULTS")
print("=" * 80)

print(
    threshold_test_result.round(5)
)

threshold_test_result.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "14_SETAR_Bootstrap_Threshold_Test.xlsx"
    ),
    index=False
)


# ================================================================
# 44. PLOT BOOTSTRAP DISTRIBUTION
# ================================================================

if len(
    bootstrap_statistics
) > 0:

    plt.figure(
        figsize=(11, 6)
    )

    plt.hist(
        bootstrap_statistics,
        bins=30
    )

    plt.axvline(
        observed_sup_wald,
        linestyle="--",
        linewidth=2,
        label="Observed sup Wald"
    )

    plt.axvline(
        bootstrap_critical_95,
        linestyle=":",
        linewidth=2,
        label="95% Bootstrap Critical Value"
    )

    plt.title(
        "Bootstrap Distribution of the Sup Wald Statistic"
    )

    plt.xlabel(
        "Sup Wald Statistic"
    )

    plt.ylabel(
        "Frequency"
    )

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_FOLDER,
            "11_Bootstrap_Sup_Wald.png"
        ),
        dpi=300
    )

    plt.show()


# ================================================================
# 45. FORECASTING FUNCTION FOR SETAR
# ================================================================

def recursive_setar_forecast(
    history,
    horizon,
    p,
    delay,
    threshold,
    model_low,
    model_high
):

    """
    Recursive multi step SETAR forecasting.

    The appropriate regime is selected using the
    estimated threshold and delay parameter.
    """

    history = list(
        np.asarray(
            history,
            dtype=float
        )
    )

    forecasts = []

    for step in range(
        horizon
    ):

        lag_values = np.array(
            [
                history[-j]
                for j in range(
                    1,
                    p + 1
                )
            ]
        )

        threshold_value = (
            history[-delay]
        )

        X = sm.add_constant(
            lag_values.reshape(
                1,
                -1
            ),
            has_constant="add"
        )

        if (
            threshold_value
            <= threshold
        ):

            forecast = float(
                model_low.predict(
                    X
                )[0]
            )

            regime = (
                "Low Inflation Regime"
            )

        else:

            forecast = float(
                model_high.predict(
                    X
                )[0]
            )

            regime = (
                "High Inflation Regime"
            )

        history.append(
            forecast
        )

        forecasts.append(
            {
                "Forecast": forecast,
                "Regime": regime
            }
        )

    return forecasts


# ================================================================
# 46. TRAIN TEST SPLIT
# ================================================================

forecast_series = (
    df[
        [
            "Date",
            MAIN_SERIES
        ]
    ]
    .dropna()
    .reset_index(
        drop=True
    )
)

y = (
    forecast_series[
        MAIN_SERIES
    ]
    .values
)

dates = (
    forecast_series[
        "Date"
    ]
)

n = len(y)

train_size = int(
    np.floor(
        n * TRAIN_RATIO
    )
)

test_size = (
    n - train_size
)

train = y[
    :train_size
]

test = y[
    train_size:
]

train_dates = dates[
    :train_size
]

test_dates = dates[
    train_size:
]

print("\n" + "=" * 80)
print("TRAIN TEST SPLIT")
print("=" * 80)

print(
    "Total observations:",
    n
)

print(
    "Training observations:",
    train_size
)

print(
    "Testing observations:",
    test_size
)

print(
    "Training period:",
    train_dates.iloc[0],
    "to",
    train_dates.iloc[-1]
)

print(
    "Testing period:",
    test_dates.iloc[0],
    "to",
    test_dates.iloc[-1]
)


# ================================================================
# 47. SELECT LINEAR AR MODEL USING TRAINING DATA
# ================================================================

ar_train_selection, ar_aic_p, ar_bic_p = (
    select_linear_ar(
        train,
        MAX_AR_LAG
    )
)

selected_ar_p = ar_bic_p

print("\n" + "=" * 80)
print("SELECTED LINEAR AR MODEL")
print("=" * 80)

print(
    "Selected AR order using BIC:",
    selected_ar_p
)


linear_ar_model = AutoReg(
    train,
    lags=selected_ar_p,
    trend="c",
    old_names=False
).fit()


# ================================================================
# 48. SEARCH SETAR USING TRAINING DATA
# ================================================================

print("\n" + "=" * 80)
print("SELECTING SETAR MODEL USING TRAINING DATA")
print("=" * 80)

train_setar_grid, train_setar = search_setar(
    train,
    max_p=MAX_AR_LAG,
    max_delay=MAX_DELAY,
    min_regime_obs=MIN_REGIME_OBS,
    lower_q=LOW_THRESHOLD_QUANTILE,
    upper_q=HIGH_THRESHOLD_QUANTILE
)

print(
    "\nSelected training SETAR:"
)

print(
    "p =",
    train_setar["p"]
)

print(
    "d =",
    train_setar["delay"]
)

print(
    "threshold =",
    train_setar["threshold"]
)

print(
    "BIC =",
    train_setar["BIC"]
)


# ================================================================
# 49. LINEAR AR FORECAST
# ================================================================

ar_forecast = (
    linear_ar_model
    .predict(
        start=train_size,
        end=n - 1,
        dynamic=False
    )
)

ar_forecast = np.asarray(
    ar_forecast
)


# ================================================================
# 50. SETAR FORECAST
# ================================================================

setar_forecast_details = recursive_setar_forecast(
    history=train,
    horizon=test_size,
    p=train_setar["p"],
    delay=train_setar["delay"],
    threshold=train_setar["threshold"],
    model_low=train_setar["model_low"],
    model_high=train_setar["model_high"]
)

setar_forecast = np.array(
    [
        item["Forecast"]
        for item in setar_forecast_details
    ]
)

setar_forecast_regimes = [
    item["Regime"]
    for item in setar_forecast_details
]


# ================================================================
# 51. FORECAST ACCURACY FUNCTIONS
# ================================================================

def calculate_mape(
    actual,
    predicted
):

    actual = np.asarray(
        actual,
        dtype=float
    )

    predicted = np.asarray(
        predicted,
        dtype=float
    )

    denominator = np.where(
        np.abs(actual) < 1e-8,
        np.nan,
        np.abs(actual)
    )

    return np.nanmean(
        np.abs(
            (
                actual - predicted
            )
            /
            denominator
        )
    ) * 100


def calculate_rmse(
    actual,
    predicted
):

    return np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )


def calculate_mae(
    actual,
    predicted
):

    return mean_absolute_error(
        actual,
        predicted
    )


def calculate_mse(
    actual,
    predicted
):

    return mean_squared_error(
        actual,
        predicted
    )


# ================================================================
# 52. FORECAST ACCURACY
# ================================================================

forecast_comparison = pd.DataFrame({

    "Model": [
        "Linear AR",
        "SETAR"
    ],

    "RMSE": [
        calculate_rmse(
            test,
            ar_forecast
        ),
        calculate_rmse(
            test,
            setar_forecast
        )
    ],

    "MAE": [
        calculate_mae(
            test,
            ar_forecast
        ),
        calculate_mae(
            test,
            setar_forecast
        )
    ],

    "MSE": [
        calculate_mse(
            test,
            ar_forecast
        ),
        calculate_mse(
            test,
            setar_forecast
        )
    ],

    "MAPE": [
        calculate_mape(
            test,
            ar_forecast
        ),
        calculate_mape(
            test,
            setar_forecast
        )
    ]
})


print("\n" + "=" * 80)
print("FORECAST ACCURACY COMPARISON")
print("=" * 80)

print(
    forecast_comparison.round(5)
)


forecast_comparison.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "15_AR_vs_SETAR_Forecast_Accuracy.xlsx"
    ),
    index=False
)


# ================================================================
# 53. FORECAST DATASET
# ================================================================

forecast_results = pd.DataFrame({

    "Date": test_dates.values,

    "Actual_Food_Inflation": test,

    "AR_Forecast": ar_forecast,

    "SETAR_Forecast": setar_forecast,

    "SETAR_Regime": setar_forecast_regimes,

    "AR_Absolute_Error": np.abs(
        test - ar_forecast
    ),

    "SETAR_Absolute_Error": np.abs(
        test - setar_forecast
    )
})

forecast_results.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "16_Forecast_Results.xlsx"
    ),
    index=False
)


# ================================================================
# 54. FORECAST GRAPH
# ================================================================

plt.figure(
    figsize=(15, 7)
)

plt.plot(
    test_dates,
    test,
    marker="o",
    label="Actual"
)

plt.plot(
    test_dates,
    ar_forecast,
    linestyle="--",
    label="Linear AR Forecast"
)

plt.plot(
    test_dates,
    setar_forecast,
    linestyle=":",
    linewidth=2,
    label="SETAR Forecast"
)

plt.title(
    "Out of Sample Forecast Comparison: AR versus SETAR"
)

plt.xlabel("Date")

plt.ylabel(
    "Food Inflation (%)"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "12_AR_vs_SETAR_Forecasts.png"
    ),
    dpi=300
)

plt.show()


# ================================================================
# 55. DETERMINE BEST FORECASTING MODEL
# ================================================================

rmse_ar = forecast_comparison.loc[
    forecast_comparison["Model"]
    == "Linear AR",
    "RMSE"
].iloc[0]

rmse_setar = forecast_comparison.loc[
    forecast_comparison["Model"]
    == "SETAR",
    "RMSE"
].iloc[0]

mape_ar = forecast_comparison.loc[
    forecast_comparison["Model"]
    == "Linear AR",
    "MAPE"
].iloc[0]

mape_setar = forecast_comparison.loc[
    forecast_comparison["Model"]
    == "SETAR",
    "MAPE"
].iloc[0]

if (
    rmse_setar < rmse_ar
    and
    mape_setar < mape_ar
):

    best_forecasting_model = "SETAR"

elif (
    rmse_ar < rmse_setar
    and
    mape_ar < mape_setar
):

    best_forecasting_model = "Linear AR"

else:

    best_forecasting_model = (
        "Mixed result; compare RMSE and MAPE jointly"
    )


print("\n" + "=" * 80)
print("FORECASTING CONCLUSION")
print("=" * 80)

print(
    "Best forecasting model:",
    best_forecasting_model
)


# ================================================================
# 56. DIEBOLD MARIANO TEST
# ================================================================

def diebold_mariano_test(
    actual,
    forecast1,
    forecast2,
    power=2
):

    """
    Simple Diebold Mariano test using loss differential.

    forecast1 = AR
    forecast2 = SETAR

    Null:
        Equal predictive accuracy.

    Alternative:
        Different predictive accuracy.
    """

    actual = np.asarray(
        actual,
        dtype=float
    )

    forecast1 = np.asarray(
        forecast1,
        dtype=float
    )

    forecast2 = np.asarray(
        forecast2,
        dtype=float
    )

    loss1 = np.abs(
        actual - forecast1
    ) ** power

    loss2 = np.abs(
        actual - forecast2
    ) ** power

    differential = (
        loss1 - loss2
    )

    mean_d = np.mean(
        differential
    )

    variance_d = np.var(
        differential,
        ddof=1
    )

    n_obs = len(
        differential
    )

    dm_stat = (
        mean_d
        /
        np.sqrt(
            variance_d
            /
            n_obs
        )
    )

    p_value = (
        2
        *
        (
            1
            -
            stats.norm.cdf(
                np.abs(dm_stat)
            )
        )
    )

    return dm_stat, p_value


dm_stat, dm_pvalue = (
    diebold_mariano_test(
        test,
        ar_forecast,
        setar_forecast
    )
)

dm_result = pd.DataFrame({
    "DM_Statistic": [
        dm_stat
    ],
    "p_value": [
        dm_pvalue
    ],
    "Significant_at_5pct": [
        dm_pvalue < ALPHA
    ]
})

print("\n" + "=" * 80)
print("DIEBOLD MARIANO TEST")
print("=" * 80)

print(
    dm_result.round(5)
)

dm_result.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "17_Diebold_Mariano_Test.xlsx"
    ),
    index=False
)


# ================================================================
# 57. ANALYSE ALL FOUR SERIES WITH SETAR
# ================================================================

print("\n" + "=" * 80)
print("SETAR ANALYSIS FOR ALL FOOD INFLATION SERIES")
print("=" * 80)

all_setar_summary = []

for variable in SERIES_TO_ANALYSE:

    print(
        f"\nAnalysing {variable}..."
    )

    series = (
        df[variable]
        .dropna()
        .astype(float)
        .values
    )

    try:

        grid, model = search_setar(
            series,
            max_p=MAX_AR_LAG,
            max_delay=MAX_DELAY,
            min_regime_obs=MIN_REGIME_OBS,
            lower_q=LOW_THRESHOLD_QUANTILE,
            upper_q=HIGH_THRESHOLD_QUANTILE
        )

        all_setar_summary.append({

            "Series": variable,

            "Selected_p": model["p"],

            "Selected_Delay": model["delay"],

            "Threshold": model["threshold"],

            "Low_Regime_N": model["n_low"],

            "High_Regime_N": model["n_high"],

            "RSS": model["RSS"],

            "AIC": model["AIC"],

            "BIC": model["BIC"],

            "Wald_Statistic": model["wald"],

            "Wald_pvalue": model["wald_pvalue"]

        })

    except Exception as e:

        print(
            f"SETAR failed for {variable}:",
            e
        )


all_setar_summary = pd.DataFrame(
    all_setar_summary
)

print(
    all_setar_summary.round(5)
)

all_setar_summary.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "18_SETAR_All_Series_Summary.xlsx"
    ),
    index=False
)


# ================================================================
# 58. GENERATE FINAL MASTER RESULTS TABLE
# ================================================================

master_results = pd.DataFrame({

    "Study_Component": [
        "Main Series",
        "Study Period",
        "Number of Observations",
        "Selected SETAR AR Order",
        "Selected SETAR Delay",
        "Estimated Threshold",
        "Low Regime Observations",
        "High Regime Observations",
        "SETAR AIC",
        "SETAR BIC",
        "Threshold Wald Statistic",
        "Threshold Wald p value",
        "Bootstrap Sup Wald p value",
        "Threshold Significant at 5%",
        "Linear AR Order",
        "Linear AR RMSE",
        "SETAR RMSE",
        "Linear AR MAPE",
        "SETAR MAPE",
        "Diebold Mariano p value",
        "Best Forecasting Model"
    ],

    "Result": [

        MAIN_SERIES,

        (
            str(
                df["Date"].min().date()
            )
            +
            " to "
            +
            str(
                df["Date"].max().date()
            )
        ),

        len(main_series),

        main_setar["p"],

        main_setar["delay"],

        main_setar["threshold"],

        main_setar["n_low"],

        main_setar["n_high"],

        main_setar["AIC"],

        main_setar["BIC"],

        main_setar["wald"],

        main_setar["wald_pvalue"],

        bootstrap_pvalue,

        (
            bootstrap_pvalue < ALPHA
            if not np.isnan(
                bootstrap_pvalue
            )
            else np.nan
        ),

        selected_ar_p,

        rmse_ar,

        rmse_setar,

        mape_ar,

        mape_setar,

        dm_pvalue,

        best_forecasting_model
    ]
})

master_results.to_excel(
    os.path.join(
        OUTPUT_FOLDER,
        "19_MASTER_SETAR_RESULTS.xlsx"
    ),
    index=False
)


# ================================================================
# 59. SAVE ALL RESULTS INTO ONE EXCEL WORKBOOK
# ================================================================

MASTER_EXCEL = os.path.join(
    OUTPUT_FOLDER,
    "SETAR_Uganda_COMPLETE_RESULTS.xlsx"
)

with pd.ExcelWriter(
    MASTER_EXCEL,
    engine="openpyxl"
) as writer:

    descriptive.to_excel(
        writer,
        sheet_name="Descriptive",
        index=True
    )

    stationarity_results.to_excel(
        writer,
        sheet_name="Stationarity",
        index=True
    )

    bds_results.to_excel(
        writer,
        sheet_name="BDS",
        index=True
    )

    setar_grid.head(100).to_excel(
        writer,
        sheet_name="SETAR_Grid",
        index=False
    )

    coef_table.to_excel(
        writer,
        sheet_name="SETAR_Coefficients",
        index=False
    )

    regime_frequency.to_excel(
        writer,
        sheet_name="Regime_Frequency",
        index=True
    )

    threshold_test_result.to_excel(
        writer,
        sheet_name="Threshold_Test",
        index=False
    )

    forecast_comparison.to_excel(
        writer,
        sheet_name="Forecast_Comparison",
        index=False
    )

    forecast_results.to_excel(
        writer,
        sheet_name="Forecasts",
        index=False
    )

    dm_result.to_excel(
        writer,
        sheet_name="DM_Test",
        index=False
    )

    all_setar_summary.to_excel(
        writer,
        sheet_name="All_SETAR",
        index=False
    )

    master_results.to_excel(
        writer,
        sheet_name="Master_Results",
        index=False
    )


# ================================================================
# 60. FINAL CONSOLE REPORT
# ================================================================

print("\n")
print("=" * 80)
print("FINAL SETAR ANALYSIS REPORT")
print("=" * 80)

print(
    "\nStudy:",
    "Modeling Nonlinear Regime Shifts in Food Price Inflation in Uganda"
)

print(
    "\nMain series:",
    MAIN_SERIES
)

print(
    "\nObservations:",
    len(main_series)
)

print(
    "\nPeriod:",
    df["Date"].min().strftime("%Y-%m"),
    "to",
    df["Date"].max().strftime("%Y-%m")
)

print(
    "\nSelected SETAR order:",
    f"SETAR({main_setar['p']},{main_setar['p']})"
)

print(
    "Delay parameter:",
    main_setar["delay"]
)

print(
    "Estimated threshold:",
    round(
        main_setar["threshold"],
        5
    )
)

print(
    "Low inflation regime:",
    main_setar["n_low"],
    "observations"
)

print(
    "High inflation regime:",
    main_setar["n_high"],
    "observations"
)

print(
    "\nBootstrap threshold p value:",
    round(
        bootstrap_pvalue,
        5
    )
)

if (
    not np.isnan(
        bootstrap_pvalue
    )
):

    if (
        bootstrap_pvalue
        < ALPHA
    ):

        print(
            "Decision: Reject the null hypothesis "
            "of no threshold nonlinearity at 5%."
        )

    else:

        print(
            "Decision: Fail to reject the null hypothesis "
            "of no threshold nonlinearity at 5%."
        )

print(
    "\nLinear AR RMSE:",
    round(
        rmse_ar,
        5
    )
)

print(
    "SETAR RMSE:",
    round(
        rmse_setar,
        5
    )
)

print(
    "\nLinear AR MAPE:",
    round(
        mape_ar,
        5
    )
)

print(
    "SETAR MAPE:",
    round(
        mape_setar,
        5
    )
)

print(
    "\nDiebold Mariano p value:",
    round(
        dm_pvalue,
        5
    )
)

print(
    "\nPreferred forecasting model:",
    best_forecasting_model
)

print(
    "\nAll results have been saved in:"
)

print(
    os.path.abspath(
        OUTPUT_FOLDER
    )
)

print(
    "\nMaster Excel workbook:"
)

print(
    os.path.abspath(
        MASTER_EXCEL
    )
)

print(
    "\nAnalysis completed successfully."
)

print("=" * 80)
```
